````python
#@title 🟢 REQUIRED · Cell 0 — START HERE (run once during the stretch; ~30 s) { display-mode: "form" }
FORCE_CACHED = False  #@param {type:"boolean"}
# Room-wide STUDIO MODE: tick the box above (or the instructor projects "FORCE_CACHED = True").
REPO_RAW = "https://raw.githubusercontent.com/thezeming/mgrcm0065-week3/main"   # class-repo base URL, baked in from w3_core.REPO_RAW_URL; empty = no fetch
_RECONNECT = False
# pinned SDK (guarded: warm runtimes skip the install)
try:
    from google import genai as _probe  # noqa
except ImportError:
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "google-genai==1.47.0"])
EMBEDDED_B64 = ""
# ---- shared core (verbatim copy of build/w3_core.py) ------------------------
# ============================================================================
#  MGRCM0065 Week 3 · "AI for Finance" — shared core
#  This file is (a) embedded verbatim into cell 0 of the student notebook and
#  (b) imported by build/build_cache.py, build/make_sample_cache.py and the
#  tests, so the notebook and the cache builder run IDENTICAL code.
#  Only the standard library + pandas + numpy + google-genai are imported.
# ============================================================================
import base64, gzip, hashlib, io, json, os, re, sys, textwrap, time, math
from datetime import datetime

import numpy as np
import pandas as pd

# ----------------------------------------------------------------------------
# Constants (brief §1). MODEL is a placeholder the instructor re-checks on
# build day and on 6 Oct at aistudio.google.com/rate-limit.
# ----------------------------------------------------------------------------
MODEL = "gemini-3.5-flash-lite"              # the configured choice; may 404 if Google withdraws it
# Chosen on 17 Sep 2026 from the instructor's own account catalogue (41 models). It is the
# newest non-preview "flash-lite" the key can see, which is the cheapest/fastest tier and so
# the one with the largest free daily allowance — what matters with ~100 students. The
# previous value, gemini-2.5-flash-lite, returned 404 from an older SDK on 17 Sep 2026.
# pick_model() independently selects this same model from that catalogue.
MODEL_FALLBACK = True                        # on a 404, ask the key what it CAN use and switch (loudly)
MODEL_IN_USE = MODEL                         # the model actually talking to us; build_cache records THIS
SDK_VERSION = "2.24.0"                       # google-genai version this code was written against
SDK_MIN_VERSION = "1.20.0"                   # oldest google-genai whose API surface we verified by hand
MODEL_EXCLUDE = ("pro", "ultra")             # never auto-pick an expensive tier
GROQ_MODEL = "openai/gpt-oss-20b"
TICKERS = ["AAPL", "MSFT", "NVDA", "TSLA", "AMZN", "META", "JPM", "BA"]
COMMITTEE_TICKER = "NVDA"
FLAT_BAND = 0.005                            # ±0.5 % is FLAT
PACE_SECONDS = 7                             # client-side spacing between live requests
TOOL_CAP = 3                                 # READ-ONLY tool calls per agent turn
ORDER_CAP = 2                                # place_order calls per turn (refused + retry smaller)
# The two budgets are deliberately separate. A single shared cap looks tidier but starves the
# mandate: the model researches all three tickers in ONE parallel batch, which spends three
# calls, and place_order then comes back "CAP REACHED" with an empty order book — observed
# against the live API on 17 Sep 2026. Lab 3 exists to show an order being filled and then
# refused, so the order must never be the thing the budget runs out on. ORDER_CAP is 2 because
# the 3.3 demonstration needs a refusal followed by a smaller retry, which is two calls.
MANDATE_GBP = 1_000_000                      # paper mandate, GBP notional only
# Raw-GitHub base URL of the class repo, e.g.
#   "https://raw.githubusercontent.com/<user>/<repo>/main"
# make_notebook.py bakes this into cell 0. Empty = the notebook never phones home and always uses
# its own embedded copy. Set it once here, then `python build/refresh.py --publish`.
REPO_RAW_URL = "https://raw.githubusercontent.com/thezeming/mgrcm0065-week3/main"
TRACE_WIDTH = 72
HALLUCINATION_DATE = "2023-06-30"            # inside any plausible training window; no splits since
HALLUCINATION_TICKERS = ["MSFT", "META", "JPM"]
COMPANY = {"AAPL": "Apple", "MSFT": "Microsoft", "NVDA": "Nvidia", "TSLA": "Tesla",
           "AMZN": "Amazon", "META": "Meta", "JPM": "JPMorgan", "BA": "Boeing"}

MODE_LIVE, MODE_STUDIO, MODE_ERROR = "LIVE MODE", "STUDIO MODE", "ERROR — hand up"
STUDIO_SUBLINE = ("every number, order and refusal is computed on your laptop; "
                  "the model's wording is the instructor's pre-run")
CACHED_BANNER = "[cached output — instructor's pre-run]"
SAMPLE_BANNER = "SAMPLE CACHE — regenerate with build/build_cache.py before class"

# Anonymised twins of the Lab-1 headlines (names, tickers, products and dates
# removed; written by the notebook builder for the 1.5b design, not by a model).
ANON_HEADLINES = {
    "H01": "Company X warns of flagship-product shipment delays after a Covid lockdown at a supplier's plant",
    "H02": "Company X announces multibillion-dollar investment in an AI start-up",
    "H03": "Company X guidance beats estimates on generative-AI demand",
    "H04": "Company X to cut over 18,000 jobs",
    "H05": "Company X posts first-ever drop in daily active users, issues weak guidance",
    "H06": "Company X acquires a failed regional bank after regulators seize it",
    "H07": "Regulator orders inspections, grounds Company X's aircraft type after an engine failure",
    "H08": "Company X confirmed to join a major stock index next month",
}

# ----------------------------------------------------------------------------
# Global state. Everything the cells share lives in S so that the 🔁 cells can
# rebuild it and the build scripts can inspect it.
# ----------------------------------------------------------------------------
_prev_S = globals().get("S") if isinstance(globals().get("S"), dict) else {}
S = {
    "mode": None, "provider": None, "client": None, "key_status": "unchecked",
    "n_requests": 0, "last_request": 0.0, "cache": {}, "cache_meta": {},
    "prices": None, "headlines": None, "fresh": None, "committee_md": None,
    "version": {}, "data_source": "none", "results": {}, "recording": None,
    "in_colab": False, "sdk_ok": False, "grounding_seen": False,
    "model": MODEL, "model_switched": False, "sdk_version": None,
}
# Re-running cell 0 or a 🔁 cell must NOT wipe the lock, the results or the order book.
S["results"] = dict(_prev_S.get("results", {}))
S["n_requests"] = int(_prev_S.get("n_requests", 0))
ORDER_BOOK = list(globals().get("ORDER_BOOK") or [])
GUARD = dict(globals().get("GUARD") or {"MAX_ORDER_GBP": MANDATE_GBP, "REQUIRE_HUMAN_APPROVAL": False})
PREDICTIONS_LOCKED = globals().get("PREDICTIONS_LOCKED", None)   # None or (timestamp_str, {id: call}, kind)
_BOOTSTRAPPED = bool(globals().get("_BOOTSTRAPPED", False))      # True once cell 0 / a 🔁 cell has run in THIS kernel
#   kind ∈ {"locked", "restored"}: "restored" is the placeholder a 🔁 cell sets when the runtime lost the
#   real lock — it opens Labs 2–4 but never scores, and 1.2 may overwrite it.
TODAY = None                   # last trading date in prices.csv, set by load_data()

# Values the notebook's cell 0 may define before this code runs.
FORCE_CACHED = bool(globals().get("FORCE_CACHED", False)) or os.environ.get("W3_FORCE_CACHED", "") == "1"
REPO_RAW = str(globals().get("REPO_RAW", "") or "")
EMBEDDED_B64 = globals().get("EMBEDDED_B64", None)
_RECONNECT = bool(globals().get("_RECONNECT", False))
QUIET = bool(globals().get("W3_QUIET", False))

genai = None
gtypes = None
gerrors = None


def _p(*a, **k):
    """print() that the build scripts can silence."""
    if not QUIET:
        print(*a, **k)


def _hr(title=""):
    line = "─" * TRACE_WIDTH
    _p(line if not title else f"── {title} " + "─" * max(0, TRACE_WIDTH - len(title) - 4))


def _countdown(seconds, label):
    """Every wait prints a countdown so nothing looks like a hang."""
    seconds = int(math.ceil(seconds))
    if seconds <= 0:
        return
    _p(f"{label}: ", end="", flush=True)
    for i in range(seconds, 0, -1):
        _p(f"{i}…", end=" ", flush=True)
        time.sleep(1)
    _p("go", flush=True)


def _trace(line):
    """Print a trace line wrapped so that no printed line exceeds 72 chars."""
    for chunk in textwrap.wrap(line, TRACE_WIDTH, subsequent_indent="      ", break_on_hyphens=False) or [""]:
        _p(chunk)


def _say(text, indent=""):
    """Print prose wrapped at 72 characters (plain lines, no hanging indent)."""
    for chunk in textwrap.wrap(text, TRACE_WIDTH, initial_indent=indent, subsequent_indent=indent,
                               break_on_hyphens=False) or [""]:
        _p(chunk)


def _sha(text):
    return hashlib.sha256(text.encode("utf-8")).hexdigest()


def _now():
    return datetime.now().strftime("%H:%M:%S")


# ----------------------------------------------------------------------------
# Data loading: embedded blob first, disk second (build scripts), GitHub third
# ----------------------------------------------------------------------------
def _materials_dir():
    here = os.path.dirname(os.path.abspath(globals().get("__file__", "./build/w3_core.py")))
    return os.path.abspath(os.path.join(here, ".."))


def _decode_blob(b64):
    return json.loads(gzip.decompress(base64.b64decode(b64)).decode("utf-8"))


def _read_disk_blob():
    root = _materials_dir()
    blob = {}
    for key, rel in [("prices_csv", "data/prices.csv"), ("headlines_csv", "data/headlines.csv"),
                     ("headlines_fresh_csv", "data/headlines_fresh.csv"),
                     ("responses_json", "cache/responses.json"),
                     ("committee_md", "cache/committee.md"), ("version_json", "VERSION.json")]:
        path = os.path.join(root, rel)
        blob[key] = open(path, encoding="utf-8").read() if os.path.exists(path) else None
    return blob


def _install_blob(blob, source):
    global TODAY
    if not blob.get("prices_csv") or not blob.get("headlines_csv"):
        raise ValueError("prices.csv / headlines.csv missing from the data bundle")
    prices = pd.read_csv(io.StringIO(blob["prices_csv"]))
    prices["date"] = pd.to_datetime(prices["date"]).dt.strftime("%Y-%m-%d")
    prices = prices.sort_values(["ticker", "date"]).reset_index(drop=True)
    heads = pd.read_csv(io.StringIO(blob["headlines_csv"]))
    fresh = pd.read_csv(io.StringIO(blob["headlines_fresh_csv"])) if blob.get("headlines_fresh_csv") else None
    for df in (heads, fresh):
        if df is not None:
            for c in ("date", "event_date"):
                df[c] = pd.to_datetime(df[c]).dt.strftime("%Y-%m-%d")
    cache = json.loads(blob["responses_json"]) if blob.get("responses_json") else {"_meta": {"sample": True, "empty": True}, "entries": {}}
    S.update(prices=prices, headlines=heads, fresh=fresh, data_source=source,
             cache=cache.get("entries", {}), cache_meta=cache.get("_meta", {}),
             committee_md=blob.get("committee_md"),
             version=json.loads(blob["version_json"]) if blob.get("version_json") else {})
    TODAY = str(prices["date"].max())
    globals()["TODAY"] = TODAY


def load_data():
    """Embedded copy first (always sufficient for Labs 1–4), disk copy for the build scripts."""
    if EMBEDDED_B64:
        _install_blob(_decode_blob(EMBEDDED_B64), "embedded copy inside this notebook")
    else:
        _install_blob(_read_disk_blob(), "materials/ folder on disk")


def _http_get(url, timeout=10):
    import urllib.request
    with urllib.request.urlopen(url, timeout=timeout) as r:
        return r.read().decode("utf-8")


REPO_FILES = [("prices_csv", "data/prices.csv"),
              ("headlines_csv", "data/headlines.csv"),
              ("headlines_fresh_csv", "data/headlines_fresh.csv"),
              ("responses_json", "cache/responses.json"),
              ("committee_md", "cache/committee.md")]


def bundle_last_trading_day(blob):
    prices = pd.read_csv(io.StringIO(blob["prices_csv"]))
    return str(pd.to_datetime(prices["date"]).dt.strftime("%Y-%m-%d").max())


def bundle_is_coherent(blob):
    """Do the prices and the recorded answers in this bundle belong to each other?

    A cache entry is keyed on cell_id + sha256(prompt), and the Lab 3 / Lab 4 prompts embed
    TODAY, which is the last trading day in prices.csv. So a bundle whose answers were recorded
    against a DIFFERENT day orphans every Lab 3 and Lab 4 entry: the cells still run, but a
    student without a working key gets "no cached answer" in exactly the two labs that carry the
    teaching — and only once their quota runs out, so it never shows up in testing.

    Checked before anything is installed. Returns (ok, reason).
    """
    try:
        last = bundle_last_trading_day(blob)
    except Exception as e:
        return False, f"prices.csv could not be read ({type(e).__name__})"
    try:
        meta = (json.loads(blob["responses_json"]) or {}).get("_meta", {})
    except Exception as e:
        return False, f"responses.json could not be read ({type(e).__name__})"
    recorded = str(meta.get("today_constant", "") or "")
    if not recorded:
        return False, "the recorded answers do not say which trading day they were built against"
    if recorded != last:
        return False, (f"prices run to {last} but the answers were recorded for {recorded} — "
                       "Lab 3 and Lab 4 would fall through to 'no cached answer'")
    return True, ""


def _warn_if_notebook_stale(ver, mine):
    if ver.get("notebook") != mine.get("notebook"):
        _say(f"⚠️  VERSION: this notebook is v{mine.get('notebook')}; current is v{ver.get('notebook')} — "
             "open the fresh notebook link and save a new copy (the code itself cannot self-update).")


def fetch_repo_updates():
    """Refresh data/ and cache/ from the class repo (skipped when REPO_RAW is empty).

    ALL OR NOTHING, deliberately. prices.csv and cache/responses.json are a matched pair — see
    bundle_is_coherent() — so a half-applied refresh is worse than no refresh at all. One failed
    download, or one mismatched pair, abandons the entire attempt and leaves the embedded copy
    untouched. The embedded copy is always internally consistent, because make_notebook.py built
    it from one snapshot.
    """
    if not REPO_RAW:
        _p("GitHub refresh: skipped (REPO_RAW is empty) — using the embedded copy.")
        return
    if os.environ.get("W3_NO_FETCH") == "1":
        # make_notebook.py captures the Expected-output boxes by EXECUTING each cell. Without this
        # the build would phone home four times per run, and its output would depend on the network
        # — which breaks the byte-identical-rebuild test.
        _p("GitHub refresh: (runs here — OK / already current / not reachable)")
        return
    base = REPO_RAW.rstrip("/")
    bust = f"?v={int(time.time())}"
    mine = dict(S["version"])
    try:
        ver = json.loads(_http_get(f"{base}/VERSION.json{bust}"))
    except Exception as e:
        _say(f"GitHub refresh: not reachable ({type(e).__name__}) — embedded copy is fine for today.")
        return
    if ver.get("data") == mine.get("data"):
        _say(f"GitHub refresh: already current (data v{mine.get('data')}) — nothing to download.")
        _warn_if_notebook_stale(ver, mine)
        return
    fetched = {}
    for key, rel in REPO_FILES:
        try:
            fetched[key] = _http_get(f"{base}/{rel}{bust}")
        except Exception as e:
            _say(f"GitHub refresh: {rel} did not download ({type(e).__name__}) — abandoning the whole "
                 "refresh rather than mixing new prices with older answers; embedded copy stands.")
            _warn_if_notebook_stale(ver, mine)
            return
    blob = dict(_read_disk_blob()) if not EMBEDDED_B64 else _decode_blob(EMBEDDED_B64)
    blob.update(fetched)
    blob["version_json"] = json.dumps(ver)
    ok, why = bundle_is_coherent(blob)
    if not ok:
        _say(f"GitHub refresh: REFUSED — {why}. Embedded copy stands.")
        _warn_if_notebook_stale(ver, mine)
        return
    try:
        _install_blob(blob, f"GitHub ({base})")
    except Exception as e:
        _say(f"GitHub refresh: bundle rejected ({type(e).__name__}) — embedded copy stands.")
        _warn_if_notebook_stale(ver, mine)
        return
    _say(f"GitHub refresh: OK — data v{mine.get('data')} \u2192 v{ver.get('data')}; prices to {TODAY}, "
         "answers recorded for the same day.")
    _warn_if_notebook_stale(ver, mine)


# ----------------------------------------------------------------------------
# SDK install / import (pinned; guarded so warm runtimes skip it)
# ----------------------------------------------------------------------------
def ensure_sdk():
    global genai, gtypes, gerrors
    try:
        from google import genai as _g
    except ImportError:
        _p(f"Installing google-genai=={SDK_VERSION} (about 20 seconds)…")
        os.system(f"{sys.executable} -m pip install -q google-genai=={SDK_VERSION}")
        try:
            from google import genai as _g
        except ImportError:
            _say("Runtime → Restart session, then run this cell again — normal, 20 seconds.")
            S["sdk_ok"] = False
            return False
    from google.genai import types as _t, errors as _e
    genai, gtypes, gerrors = _g, _t, _e
    ver = getattr(_g, "__version__", "?")
    S["sdk_version"] = ver
    if not sdk_version_ok(ver):
        S["sdk_ok"] = False
        _p(f"SDK: google-genai {ver} — TOO OLD")
        _say(f"This notebook needs google-genai {SDK_MIN_VERSION} or newer (it was written "
             f"against {SDK_VERSION}); you have {ver}, whose API is different enough to fail "
             "in the middle of a cell rather than here. Fix it with one line:")
        _p(f"  {sys.executable} -m pip install -U 'google-genai>={SDK_MIN_VERSION}'")
        _say("If pip says no matching distribution, the cause is your Python, not your "
             "network: modern google-genai needs Python 3.10 or newer, and 'python3' on a Mac "
             "is often the system 3.9. Check with  python3 -V  and rebuild the environment "
             "with an explicit modern interpreter, e.g. /opt/homebrew/bin/python3.12.")
        _say("STUDIO MODE meanwhile — every number, order and refusal still runs.")
        return False
    S["sdk_ok"] = True
    _p(f"SDK: google-genai {ver}")
    if ver != SDK_VERSION:
        _say(f"(written against {SDK_VERSION}; {SDK_MIN_VERSION} and newer are supported — "
             "the calls this notebook makes are identical on both.)")
    return True


def _ver_tuple(v):
    """('1.47.0' → (1, 47, 0)); anything unparseable sorts as (0,) so it fails the floor check."""
    parts = []
    for chunk in str(v).split(".")[:3]:
        digits = "".join(c for c in chunk if c.isdigit())
        if digits == "":
            break
        parts.append(int(digits))
    return tuple(parts) or (0,)


def sdk_version_ok(ver):
    """True when the installed google-genai is at least SDK_MIN_VERSION."""
    return _ver_tuple(ver) >= _ver_tuple(SDK_MIN_VERSION)


# ----------------------------------------------------------------------------
# Key discovery and the three-attempt verification
# ----------------------------------------------------------------------------
def find_key():
    """Returns (key, provider, status). Distinguishes no key / NotebookAccessError."""
    try:
        from google.colab import userdata  # noqa
        S["in_colab"] = True
    except ImportError:
        S["in_colab"] = False
        userdata = None
    for name, provider in (("GEMINI_API_KEY", "gemini"), ("GROQ_API_KEY", "groq")):
        if userdata is not None:
            try:
                val = userdata.get(name)
                if val:
                    return val.strip(), provider, "colab-secret"
            except Exception as e:  # SecretNotFoundError / NotebookAccessError
                if "NotebookAccess" in type(e).__name__:
                    return None, provider, "notebook-access"
        val = os.environ.get(name, "").strip()
        if val:
            return val, provider, "env"
    return None, None, "none"


# Remedies, in the order _classify_error tests for them. Each one has its own branch and its own
# sentence on screen, because telling someone to check their Wi-Fi when the model name is wrong
# wastes the whole room's time.
#   rate-limit      429 / RESOURCE_EXHAUSTED — the key works, the free quota is busy
#   model-not-found 404 / NOT_FOUND, or a 403 whose message names models/… — the MODEL name is
#                   not available to this key (deterministic: never retried, because waiting
#                   cannot make a withdrawn model come back)
#   bad-key         400/401/403 naming the key or a permission — the key itself is wrong
#   server          5xx at the provider
#   network         a transport-level failure: DNS, TCP, TLS, timeout
#   other           anything we have not met before; never described as a network problem
def _classify_error(e):
    """Map any exception from the SDK to exactly one remedy. Specific codes first; the
    name-sniffing network test last, so 'network' only ever means an actual network fault.

    Works for both SDKs: google-genai puts an int in .code, Groq/OpenAI-style SDKs put a
    string error code there and the HTTP status in .status_code, so we normalise first."""
    name = type(e).__name__
    code = getattr(e, "code", None)
    if not isinstance(code, int) or isinstance(code, bool):
        try:
            code = int(code)
        except (TypeError, ValueError):
            code = None
    if code is None:
        sc = getattr(e, "status_code", None)
        code = sc if isinstance(sc, int) and not isinstance(sc, bool) else None
    status = str(getattr(e, "status", "") or "")
    msg = str(e)
    up = msg.upper()
    if code == 429 or "RESOURCE_EXHAUSTED" in status or "429" in msg[:12] or "RateLimit" in name:
        return "rate-limit"
    if code == 404 or "NOT_FOUND" in status or "NOT_FOUND" in up[:200] or "NotFound" in name:
        return "model-not-found"
    # A 403 PERMISSION_DENIED that names a model ("Permission denied on resource
    # models/gemini-2.5-flash-lite") is the model being withdrawn from this key, not a bad
    # key — Google returns either code for it. Tested before the key branch on purpose.
    if code in (403, 404) and "MODELS/" in up and "API_KEY" not in up:
        return "model-not-found"
    if code in (400, 401, 403) and ("API_KEY" in up or "PERMISSION" in up or code in (401, 403)):
        return "bad-key"
    if code is None and any(w in name for w in ("Authentication", "PermissionDenied")):
        return "bad-key"
    if code is not None and 500 <= int(code) < 600:
        return "server"
    if any(w in name for w in ("Timeout", "Connect", "Network", "Remote", "Transport", "OSError", "URLError")):
        return "network"
    return "other"


def _key_console(provider):
    """Where to re-copy a rejected key from — the provider's own console, never Google's
    for a Groq key."""
    return "console.groq.com/keys" if provider == "groq" else "aistudio.google.com/apikey"


def model_from_error(e, default=None):
    """Pull the model name out of a 404 message ('models/gemini-2.5-flash-lite is not found')."""
    m = re.search(r"models/([A-Za-z0-9._\-]+)", str(e))
    # ".rstrip('.')" because a 403 states it as a sentence: "…on resource models/x-y-z."
    return m.group(1).rstrip(".") if m else (default if default is not None else MODEL_IN_USE)


# ----------------------------------------------------------------------------
# Model discovery. Google publishes no static free-tier table (design-notes/
# research_tooling.md), so a 404 on the configured MODEL is an expected failure,
# not an exotic one. list_text_models() asks the key what it can actually use;
# pick_model() chooses between them by a rule that is fixed and written down.
# ----------------------------------------------------------------------------
def list_text_models(client=None):
    """Every model this API key can call with generateContent, as plain names
    ('gemini-2.5-flash'). A listing entry that advertises no supported actions at all is
    skipped rather than kept. Raises the SDK's own exception so the caller can classify it."""
    client = client or S["client"]
    out = []
    for m in client.models.list(config={"query_base": True}):
        actions = list(getattr(m, "supported_actions", None) or [])
        if not any(a.lower() == "generatecontent" for a in actions):
            continue   # a model that advertises nothing is treated as unusable, not as a candidate
        name = (getattr(m, "name", "") or "").split("/")[-1]
        if name:
            out.append(name)
    return sorted(set(out))


def _version_tuple(n):
    """The version that belongs to the family name, as a fixed-width tuple (largest first).

    Only the dotted or whole numbers immediately after the first word count, and we stop at
    the first token that is not a number, so a date suffix ('-preview-11-2025') or a build
    number ('-001') is never read as a version. Fixed width so the comparison is positional:
    without the padding, ('gemini-3',) would sort ahead of ('gemini-3.5',) simply because the
    shorter tuple is a prefix of the longer one."""
    nums = []
    for tok in n.split("-")[1:]:
        if re.fullmatch(r"\d+(\.\d+)*", tok):
            nums.extend(int(x) for x in tok.split("."))
        else:
            break
    return tuple(-x for x in (nums + [0, 0, 0])[:3])


def _model_rank(name):
    """Sort key for pick_model. Smaller is better, and it is entirely determined by the
    name, so the same list always yields the same choice:
      1. family      flash-lite (0) before flash (1) before anything else (2);
      2. version     the numbers in the name, largest first, compared position by position
                     so 3.5 beats a bare 3 (3.6 beats 2.5);
      3. maturity    a plain name before a preview/experimental/dated one;
      4. simplicity  the shorter name, then alphabetical — a pure tie-break."""
    n = name.lower()
    family = 0 if "flash-lite" in n else (1 if "flash" in n else 2)
    nums = _version_tuple(n)
    preview = 1 if any(w in n for w in ("preview", "exp", "-rc", "latest")) else 0
    return (family, nums, preview, len(name), name)


def pick_model(names):
    """The best model to fall back to, or None. Never picks a pro/ultra tier: an expensive
    model chosen silently is a bill, not a rescue."""
    usable = [n for n in names if not any(w in n.lower() for w in MODEL_EXCLUDE)]
    return sorted(usable, key=_model_rank)[0] if usable else None


def use_model(name, reason=""):
    """Switch the model every later request uses, and say so in one line on screen."""
    global MODEL_IN_USE
    MODEL_IN_USE = name
    S["model"] = name
    S["model_switched"] = (name != MODEL)
    if S["model_switched"]:
        _say(f"MODEL {MODEL} is unavailable; using {name} instead{reason}")
    return name


def resolve_model(client=None, configured=None):
    """Called when the configured model 404s. Asks the key what it can use, announces the
    switch and returns the new name, or None if nothing usable came back."""
    configured = configured or MODEL
    if not MODEL_FALLBACK:
        _say(f"MODEL_FALLBACK is off, so {configured} was not replaced automatically.")
        return None
    try:
        names = list_text_models(client)
    except Exception as e:
        _say(f"could not ask the provider which models this key can use "
             f"({type(e).__name__}: {str(e)[:60]})")
        return None
    if not names:
        _say("the provider listed no text-generation models for this key.")
        return None
    pick = pick_model(names)
    if pick is None:
        _say("the only models this key can use are pro/ultra tiers; not switching to one "
             "automatically. Set MODEL in build/w3_core.py by hand if you want one.")
        return None
    return use_model(pick, f" ({len(names)} models offered to this key)")


def verify_key(key, provider):
    """One live request, up to three attempts with a printed backoff (5 s, 10 s) — but only
    a network fault is ever retried; every other outcome is decided on the first attempt.

    Returns exactly one of: ok · rate-limited · bad-key · model-not-found · network · server ·
    provider-error, for BOTH providers (the Groq arm classifies through the same
    _classify_error, so a Groq timeout is never reported as a rejected key). A 404 is NOT retried — it is deterministic, so a second attempt would
    waste thirty seconds to fail identically; instead we ask the key which models it can
    use and switch (MODEL_FALLBACK), which is the only thing that can actually fix it.
    """
    if provider == "groq":
        try:
            import groq  # noqa
        except ImportError:
            _p("Installing groq SDK (Groq key found, no Gemini key)…")
            os.system(f"{sys.executable} -m pip install -q groq")
        try:
            import groq
            S["client"] = groq.Groq(api_key=key)
            _pace()
            S["client"].chat.completions.create(model=GROQ_MODEL, max_tokens=5,
                                                messages=[{"role": "user", "content": "Reply with the single word OK."}])
            S["n_requests"] += 1
            return "ok"
        except Exception as e:
            kind = _classify_error(e)
            _say(f"Groq key check: {kind} ({type(e).__name__}: {str(e)[:90]})")
            # Same honesty rule as the Gemini arm: a Groq timeout is a network problem and a
            # Groq 429 is a busy quota — neither is a rejected key.
            if kind == "rate-limit":
                return "rate-limited"
            if kind in ("network", "server", "model-not-found", "bad-key"):
                return kind
            return "provider-error"
    S["client"] = genai.Client(api_key=key, http_options=gtypes.HttpOptions(timeout=30000))
    use_model(MODEL)                       # a re-run of cell 0 starts from the configured name again
    switched = False
    last = "network"
    for attempt in (1, 2, 3):
        try:
            _pace()
            resp = S["client"].models.generate_content(
                model=MODEL_IN_USE, contents="Reply with the single word OK.",
                config=gtypes.GenerateContentConfig(temperature=0, max_output_tokens=5))
            S["n_requests"] += 1
            _ = _safe_text(resp)
            return "ok"
        except Exception as e:
            kind = _classify_error(e)
            last = kind
            _say(f"key check attempt {attempt}/3: {kind} ({type(e).__name__}: {str(e)[:90]})")
            if kind == "rate-limit":
                return "rate-limited"          # the key works; quota is busy right now
            if kind == "bad-key":
                return "bad-key"
            if kind == "model-not-found":
                # Deterministic: do not back off and try the same name again.
                if switched or resolve_model(S["client"], model_from_error(e)) is None:
                    return "model-not-found"
                switched = True
                continue
            if attempt < 3:
                _countdown(5 * attempt, "  backing off")
    return last if last in ("network", "server") else "provider-error"


def _pace():
    """≥ 7 s between live requests, with a visible countdown."""
    wait = PACE_SECONDS - (time.time() - S["last_request"])
    if wait > 0:
        _countdown(wait, "free-tier pacing")
    S["last_request"] = time.time()


# ----------------------------------------------------------------------------
# Cell 0 — bootstrap
# ----------------------------------------------------------------------------
def bootstrap(reconnect=False):
    t0 = time.time()
    _hr("MGRCM0065 Week 3 · AI for Finance — START HERE" if not reconnect else "🔁 reconnect")
    try:
        load_data()
    except Exception as e:
        S["mode"] = MODE_ERROR
        _p(f"Data error: {type(e).__name__}: {e}")
        _p(f"=== {MODE_ERROR} ===")
        _say("data failed to load — watch the projector; we fix this at the break")
        return
    ver = S["version"]
    cache_kind = "SAMPLE" if S["cache_meta"].get("sample", True) else "real"
    _p(f"Notebook v{ver.get('notebook', '?')} · data v{ver.get('data', '?')} · cache: {cache_kind}")
    _p(f"prices to {TODAY} · {len(S['headlines'])} headlines + "
       f"{0 if S['fresh'] is None else len(S['fresh'])} fresh")
    _p(f"Data source: {S['data_source']}")
    _recorded = str(S["cache_meta"].get("today_constant", "") or "")
    if _recorded and _recorded != TODAY:
        _p(f"⚠️  recorded answers are for {_recorded} but prices run to {TODAY} — Labs 3 and 4 "
           "would fall through to 'no cached answer'; tell the instructor.")
    if not reconnect:
        fetch_repo_updates()
    sdk = ensure_sdk()
    if not sdk:
        S["mode"] = MODE_STUDIO
    key, provider, status = find_key()
    S["provider"] = provider
    S["key_status"] = status
    if FORCE_CACHED:
        S["mode"] = MODE_STUDIO
        _say("FORCE_CACHED is on → STUDIO MODE regardless of your key.")
    elif not sdk:
        pass
    elif status == "notebook-access":
        S["mode"] = MODE_STUDIO
        _say("Key present but THIS copy cannot see it (NotebookAccessError): click 🔑 in the left "
             "sidebar, toggle 'Notebook access' ON for GEMINI_API_KEY, then re-run this cell.")
    elif key is None:
        if S["in_colab"] and not reconnect:
            try:
                from getpass import getpass
                pasted = getpass("No key found in Secrets. Paste your GEMINI_API_KEY now, or press Enter for STUDIO MODE: ")
                if pasted.strip():
                    key, provider, status = pasted.strip(), "gemini", "pasted"
                    S["provider"] = provider
            except Exception:
                pass
        if key is None:
            S["mode"] = MODE_STUDIO
            _say("No key found (Secrets: GEMINI_API_KEY / GROQ_API_KEY; or the environment). "
                 "STUDIO MODE is fine — see the sub-line below.")
    if key is not None and S["mode"] is None:
        result = verify_key(key, provider)
        if result in ("ok", "rate-limited"):
            S["mode"] = MODE_LIVE
            if result == "rate-limited":
                _say("Key works but the free-tier quota is busy right now — cached twins will fill any gap.")
        elif result == "bad-key":
            S["mode"] = MODE_STUDIO
            _say(f"Key rejected by the provider: re-copy it from {_key_console(provider)} (no spaces), "
                 "save the Secret, re-run this cell. STUDIO MODE meanwhile.")
        elif result == "model-not-found":
            S["mode"] = MODE_STUDIO
            _say(f"The model {GROQ_MODEL if provider == 'groq' else MODEL} is not available to this "
                 "API key. This usually means the provider has renamed or withdrawn it. Your key and "
                 "your Wi-Fi are both fine. Tell the instructor, who runs build/list_models.py to see "
                 "what the key can use and then sets MODEL in build/w3_core.py. STUDIO MODE "
                 "meanwhile — every number, order and refusal in today's session still runs.")
        elif result == "server":
            S["mode"] = MODE_STUDIO
            _say("The provider's own servers returned an error three times (nothing to do with "
                 "your key or your Wi-Fi). Wait a minute, then re-run this cell — that is the "
                 "retry button. STUDIO MODE meanwhile.")
        elif result == "network":
            S["mode"] = MODE_STUDIO
            _say("Network timed out three times: check Wi-Fi, then re-run this cell (that is the retry "
                 "button). STUDIO MODE meanwhile.")
        else:
            S["mode"] = MODE_STUDIO
            _say("The provider returned an error we do not have a specific remedy for — the exact "
                 "wording is in the attempt lines above; show them to the instructor. Re-running "
                 "this cell is the retry button. STUDIO MODE meanwhile.")
    _hr()
    _p(f"=== {S['mode']} ===")
    if S["mode"] == MODE_STUDIO:
        _say(STUDIO_SUBLINE)
        meta = S["cache_meta"]
        if meta.get("sample", True):
            _p(f"⚠️  {SAMPLE_BANNER}")
            _say(f"cached wording is stand-in text; the real cache will be produced with: "
                 f"{meta.get('model_placeholder', MODEL)}")
        else:
            _say(f"cached answers were produced with: {meta.get('model', MODEL)} "
                 f"(instructor's pre-run on {meta.get('generated_on', '?')})")
    if S["mode"] == MODE_LIVE:
        if provider == "groq":
            _say(f"provider: Groq · model {GROQ_MODEL} · single-turn cells live; Lab 3 replays the "
                 "instructor's trace")
        else:
            _say(f"model: {MODEL_IN_USE} · pacing ≥ {PACE_SECONDS} s between requests · "
                 f"read-only tool-call cap {TOOL_CAP} per turn (+{ORDER_CAP} orders)")
    _p("Runtime type: CPU (no GPU needed)")
    _say("Google account: Colab, the Secret and the key must all be the same PERSONAL account "
         "(check the avatar, top right).")
    _p("Do NOT use Runtime → Run all — this notebook is a sequence of reveals.")
    _p(f"(cell 0 finished in {time.time() - t0:.1f} s at {_now()})")
    globals()["_BOOTSTRAPPED"] = True


def reconnect():
    """🔁 Disconnected? Run this first. Idempotent; keeps a real lock if it survived.

    If the runtime lost the lock, a PLACEHOLDER lock is set so Labs 2–4 open. The placeholder is
    marked "restored": 1.4 never scores it (it would read 0/8) and 1.2 is allowed to overwrite it,
    so a student who wants the scoreboard back runs 1.2 → 1.3 → 1.4 once more.
    """
    global PREDICTIONS_LOCKED
    was_alive = bool(globals().get("_BOOTSTRAPPED", False))   # read BEFORE bootstrap() sets it
    bootstrap(reconnect=True)
    if PREDICTIONS_LOCKED is None:
        PREDICTIONS_LOCKED = (_now(), {}, "restored")
        globals()["PREDICTIONS_LOCKED"] = PREDICTIONS_LOCKED
        if was_alive:
            # Cell 0 already ran in this kernel: nothing was lost — the student simply never ran 1.2.
            _say("You have not locked any predictions yet — run 1.2 (then 1.3 and 1.4) if you want "
                 "the scoreboard; Labs 2–4 are open either way, on your honour.")
        else:
            _say("Your prediction lock was lost with the runtime — Labs 2–4 are open on your honour; "
                 "the calls you noted down are the record. Want the scoreboard back? Run 1.2 (set the "
                 "dropdowns to what you wrote), then 1.3 and 1.4 — the notebook allows this once.")
    _p("🔁 reconnected — carry on from where you were.")


# ----------------------------------------------------------------------------
# Gating
# ----------------------------------------------------------------------------
def require(condition, message):
    if not condition:
        _p(f"⛔ {message}")
        return False
    return True


def locked():
    """True for a real lock AND for the reconnect placeholder (both open Labs 2–4)."""
    return globals().get("PREDICTIONS_LOCKED") is not None


def lock_is_placeholder():
    """True only for the placeholder a 🔁 cell sets after the runtime lost the real lock."""
    lk = globals().get("PREDICTIONS_LOCKED")
    return lk is not None and len(lk) >= 3 and lk[2] == "restored"


# ----------------------------------------------------------------------------
# Tools (Lab 3). All slice by TODAY. GBP is notional only (no FX applied).
# ----------------------------------------------------------------------------
def _series(ticker):
    if ticker not in TICKERS:
        raise ValueError(f"unknown ticker {ticker!r}; choose from {', '.join(TICKERS)}")
    df = S["prices"][S["prices"]["ticker"] == ticker]
    return df[df["date"] <= TODAY].set_index("date")


def _prior_trading_day(s, date):
    date = pd.to_datetime(date).strftime("%Y-%m-%d")
    if date > TODAY:
        raise ValueError(f"{date} is after TODAY ({TODAY}) — the future is not available")
    idx = s.index[s.index <= date]
    if len(idx) == 0:
        raise ValueError(f"no prices before {date}")
    return idx[-1], (idx[-1] != date)


def get_price(ticker, date):
    """Closing price in USD on `date` (or the nearest PRIOR trading day)."""
    s = _series(str(ticker).upper())
    d, moved = _prior_trading_day(s, date)
    out = {"ticker": ticker.upper(), "date": d, "close_usd": round(float(s.loc[d, "close"]), 2)}
    if moved:
        out["note"] = f"{date} was not a trading day; nearest prior trading day {d} used"
    return out


def get_return(ticker, start, end):
    """Total return (adjusted closes) from `start` to `end`, both nearest-prior-trading-day."""
    s = _series(str(ticker).upper())
    d0, _ = _prior_trading_day(s, start)
    d1, _ = _prior_trading_day(s, end)
    r = float(s.loc[d1, "adj_close"] / s.loc[d0, "adj_close"] - 1)
    return {"ticker": ticker.upper(), "start": d0, "end": d1, "return_pct": round(100 * r, 2)}


def get_volatility(ticker, window_days=30):
    """Annualised volatility (%) of daily log returns over the last `window_days` trading days."""
    s = _series(str(ticker).upper())["adj_close"]
    w = int(window_days)
    lr = np.log(s).diff().dropna().tail(w)
    return {"ticker": ticker.upper(), "window_days": w, "as_of": TODAY,
            "annualised_vol_pct": round(float(lr.std() * np.sqrt(252) * 100), 2)}


def get_headlines(ticker):
    """Headlines for `ticker` dated ≤ TODAY: date + text only (no returns, no labels)."""
    t = str(ticker).upper()
    rows = [S["headlines"]] + ([S["fresh"]] if S["fresh"] is not None else [])
    df = pd.concat(rows, ignore_index=True)
    df = df[(df["ticker"] == t) & (df["date"] <= TODAY)].sort_values("date")
    return {"ticker": t, "headlines": [{"date": r.date, "text": r.headline} for r in df.itertuples()]}


def guard_check(ticker, gbp_notional, side):
    """The governance guard lives HERE, not in the mandate text."""
    side = str(side).upper()
    if str(ticker).upper() not in TICKERS:
        return f"REFUSED: unknown ticker {ticker}"
    if side != "BUY":
        return f"REFUSED: long-only mandate — {side} orders are not permitted"
    if float(gbp_notional) > GUARD["MAX_ORDER_GBP"]:
        return f"REFUSED: £{float(gbp_notional):,.0f} exceeds MAX_ORDER_GBP £{GUARD['MAX_ORDER_GBP']:,.0f}"
    if GUARD["REQUIRE_HUMAN_APPROVAL"]:
        return "REFUSED: REQUIRE_HUMAN_APPROVAL is on — no human has approved this order"
    return "FILLED"


def place_order(ticker, gbp_notional, side="BUY"):
    """Paper order (GBP notional, long-only). Every call is written to ORDER_BOOK."""
    status = guard_check(ticker, gbp_notional, side)
    rec = {"time": _now(), "ticker": str(ticker).upper(), "side": str(side).upper(),
           "gbp_notional": float(gbp_notional), "status": status}
    if status == "FILLED":
        rec["price_usd"] = get_price(ticker, TODAY)["close_usd"]
    ORDER_BOOK.append(rec)
    return {"status": status, "ticker": rec["ticker"], "gbp_notional": rec["gbp_notional"], "side": rec["side"]}


TOOLS = {"get_price": get_price, "get_return": get_return, "get_volatility": get_volatility,
         "get_headlines": get_headlines, "place_order": place_order}
READ_ONLY_TOOLS = ["get_price", "get_return", "get_volatility", "get_headlines"]
MANDATE_TOOLS = READ_ONLY_TOOLS + ["place_order"]

_TOOL_DECLS = {
    "get_price": ("Closing share price in USD of a ticker on a date (YYYY-MM-DD); uses the nearest prior trading day.",
                  {"ticker": ("STRING", "one of " + ", ".join(TICKERS)), "date": ("STRING", "YYYY-MM-DD, not after today")},
                  ["ticker", "date"]),
    "get_return": ("Total return in percent between two dates (YYYY-MM-DD) using adjusted closes.",
                   {"ticker": ("STRING", "ticker"), "start": ("STRING", "YYYY-MM-DD"), "end": ("STRING", "YYYY-MM-DD")},
                   ["ticker", "start", "end"]),
    "get_volatility": ("Annualised volatility in percent over the last N trading days.",
                       {"ticker": ("STRING", "ticker"), "window_days": ("INTEGER", "number of trading days, e.g. 30")},
                       ["ticker", "window_days"]),
    "get_headlines": ("Recent news headlines for a ticker (date and text only).",
                      {"ticker": ("STRING", "ticker")}, ["ticker"]),
    "place_order": ("Place a paper (simulated) order in GBP notional. Long-only: side must be BUY.",
                    {"ticker": ("STRING", "ticker"), "gbp_notional": ("NUMBER", "order size in GBP"),
                     "side": ("STRING", "BUY")}, ["ticker", "gbp_notional", "side"]),
}


def tool_declarations(names):
    """google-genai function declarations (verified against types.py 2.24.0)."""
    decls = []
    for n in names:
        desc, props, req = _TOOL_DECLS[n]
        decls.append(gtypes.FunctionDeclaration(
            name=n, description=desc,
            parameters=gtypes.Schema(
                type="OBJECT",
                properties={k: gtypes.Schema(type=t, description=d) for k, (t, d) in props.items()},
                required=req)))
    return [gtypes.Tool(function_declarations=decls)]


# Models invent argument names. The live run on 17 Sep 2026 called
# place_order(size=200000, ticker='AAPL') — 'size' for 'gbp_notional', and no 'side' at all.
# run_tool would answer with a TypeError message, which costs a turn and, worse, leaves the
# recorded step with no status so run_3_2 prints "wants to buy £0". We canonicalise the names
# BEFORE the call so both the execution and the recorded script use the declared ones.
_ARG_ALIASES = {
    "place_order": {"size": "gbp_notional", "size_gbp": "gbp_notional", "gbp": "gbp_notional",
                    "gbp_size": "gbp_notional", "gbp_amount": "gbp_notional", "amount": "gbp_notional",
                    "notional": "gbp_notional", "order_size": "gbp_notional", "quantity": "gbp_notional",
                    "value": "gbp_notional", "symbol": "ticker", "stock": "ticker",
                    "direction": "side", "action": "side"},
    "get_price": {"symbol": "ticker", "stock": "ticker", "as_of": "date", "day": "date"},
    "get_return": {"symbol": "ticker", "stock": "ticker", "from": "start", "to": "end",
                   "start_date": "start", "end_date": "end"},
    "get_volatility": {"symbol": "ticker", "stock": "ticker", "window": "window_days",
                       "days": "window_days", "lookback": "window_days"},
    "get_headlines": {"symbol": "ticker", "stock": "ticker"},
}


def _clean_number(v):
    """'£200,000' / '200000 GBP' -> 200000.0; anything else is handed back untouched."""
    if isinstance(v, str):
        t = re.sub(r"[£$,\s]|gbp|usd", "", v, flags=re.I)
        try:
            return float(t)
        except ValueError:
            return v
    return v


def normalise_args(name, args):
    """Map the argument names a model actually produces onto the ones we declared."""
    alias = _ARG_ALIASES.get(name, {})
    out = {}
    for k, v in (args or {}).items():
        out[alias.get(str(k).lower(), k)] = v
    if name == "place_order":
        if "gbp_notional" in out:
            out["gbp_notional"] = _clean_number(out["gbp_notional"])
        out.setdefault("side", "BUY")
    return out


def run_tool(name, args):
    if name not in TOOLS:
        return {"error": f"unknown tool {name}"}
    try:
        return TOOLS[name](**args)
    except TypeError as e:
        return {"error": f"bad arguments for {name}: {e}"}
    except Exception as e:
        return {"error": f"{type(e).__name__}: {e}"}


def _fmt_args(args):
    return ", ".join(f"{k}={v!r}" for k, v in args.items())


def _print_tool_call(name, args, result):
    res = json.dumps(result, default=str, ensure_ascii=False)
    if len(res) > 400:
        res = res[:400] + " …"
    _trace(f"🔧 TOOL CALL {name}({_fmt_args(args)}) → {res}")


# ----------------------------------------------------------------------------
# Safe text extraction and the three-tier parser
# ----------------------------------------------------------------------------
def _safe_text(resp):
    """Text from the first candidate; None if empty/blocked. Never raises."""
    try:
        cands = resp.candidates or []
        if not cands or cands[0].content is None or not cands[0].content.parts:
            return None
        gm = getattr(cands[0], "grounding_metadata", None)
        if gm is not None and (getattr(gm, "web_search_queries", None) or getattr(gm, "grounding_chunks", None)):
            S["grounding_seen"] = True
        parts = [p.text for p in cands[0].content.parts if getattr(p, "text", None) and not getattr(p, "thought", False)]
        text = "\n".join(parts).strip()
        return text or None
    except Exception:
        return None


CALL_SCHEMA = {
    "type": "ARRAY",
    "items": {"type": "OBJECT",
              "properties": {"id": {"type": "STRING"},
                             "call": {"type": "STRING", "enum": ["UP", "DOWN", "FLAT"]},
                             "reason": {"type": "STRING"}},
              "required": ["id", "call", "reason"]},
}


def parse_calls(text, ids):
    """Tier 1 JSON → tier 2 regex → tier 3 UNPARSED. Returns {id: (call, reason)}. Never raises."""
    out = {}
    if text:
        try:  # tier 1
            data = json.loads(text)
            if isinstance(data, dict):
                data = data.get("calls") or data.get("items") or list(data.values())
            for row in data:
                i = str(row.get("id", "")).strip().upper()
                c = str(row.get("call", "")).strip().upper()
                if i in ids and c in ("UP", "DOWN", "FLAT"):
                    out[i] = (c, str(row.get("reason", "")).strip())
        except Exception:
            pass
        if len(out) < len(ids):  # tier 2
            for i in ids:
                if i in out:
                    continue
                m = re.search(rf"{i}\b[^A-Za-z]{{0,20}}(UP|DOWN|FLAT)\b", text, re.I)
                if m:
                    out[i] = (m.group(1).upper(), "(parsed from free text)")
    for i in ids:  # tier 3
        if i not in out:
            out[i] = ("UNPARSED", (text or "(no text)")[:80])
    return out


# ----------------------------------------------------------------------------
# ask(): the only path to a model
# ----------------------------------------------------------------------------
def _cache_key(cell_id, prompt):
    return f"{cell_id}:{_sha(prompt)}"


def _cached_entry(cell_id, prompt):
    return S["cache"].get(_cache_key(cell_id, prompt))


def _record(cell_id, prompt, text, tool_script=None, extra=None):
    """build_cache.py switches recording on; live results are collected here."""
    if S["recording"] is None:
        return
    key = _cache_key(cell_id, prompt)
    entry = S["recording"].setdefault(key, {"cell_id": cell_id, "prompt": prompt, "text": None,
                                            "tool_script": [], "variants": {}})
    if text is not None and entry["text"] is None:
        entry["text"] = text
    if tool_script:
        entry["tool_script"] = tool_script
    if extra:
        for k, v in extra.items():
            if isinstance(v, dict):
                entry.setdefault(k, {}).update(v)
            else:
                entry[k] = v


def _replay(entry, tools):
    """Replay a stored tool script through the REAL tool functions (200 ms per step)."""
    script = entry.get("tool_script") or []
    alt = entry.get("tool_script_refusal") or []
    # The refusal script is swapped in only when the fill script's first order would be refused
    # under the student's guard AND the refusal script itself places an order — otherwise the fill
    # script replays and the guard refuses it, so the book and the closing text still agree.
    if alt and any(step["name"] == "place_order" for step in alt) \
            and any(step["name"] == "place_order" for step in script):
        first = next(step for step in script if step["name"] == "place_order")
        if guard_check(first["args"].get("ticker"), first["args"].get("gbp_notional", 0),
                       first["args"].get("side", "BUY")) != "FILLED":
            script = alt
    executed = []
    for step in script:
        time.sleep(0.2)
        result = run_tool(step["name"], dict(step["args"]))
        _print_tool_call(step["name"], step["args"], result)
        executed.append({"name": step["name"], "args": step["args"], "result": result})
    text = entry.get("text")
    orders = [e for e in executed if e["name"] == "place_order"]
    variants = entry.get("variants") or {}
    if orders:
        statuses = [o["result"].get("status", "") for o in orders]
        if all(st == "FILLED" for st in statuses):
            text = variants.get("after_fill") or text
        elif statuses[-1] == "FILLED":
            text = variants.get("after_partial") or (
                "(no cached closing text for this outcome — the instructor's pre-run recorded a fill "
                "and a refusal but not a refused-then-filled turn; the ORDER BOOK above is the record)")
        else:
            text = variants.get("after_refusal") or text
    return text, executed


def _groq_generate(prompt, temperature, persona, schema):
    msgs = ([{"role": "system", "content": persona}] if persona else []) + [{"role": "user", "content": prompt}]
    kw = {"model": GROQ_MODEL, "messages": msgs, "temperature": float(temperature), "max_tokens": 1200}
    if schema:
        kw["response_format"] = {"type": "json_object"}
        msgs[-1]["content"] += "\nReturn JSON: {\"calls\": [{\"id\": …, \"call\": UP|DOWN|FLAT, \"reason\": …}]}"
    _pace()
    r = S["client"].chat.completions.create(**kw)
    S["n_requests"] += 1
    return r.choices[0].message.content


def _gemini_generate(cell_id, prompt, temperature, persona, schema, tools):
    """One turn, including our OWN tool loop (automatic function calling disabled)."""
    cfg = gtypes.GenerateContentConfig(
        temperature=float(temperature),
        system_instruction=persona if persona else None,
        automatic_function_calling=gtypes.AutomaticFunctionCallingConfig(disable=True))
    if schema:
        cfg.response_mime_type = "application/json"
        cfg.response_schema = schema
    if tools:
        cfg.tools = tool_declarations(tools)
        _say(f"tool-call cap applied: {TOOL_CAP} read-only calls per turn"
             + (f" plus up to {ORDER_CAP} orders" if any(t not in READ_ONLY_TOOLS for t in tools) else "")
             + " (automatic function calling OFF; loop is ours)")
    contents = [gtypes.Content(role="user", parts=[gtypes.Part.from_text(text=prompt)])]
    tool_script, n_reads, n_orders = [], 0, 0
    for _round in range(TOOL_CAP + ORDER_CAP + 2):
        _pace()
        resp = S["client"].models.generate_content(model=MODEL_IN_USE, contents=contents, config=cfg)
        S["n_requests"] += 1
        _p(f"   [request #{S['n_requests']} · {_now()}]")
        cand = (resp.candidates or [None])[0]
        parts = list(cand.content.parts) if (cand is not None and cand.content is not None and cand.content.parts) else []
        calls = [p.function_call for p in parts if getattr(p, "function_call", None)]
        if not calls:
            return _safe_text(resp), tool_script
        contents.append(cand.content)
        responses = []
        for fc in calls:
            args = normalise_args(fc.name, dict(fc.args or {}))
            is_order = fc.name not in READ_ONLY_TOOLS
            spent = n_orders if is_order else n_reads
            cap = ORDER_CAP if is_order else TOOL_CAP
            if spent >= cap:
                what = "order cap" if is_order else "read-only tool-call cap"
                result = {"error": f"{what} of {cap} reached — answer with what you have"}
                _trace(f"🔧 TOOL CALL {fc.name}({_fmt_args(args)}) → CAP REACHED ({cap})")
            else:
                result = run_tool(fc.name, args)
                if is_order:
                    n_orders += 1
                else:
                    n_reads += 1
                _print_tool_call(fc.name, args, result)
                tool_script.append({"name": fc.name, "args": args, "result": result})
            responses.append(gtypes.Part.from_function_response(name=fc.name, response={"result": result}))
        contents.append(gtypes.Content(role="user", parts=responses))
        # Silence function calling only once every tool THIS turn was given has run out of budget.
        # Gating it on the read budget alone (as it was) forbade the model from ever emitting the
        # order, whatever the order cap said; gating it on both counters regardless of which tools
        # were offered would leave a read-only turn burning rounds on calls that can only be refused.
        reads_left = n_reads < TOOL_CAP and any(t in READ_ONLY_TOOLS for t in tools)
        orders_left = n_orders < ORDER_CAP and any(t not in READ_ONLY_TOOLS for t in tools)
        if not reads_left and not orders_left:
            cfg.tool_config = gtypes.ToolConfig(function_calling_config=gtypes.FunctionCallingConfig(mode="NONE"))
    return None, tool_script


def ask(cell_id, prompt, temperature=0, tools=None, persona=None, schema=None):
    """Returns {'text', 'source' ∈ live|cached|miss, 'tool_calls', 'sample'}. Never raises for cache paths."""
    entry = _cached_entry(cell_id, prompt)
    sample = bool(S["cache_meta"].get("sample", True))
    live = (S["mode"] == MODE_LIVE) and not FORCE_CACHED
    if live and S["provider"] == "groq" and tools:
        _say("Groq provider: Lab 3 replays the instructor's trace (Groq path is single-turn only).")
        live = False
    book_before = list(ORDER_BOOK)
    if live:
        attempt, switched = 0, False
        while attempt < 2:
            attempt += 1
            # Rewind the book before EVERY attempt. The retry re-runs the whole tool loop, so a
            # 5xx arriving after a fill would otherwise book the same order twice and put two
            # £1,000,000 fills on the projector against a £1,000,000 mandate.
            del ORDER_BOOK[:]
            ORDER_BOOK.extend(book_before)
            try:
                if S["provider"] == "groq":
                    text, script = _groq_generate(prompt, temperature, persona, schema), []
                else:
                    text, script = _gemini_generate(cell_id, prompt, temperature, persona, schema, tools)
                if text is None:
                    _p("empty/blocked response from the model → serving the cached twin")
                    break
                _record(cell_id, prompt, text, script)
                return {"text": text, "source": "live", "tool_calls": script, "sample": False}
            except Exception as e:
                kind = _classify_error(e)
                if kind == "rate-limit":
                    _p(f"429 / RESOURCE_EXHAUSTED (free-tier limit) → {CACHED_BANNER}")
                    break
                if kind == "model-not-found":
                    # The model went away mid-session. Retrying the same name cannot help,
                    # so either switch to one this key can use or serve the cached twin.
                    # A successful switch buys one extra pass (on the NEW name) even if the
                    # 404 arrived on the second attempt; `switched` makes that a one-shot,
                    # so this cannot loop.
                    if S["provider"] == "gemini" and not switched and resolve_model(
                            S["client"], model_from_error(e)) is not None:
                        switched = True
                        attempt -= 1
                        continue
                    _say(f"the model {MODEL_IN_USE} is not available to this key (this is not a "
                         f"network problem) → {CACHED_BANNER}")
                    break
                _p(f"model error ({type(e).__name__}: {str(e)[:100]})")
                if attempt == 1:
                    _countdown(30, "one retry after")
                else:
                    _p(f"second failure → {CACHED_BANNER}")
        del ORDER_BOOK[:]
        ORDER_BOOK.extend(book_before)
    if entry is None:
        _say("no cached answer for your edited prompt — reset to default or use a live laptop.")
        return {"text": None, "source": "miss", "tool_calls": [], "sample": sample}
    _p(CACHED_BANNER)
    if sample:
        _p("   (SAMPLE — stand-in text, not model output)")
    if tools:
        text, executed = _replay(entry, tools)
    else:
        text, executed = entry.get("text"), []
    return {"text": text, "source": "cached", "tool_calls": executed, "sample": sample}


# ----------------------------------------------------------------------------
# Prompts (shared by the notebook and the cache builders — the hash is the key)
# ----------------------------------------------------------------------------
def flat_rule():
    return (f"FLAT means the return over the first trading session that could react to the headline "
            f"is between -{FLAT_BAND*100:.1f}% and +{FLAT_BAND*100:.1f}%.")


def prompt_lab1(rows, anonymised=False):
    lines = []
    for r in rows.itertuples():
        if anonymised:
            lines.append(f"{r.id}: {ANON_HEADLINES.get(r.id, r.headline)}")
        else:
            lines.append(f"{r.id} ({r.date}, {COMPANY.get(r.ticker, r.ticker)}, {r.ticker}): {r.headline}")
    head = ("You are a financial expert with stock recommendation experience. For each headline below, "
            "decide whether the company's share price went UP, DOWN or stayed FLAT over the first "
            f"trading session that could react to it. {flat_rule()} "
            "Use only what the headline says. Give a one-sentence reason per headline. "
            "Return a JSON array of objects with keys id, call (UP, DOWN or FLAT) and reason.")
    if anonymised:
        head += " The company names, tickers and dates have been removed on purpose."
    return head + "\n\n" + "\n".join(lines)


def prompt_recall(row):
    return (f"Here is a news headline published on {row['date']} about {COMPANY.get(row['ticker'], row['ticker'])} "
            f"({row['ticker']}): \"{row['headline']}\".\n"
            "In one paragraph of at most five sentences, describe what happened to the company's share price and "
            "to the company in the weeks and months that followed this headline.")


def prompt_hallucination():
    names = ", ".join(f"{COMPANY[t]} ({t})" for t in HALLUCINATION_TICKERS)
    return (f"Give your best estimate of the closing share price in US dollars of {names} on {HALLUCINATION_DATE}. "
            "Answer with exactly three lines in the form 'TICKER: price', for example 'XYZ: 123.45'. "
            "Do not add disclaimers, ranges or caveats — a single best estimate per line.")


PERSONAS = {
    "risk": "You are a cautious risk officer at a UK pension fund. You answer in three sentences.",
    "pm": "You are a bullish hedge-fund portfolio manager. You answer in three sentences.",
}


def prompt_persona():
    return (f"Should a UK pension fund buy {COMPANY[COMMITTEE_TICKER]} ({COMMITTEE_TICKER}) shares this month? "
            "End with one line 'RECOMMENDATION: BUY' or 'RECOMMENDATION: DO NOT BUY'.")


def prompt_temperature():
    return (f"In one sentence, what is the biggest risk of holding {COMPANY[COMMITTEE_TICKER]} "
            f"({COMMITTEE_TICKER}) shares for the next year?")


ANALYST_PROMPT_TEMPLATE = (
    "Today is {today}. You are a junior equity analyst with access to tools. Write a short briefing "
    "(at most 120 words) on {company} ({ticker}) for a portfolio manager: current price, performance over "
    "the last month and the last three months, current volatility, and the most recent news. "
    "Use the tools for every number you state. Do not place any orders.")


def prompt_analyst():
    return ANALYST_PROMPT_TEMPLATE.format(today=TODAY, company=COMPANY[COMMITTEE_TICKER], ticker=COMMITTEE_TICKER)


MANDATE_PREAMBLE = ("Paper-trading classroom simulation; no real money. You MUST call place_order exactly once "
                    "with your chosen ticker and size before your final answer.")
MANDATE_BODY = (" Today is {today}. You manage a £{gbp:,} paper mandate, long-only, and may invest in any of "
                "{tickers}. Using your tools, choose ONE stock and ONE order size in GBP, place the order, "
                "then explain your choice in at most 80 words. If an order is refused, say so and explain "
                "what you will do next.")


def mandate_text(mandatory=True, extra=""):
    body = MANDATE_BODY.format(today=TODAY, gbp=MANDATE_GBP, tickers=", ".join(TICKERS))
    return ((MANDATE_PREAMBLE if mandatory else
             "Paper-trading classroom simulation; no real money. You have a place_order tool.") + body + extra).strip()


# ----------------------------------------------------------------------------
# Scoring helpers
# ----------------------------------------------------------------------------
def score(calls, rows):
    """calls: {id: call}; returns (hits, n)."""
    hits = sum(1 for r in rows.itertuples() if calls.get(r.id, "").upper() == str(r.label).upper())
    return hits, len(rows)


def _label_word(label):
    return {"UP": "UP", "DOWN": "DOWN", "FLAT": "FLAT"}.get(str(label).upper(), str(label))


def _fmt_pct(x):
    return "n/a" if x is None or (isinstance(x, float) and math.isnan(x)) else f"{100*float(x):+.2f}%"


# ----------------------------------------------------------------------------
# Cell runners — each notebook code cell calls exactly one of these
# ----------------------------------------------------------------------------
def run_1_1():
    """Print the 8 headlines with the FLAT rule."""
    h = S["headlines"]
    _hr("Lab 1 · 8 real, dated headlines")
    _say(f"FLAT_BAND = ±{FLAT_BAND*100:.1f}%  — {flat_rule()}")
    _say("Scoring uses the return over the first trading session that could react to the headline.")
    _p("")
    for r in h.itertuples():
        _trace(f"{r.id} · {r.date} · {COMPANY.get(r.ticker, r.ticker)} ({r.ticker})")
        _trace(f"     {r.headline}")
    _p("")
    _say("Note your 8 calls somewhere of your own, then set the dropdowns in 1.2 and run it to lock.")


def lock_predictions(calls):
    """1.2 — lock once, with a timestamp; refuse to re-lock."""
    global PREDICTIONS_LOCKED
    restoring = lock_is_placeholder()
    if locked() and not restoring:
        ts = PREDICTIONS_LOCKED[0]
        _say(f"⛔ locked at {ts} — your score stands; the dropdowns will not re-lock.")
        return
    calls = {k: str(v).upper() for k, v in calls.items()}
    bad = [k for k, v in calls.items() if v not in ("UP", "DOWN", "FLAT")]
    if bad:
        _p(f"⛔ {', '.join(bad)} not set to UP/DOWN/FLAT — fix the dropdowns and run again.")
        return
    PREDICTIONS_LOCKED = (_now(), calls, "locked")
    globals()["PREDICTIONS_LOCKED"] = PREDICTIONS_LOCKED
    if restoring:
        _say(f"🔒 predictions re-entered at {PREDICTIONS_LOCKED[0]} after the reconnect — on your honour, "
             "these should be the calls you noted before the reveal. Now run 1.3 and 1.4.")
    else:
        _say(f"🔒 predictions locked at {PREDICTIONS_LOCKED[0]} — on your honour; the notebook will notice a re-lock.")
    _say(" ".join(f"{k}:{v}" for k, v in calls.items()), indent="   ")


def run_1_3():
    """One batched, structured call: all 8 headlines to the model."""
    if not require(locked(), "run 1.2 first (lock your predictions) — the model is not called until you do"):
        return
    h = S["headlines"]
    prompt = prompt_lab1(h)
    _hr("Lab 1 · the model's calls (one request, JSON output)")
    _say(f"prompt includes: {flat_rule()}")
    res = ask("1.3", prompt, temperature=0, schema=CALL_SCHEMA)
    parsed = parse_calls(res["text"], list(h["id"]))
    S["results"]["1.3"] = {"parsed": parsed, "source": res["source"]}
    for i, (c, reason) in parsed.items():
        _trace(f"{i} → {c:<8} {reason}")
    if any(c == "UNPARSED" for c, _ in parsed.values()):
        _say("some rows are UNPARSED (scored as misses) — raw text shown; that is the finding, not a bug.")


def run_1_4():
    """Scoreboard: You k/8 · Model m/8 (flat band ±0.5 %)."""
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    if not require("1.3" in S["results"], "run 1.3 first (the model's calls)"):
        return
    if lock_is_placeholder():
        _hr("Lab 1 · scoreboard")
        _say("⛔ no scoreboard: your lock was lost with the runtime, so the notebook has no record of "
             "your calls — the calls you noted down are the record. To restore the scoreboard, run 1.2 "
             "(set the dropdowns to what you wrote), then 1.3 and 1.4 again.")
        return
    h = S["headlines"]
    you = PREDICTIONS_LOCKED[1]
    model = {i: c for i, (c, _) in S["results"]["1.3"]["parsed"].items()}
    ky, n = score(you, h)
    km, _ = score(model, h)
    S["results"]["1.4"] = {"you": ky, "model": km, "n": n}
    _hr("Lab 1 · scoreboard")
    _p(f"You: {ky}/{n} · Model: {km}/{n} (flat band ±{FLAT_BAND*100:.1f}%)")
    _p(f"model answer: {S['results']['1.3']['source']}")
    _p("")
    _p(f"{'id':<4} {'ticker':<6} {'truth':<6} {'ret':>8}  {'you':<6} {'model':<8} note")
    for r in h.itertuples():
        y, m = you.get(r.id, "—"), model.get(r.id, "—")
        note = "" if y == m else "← you disagreed with the model"
        if y != r.label and m == r.label:
            note = "← model right, you wrong"
        elif y == r.label and m != r.label:
            note = "← you right, model wrong"
        _p(f"{r.id:<4} {r.ticker:<6} {r.label:<6} {_fmt_pct(r.ret_event):>8}  {y:<6} {m:<8} {note}")
    _p("")
    _p("Q1 — Quote the headline where the model's reason surprised you.")


def _seed_row():
    h = S["headlines"]
    seed = h[h["group"] == "seed2020"]
    return (seed if len(seed) else h.tail(1)).iloc[0].to_dict()


def run_1_5a():
    """The reveal: the model 'predicts' the aftermath of the 2020 headline, beside the truth."""
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    row = _seed_row()
    _hr("Lab 1b · 1.5a — the 2020 headline")
    _trace(f"{row['id']} · {row['date']} · {COMPANY.get(row['ticker'], row['ticker'])}: {row['headline']}")
    res = ask("1.5a", prompt_recall(row), temperature=0)
    S["results"]["1.5a"] = res
    _p("")
    _p("MODEL SAYS:")
    for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
        _p("  " + line)
    _p("")
    _p("TRUTH FROM prices.csv (adjusted closes):")
    s = S["prices"][S["prices"]["ticker"] == row["ticker"]].set_index("date")["adj_close"]
    ev = row["event_date"]
    prior = s[s.index < ev]
    if len(prior) == 0:
        _say(f"  the price file does not go back far enough to score this headline (no close before "
             f"{ev} for {row['ticker']}) — pull prices from an earlier start date.")
    else:
        base = prior.iloc[-1]
        def ret_to(days):
            target = (pd.to_datetime(ev) + pd.Timedelta(days=days)).strftime("%Y-%m-%d")
            later = s[s.index <= target]
            return later.iloc[-1] / base - 1, later.index[-1]
        _p(f"  first reacting session {ev}: {_fmt_pct(row['ret_event'])}  (label {row['label']})")
        _p(f"  next session: {_fmt_pct(row['ret_next'])}")
        for d, name in ((35, "≈ 5 weeks later"), (182, "≈ 6 months later"), (365, "≈ 12 months later")):
            r, dd = ret_to(d)
            _p(f"  {name} ({dd}): {_fmt_pct(r)}")
    _p("")
    _p("Q2 — Quote the sentence in which the model tells you the future.")
    _say("A headline from 2020 is inside the model's reading; this is recall, not prediction.")


def _scored_arm(cell_id, rows, anonymised, title):
    prompt = prompt_lab1(rows, anonymised=anonymised)
    res = ask(cell_id, prompt, temperature=0, schema=CALL_SCHEMA)
    parsed = parse_calls(res["text"], list(rows["id"]))
    calls = {i: c for i, (c, _) in parsed.items()}
    k, n = score(calls, rows)
    S["results"][cell_id] = {"parsed": parsed, "hits": k, "n": n, "source": res["source"]}
    _p(f"{title}: Model {k}/{n} (flat band ±{FLAT_BAND*100:.1f}%) · {res['source']}")
    for r in rows.itertuples():
        c, reason = parsed[r.id]
        shown = ANON_HEADLINES.get(r.id, r.headline) if anonymised else r.headline
        _trace(f"{r.id} truth {r.label:<5} model {c:<8} {shown[:40]}…")
    return k, n


def run_1_5b():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5b — the same 8 headlines, names replaced by 'Company X'")
    _scored_arm("1.5b", S["headlines"], True, "Anonymised")


def run_1_5c():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5c — fresh headlines (last three weeks before build)")
    f = S["fresh"]
    if f is None or len(f) == 0:
        _say("fresh headlines unavailable — watch the projector (the instructor runs the canonical fresh arm).")
        S["results"]["1.5c"] = {"hits": None, "n": 0, "source": "unavailable"}
        return
    for r in f.itertuples():
        _trace(f"{r.id} · {r.date} · {r.ticker}: {r.headline}")
    _p("")
    _scored_arm("1.5c", f, False, "Fresh")
    if S.get("grounding_seen"):
        _say("note: the model reported search grounding — this is retrieval, not memory.")


def run_1_5d():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5d — comparison")
    R = S["results"]
    k13, n13 = (None, None)
    if "1.3" in R:   # scored from 1.3's own parsed calls, so the row is filled whenever 1.3 has run
        k13, n13 = score({i: c for i, (c, _) in R["1.3"]["parsed"].items()}, S["headlines"])
    rows = [("named, old (1.3)", k13, n13, R.get("1.3", {}).get("source")),
            ("anonymised, old (1.5b)", R.get("1.5b", {}).get("hits"), R.get("1.5b", {}).get("n"), R.get("1.5b", {}).get("source")),
            ("named, fresh (1.5c)", R.get("1.5c", {}).get("hits"), R.get("1.5c", {}).get("n"), R.get("1.5c", {}).get("source"))]
    _p(f"{'arm':<26}{'model':>8}   tag")
    for name, k, n, src in rows:
        val = "not run" if k is None else f"{k}/{n}"
        _p(f"{name:<26}{val:>8}   {src or '—'}")
    if "1.4" in R:
        _p(f"You (named, old): {R['1.4']['you']}/{R['1.4']['n']}")
    else:
        _p("You (named, old): not scored yet — run 1.4")
    _p("")
    _say("n = 8 is a design demonstration, not a result. Named vs anonymised speaks to recognition / "
         "distraction; fresh speaks to cut-off contamination — and we cannot verify this model's cut-off.")


def run_2A_1():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2A.1 — 'best estimate, no disclaimers'")
    res = ask("2A.1", prompt_hallucination(), temperature=0)
    S["results"]["2A.1"] = res
    _p("MODEL SAYS:")
    for line in (res["text"] or "(no answer)").splitlines():
        _trace("  " + line)


def run_2A_2():
    if not require("2A.1" in S["results"], "run 2A.1 first"):
        return
    _hr("Lab 2 · 2A.2 — the truth beside the model's numbers")
    text = (S["results"]["2A.1"]["text"] or "").replace(",", "")   # '1,234.56' → '1234.56' (no comma decimals)
    _p(f"{'ticker':<7}{'model':>10}{'true close':>12}{'error':>9}")
    any_number = False
    for t in HALLUCINATION_TICKERS:
        true = get_price(t, HALLUCINATION_DATE)["close_usd"]
        m = re.search(rf"{t}\b[^0-9\n]{{0,25}}\$?\s*([0-9]{{1,6}}(?:\.[0-9]{{1,2}})?)", text)
        if m:
            any_number = True
            val = float(m.group(1))
            _p(f"{t:<7}{val:>10.2f}{true:>12.2f}{100*(val/true-1):>8.1f}%")
        else:
            _p(f"{t:<7}{'declined':>10}{true:>12.2f}{'—':>9}")
    _say(f"(true closes from prices.csv, as quoted on {HALLUCINATION_DATE}; no split adjustment needed "
         "for these three)")
    if not any_number:
        _say("The model declined to give numbers — that refusal is the finding. Q3: which control does it replace?")
    else:
        _say("Q3 — Circle every wrong closing price. What went wrong, and which finance control catches it?")


def run_2B():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2B — same question, two job titles")
    out = {}
    for key, label in (("risk", "cautious risk officer"), ("pm", "bullish hedge-fund PM")):
        res = ask(f"2B.{key}", prompt_persona(), temperature=0, persona=PERSONAS[key])
        out[key] = res["text"]
        _p(f"▶ {label}:")
        for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
            _p("  " + line)
        _p("")
    S["results"]["2B"] = out
    _say("Q4 — Write the two recommendations (cautious risk officer / bullish hedge-fund PM). "
         "What changed, and which finance control catches it?")


def run_2C():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2C (stretch) — temperature 0 vs 1, three draws each")
    for temp in (0, 1):
        _p(f"temperature = {temp}")
        for d in (1, 2, 3):
            res = ask(f"2C.t{temp}.d{d}", prompt_temperature(), temperature=temp)
            _trace(f"  draw {d}: {res['text'] or '(no answer)'}")
    _say("At 0 the drafts are nearly (not exactly) identical; at 1 each draw is a new draft.")


def run_3_1():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 3 · 3.1 — analyst mode (four read-only tools)")
    _say(f"TODAY = {TODAY} · tools: {', '.join(READ_ONLY_TOOLS)}")
    res = ask("3.1", prompt_analyst(), temperature=0, tools=READ_ONLY_TOOLS)
    S["results"]["3.1"] = res
    _p("")
    _p(f"tool calls this turn: {len(res['tool_calls'])}")
    _p("FINAL ANSWER:")
    for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
        _p("  " + line)
    _p("")
    _say("Q5 — Tool calls: ___ . Did every number in the final answer come from a tool? YES / NO — "
         "if NO, write the unsourced number.")
    if S["mode"] == MODE_LIVE and not FORCE_CACHED:
        _say("Wait for the 20-second countdown before the next cell — the model is rate-limited, not broken.")
        _countdown(20, "rate-limit breather")
    else:
        _p("(no countdown needed in STUDIO MODE)")


def _snapshot_receipt(cell_id):
    """Remember the student's own 3.2/3.3 run (limit, approval, order book) for the Receipt cell."""
    S["results"]["receipt"] = {"cell": cell_id, "time": _now(), "guard": dict(GUARD),
                               "book": [dict(o) for o in ORDER_BOOK]}


def run_3_2(max_order_gbp, require_human_approval, mandate=None, cell_id="3.2"):
    """Mandate mode. Widgets are read INSIDE the cell that calls this; ORDER_BOOK cleared each run.

    cell_id "3.2" (3.2 and 3.3) is the run the Receipt reports. Any other cell_id (the 🟡 3.4 sceptic
    card, the 🔴 instructor demo) is a SCRATCH run: it prints its own trace and book, then puts the
    student's 3.2/3.3 limit and ORDER BOOK back so the Receipt is unaffected.
    """
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    scratch = cell_id != "3.2"
    saved = (dict(GUARD), [dict(o) for o in ORDER_BOOK])
    GUARD["MAX_ORDER_GBP"] = int(max_order_gbp)
    GUARD["REQUIRE_HUMAN_APPROVAL"] = bool(require_human_approval)
    del ORDER_BOOK[:]
    _p(f"ACTIVE LIMIT: £{GUARD['MAX_ORDER_GBP']:,} · HUMAN APPROVAL: {'ON' if GUARD['REQUIRE_HUMAN_APPROVAL'] else 'OFF'}")
    _say(f"paper mandate £{MANDATE_GBP:,} (GBP notional only; 1 USD ≈ 0.78 GBP is NOT applied) · "
         f"long-only · TODAY {TODAY}")
    if scratch:
        _say(f"(scratch run for cell {cell_id} — your 3.2/3.3 limit and ORDER BOOK are kept for the Receipt)")
    text = mandate or mandate_text()
    # The 🔴 demo cell passes cell_id="instructor": a scratch run that replays the 3.2 cache entry
    # (the cache key is cell_id + prompt hash, and the demo prompt IS the 3.2 prompt).
    res = ask("3.2" if cell_id == "instructor" else cell_id, text, temperature=0, tools=MANDATE_TOOLS)
    S["results"][cell_id] = res
    try:
        if res["source"] == "miss":
            print_order_book()
            return
        orders = [c for c in res["tool_calls"] if c["name"] == "place_order"]
        _p("")
        # An order step with no "status" never reached guard_check — the arguments were malformed.
        # Say so, rather than printing "£0" as though the agent had asked for nothing.
        good = [c for c in orders if isinstance(c.get("result"), dict) and c["result"].get("status")]
        if good:
            o = good[0]["args"]
            _p(f"the agent wants to buy £{float(o.get('gbp_notional', 0)):,.0f} of {str(o.get('ticker', '?')).upper()}")
        elif orders:
            _p("the agent's order was rejected as malformed:")
            for line in textwrap.wrap(_fmt_args(orders[0].get("args") or {}), TRACE_WIDTH - 4):
                _p("  " + line)
        else:
            _p("the agent placed NO order this turn")
        _p("FINAL ANSWER:")
        for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
            _p("  " + line)
        print_order_book()
        refused = [o["result"].get("status", "") for o in good if o["result"].get("status", "").startswith("REFUSED")]
        if refused:
            _say(f"guard that fired: {', '.join(_guards_fired(refused))}"
                 + (" (cached wording may name another control — the ORDER BOOK is the record)"
                    if res["source"] == "cached" else ""))
            _say("Q6 — When refused, the agent (a) tried a smaller order (b) argued (c) complied silently "
                 "(d) other — quote one sentence if there is one.")
        elif good and not scratch:
            _say("3.3 — set MAX_ORDER_GBP BELOW the amount above, tick REQUIRE_HUMAN_APPROVAL, and run this "
                 "cell again.")
    finally:
        if scratch:
            GUARD.clear()
            GUARD.update(saved[0])
            del ORDER_BOOK[:]
            ORDER_BOOK.extend(saved[1])
        else:
            _snapshot_receipt(cell_id)


def _guards_fired(statuses):
    """Which control(s) produced these REFUSED statuses (guard_check wording → control name)."""
    names = []
    for st in statuses:
        name = ("MAX_ORDER_GBP" if "MAX_ORDER_GBP" in st else
                "REQUIRE_HUMAN_APPROVAL" if "REQUIRE_HUMAN_APPROVAL" in st else
                "long-only mandate" if "long-only" in st else "ticker universe")
        if name not in names:
            names.append(name)
    return names


def print_order_book(book=None):
    book = ORDER_BOOK if book is None else book
    _p("ORDER BOOK:")
    if not book:
        _p("  (empty)")
    for o in book:
        px = f" @ ${o['price_usd']:.2f}" if "price_usd" in o else ""
        _trace(f"  {o['time']} {o['side']} {o['ticker']} £{o['gbp_notional']:,.0f}{px} — {o['status']}")


def run_receipt():
    """Receipt: scoreboard + the student's OWN 3.2/3.3 run (not the 3.4 scratch run)."""
    _hr("Receipt — keep this for your reflection")
    R = S["results"]
    if "1.4" in R:
        _p(f"Scoreboard: You {R['1.4']['you']}/{R['1.4']['n']} · Model {R['1.4']['model']}/{R['1.4']['n']} "
           f"(flat band ±{FLAT_BAND*100:.1f}%)")
    elif lock_is_placeholder():
        _say("Scoreboard: not in this runtime (lock lost over the break) — the score you noted down is the record.")
    else:
        _say("Scoreboard: run 1.4 first (1.2 → 1.3 → 1.4) — then re-run this cell.")
    for cid, name in (("1.5b", "anonymised"), ("1.5c", "fresh")):
        if cid in R and R[cid].get("hits") is not None:
            _p(f"  {name}: Model {R[cid]['hits']}/{R[cid]['n']}")
    _p(f"Mode: {S['mode']} · live requests this session: {S['n_requests']}")
    snap = R.get("receipt")
    if snap:
        g = snap["guard"]
        _p(f"Last 3.2/3.3 run at {snap['time']}: limit £{g['MAX_ORDER_GBP']:,} · "
           f"approval {'ON' if g['REQUIRE_HUMAN_APPROVAL'] else 'OFF'}")
        print_order_book(snap["book"])
    else:
        _p("No 3.2/3.3 run in this runtime yet — run 3.2 (and 3.3) first.")


def run_4_1():
    _hr(f"Lab 4 · the committee is convened on {COMMITTEE_TICKER}")
    md = S["committee_md"]
    if not md:
        md = ("# PLACEHOLDER — committee transcript not generated yet\n\n"
              "Run build/build_cache.py with a key (or build/make_sample_cache.py) to create cache/committee.md.")
    ip = None
    try:
        import IPython
        ip = IPython.get_ipython()
    except Exception:
        pass
    if ip is not None:
        from IPython.display import Markdown, display
        display(Markdown(md))
    else:
        _p(md)
    return md


def committee_context():
    """The Lab-3 tool outputs every committee role shares (real tool functions)."""
    t = COMMITTEE_TICKER
    d = pd.to_datetime(TODAY)
    ctx = {
        "price": get_price(t, TODAY),
        "ret_1m": get_return(t, (d - pd.Timedelta(days=30)).strftime("%Y-%m-%d"), TODAY),
        "ret_3m": get_return(t, (d - pd.Timedelta(days=91)).strftime("%Y-%m-%d"), TODAY),
        "ret_12m": get_return(t, (d - pd.Timedelta(days=365)).strftime("%Y-%m-%d"), TODAY),
        "vol_30d": get_volatility(t, 30),
        "headlines": get_headlines(t),
    }
    return ctx


COMMITTEE_ROLES = [
    ("bull", "Bull analyst", "Argue the case FOR buying {company} ({ticker}) now, in at most 120 words, using only the data below."),
    ("bear", "Bear analyst", "Argue the case AGAINST buying {company} ({ticker}) now, in at most 120 words, using only the data below."),
    ("risk", "Risk manager", "Given the bull and bear cases and the data below, state the two biggest risks and a position limit you would impose, in at most 120 words."),
    ("trader", "Trader", "Given the bull, bear and risk-manager notes and the data below, propose ONE paper order for {company} ({ticker}) within a £1,000,000 long-only mandate: ticker, GBP size, and a two-sentence rationale. State the order as 'ORDER: BUY £<size> of {ticker}'."),
]


def committee_prompts(ctx):
    """Four sequential prompts sharing the tool outputs; later roles see earlier notes."""
    data = json.dumps(ctx, default=str, indent=0)
    prompts = {}
    for key, role, instr in COMMITTEE_ROLES:
        p = f"Today is {TODAY}. You are the {role} on a paper-trading investment committee. "
        p += instr.format(company=COMPANY[COMMITTEE_TICKER], ticker=COMMITTEE_TICKER)
        p += "\n\nDATA (from the Lab-3 tools):\n" + data
        prompts[key] = p
    return prompts


def committee_prompt(key, prompts, notes):
    """Later roles see the earlier notes (the chain is the same in the notebook and the builders)."""
    p = prompts[key]
    if key == "risk":
        p += f"\n\nBULL:\n{notes.get('bull', '')}\n\nBEAR:\n{notes.get('bear', '')}"
    if key == "trader":
        p += (f"\n\nBULL:\n{notes.get('bull', '')}\n\nBEAR:\n{notes.get('bear', '')}"
              f"\n\nRISK MANAGER:\n{notes.get('risk', '')}")
    return p


def run_committee_live(cell_prefix="4.2"):
    """🟡 Regenerate the transcript live (4 calls). Also used by build_cache.py."""
    ctx = committee_context()
    prompts = committee_prompts(ctx)
    notes = {}
    for key, role, _ in COMMITTEE_ROLES:
        p = committee_prompt(key, prompts, notes)
        res = ask(f"{cell_prefix}.{key}", p, temperature=0)
        notes[key] = res["text"] or "(no answer)"
        _p(f"▶ {role} ({res['source']}):")
        for line in textwrap.wrap(notes[key], TRACE_WIDTH - 2):
            _p("  " + line)
        _p("")
    return ctx, notes


def committee_evidence_rows(ctx):
    """The shared evidence as (row-number, tool call, returned value) triples.

    One row per Lab-3 tool output, each tagged with the tool call (brief §5) that
    produced it, so a reader can check any number in the transcript against the call
    that returned it. This is what the Analyst role is asked to do in committee.
    """
    t = COMMITTEE_TICKER
    rows = []
    p, h = ctx["price"], ctx["headlines"]
    rows.append((f'get_price("{t}", "{p["date"]}")', f'close_usd = {p["close_usd"]:.2f} USD (unadjusted close)'))
    for key, label in (("ret_1m", "one month"), ("ret_3m", "three months"), ("ret_12m", "twelve months")):
        r = ctx[key]
        rows.append((f'get_return("{t}", "{r["start"]}", "{r["end"]}")',
                     f'return_pct = {r["return_pct"]:+.2f} % ({label}, adjusted closes)'))
    v = ctx["vol_30d"]
    rows.append((f'get_volatility("{t}", {v["window_days"]})',
                 f'annualised_vol_pct = {v["annualised_vol_pct"]:.2f} % (as of {v["as_of"]})'))
    rows.append((f'get_headlines("{t}")',
                 f'{len(h["headlines"])} headline(s), date + text only, dated <= TODAY'))
    for i, item in enumerate(h["headlines"], 1):
        rows.append((f'  \u2514 headline {i} (from get_headlines)', f'{item["date"]} \u2014 {item["text"]}'))
    rows.append(("(constant) TODAY", f"{TODAY} \u2014 no tool may return anything dated after this"))
    rows.append(("(constant) mandate", f"\u00a3{MANDATE_GBP:,.0f} paper notional, long-only, exactly one place_order call"))
    return [(i + 1, a, b) for i, (a, b) in enumerate(rows)]


def committee_evidence_table(ctx):
    rows = committee_evidence_rows(ctx)
    w = max(len(a) for _, a, _ in rows)
    out = [f"{'#':>2}  {'TOOL CALL (brief §5)'.ljust(w)}  RETURNED",
           f"{'--':>2}  {'-' * w}  {'-' * 34}"]
    for n, a, b in rows:
        out.append(f"{n:>2}  {a.ljust(w)}  {b}")
    return "\n".join(out)


def committee_markdown(ctx, notes, generated_by, sample, appendix=None):
    """Transcript: evidence table (each row tagged to its tool call), four notes, appendix.

    `appendix` is an optional dict with keys "numbers" and "unsourced": lists of strings
    separating the numbers that came from a tool row above from the claims that did not.
    Both builders (build_cache.py, make_sample_cache.py) call this one function, so the
    long and short forms of the transcript cannot diverge.
    """
    md = f"# Investment committee transcript — {COMPANY[COMMITTEE_TICKER]} ({COMMITTEE_TICKER})\n\n"
    md += f"*Generated {datetime.now().strftime('%Y-%m-%d')} by {generated_by}; data as of {TODAY}.*  \n"
    if sample:
        md += "**SAMPLE — stand-in text written by Claude (Anthropic), not model output; regenerate with build/build_cache.py.**\n\n"
    md += ("## Shared evidence (Lab-3 tool outputs)\n\n"
           "Every agent below saw exactly this, and nothing else. Each row is tagged with "
           "the tool call that produced it.\n\n```\n" + committee_evidence_table(ctx) + "\n```\n\n")
    for key, role, _ in COMMITTEE_ROLES:
        md += f"## {role}\n\n{notes.get(key, '')}\n\n"
    if appendix:
        md += "## Appendix — check the transcript against the table\n\n"
        md += "### Numbers used (each one traces to a row above)\n\n"
        for line in appendix.get("numbers", []):
            md += f"- {line}\n"
        md += "\n### Claims made without tool support (nothing in the table returns these)\n\n"
        for line in appendix.get("unsourced", []):
            md += f"- {line}\n"
        md += "\n*For the Analyst role: the second list is the answer, and the reason the trace is printed.*\n"
    return md


def reset_all():
    """Reset cell: clears the lock, the order book, results and counters (instructor / demo use)."""
    global PREDICTIONS_LOCKED
    PREDICTIONS_LOCKED = None
    globals()["PREDICTIONS_LOCKED"] = None
    del ORDER_BOOK[:]
    S["results"] = {}
    S["n_requests"] = 0
    GUARD.update(MAX_ORDER_GBP=MANDATE_GBP, REQUIRE_HUMAN_APPROVAL=False)
    _say("reset: lock cleared, ORDER BOOK emptied, results and request counter zeroed.")


def run_instructor_unprompted(n=5):
    """🔴 Would it trade unprompted? n re-runs of 3.2 without the mandatory-order line."""
    if not require(locked(), "run 1.2 first (any dropdown values) — the gate applies to instructors too"):
        return
    saved = (dict(GUARD), [dict(o) for o in ORDER_BOOK])   # scratch runs: the Receipt is untouched
    try:
        for i in range(1, n + 1):
            _hr(f"🔴 unprompted run {i}/{n}")
            GUARD.update(MAX_ORDER_GBP=MANDATE_GBP, REQUIRE_HUMAN_APPROVAL=False)
            del ORDER_BOOK[:]
            res = ask(f"X.unprompted.d{i}", mandate_text(mandatory=False), temperature=1, tools=MANDATE_TOOLS)
            orders = [c for c in res["tool_calls"] if c["name"] == "place_order"]
            _p(f"orders placed: {len(orders)} · source: {res['source']}")
            for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
                _p("  " + line)
    finally:
        GUARD.clear()
        GUARD.update(saved[0])
        del ORDER_BOOK[:]
        ORDER_BOOK.extend(saved[1])

# ---- go -------------------------------------------------------------------
bootstrap()
````

## Glossary 🟢 REQUIRED — *read, nothing to run* (finance terms for what you are about to do)

| Term | In finance language |
|---|---|
| **LLM** (large language model) | A very well-read intern who has read every annual report and news article up to a cut-off date, has never traded, and answers in the same confident tone whether right or wrong. |
| **Prompt** | The instruction memo you hand the intern. |
| **Temperature** | Noise added to the answer: 0 = *nearly* the same memo every time; 1 = a different draft each time. A simulation seed, not a correctness dial. |
| **Knowledge cut-off** | The date the intern stopped reading. We cannot verify it for any model. |
| **Context window** | The intern's working memory in one sitting. |
| **Hallucination** | Confidently wrong rather than "I don't know" — a spuriously significant coefficient that can talk. |
| **Structured output** | Making the intern fill in a form (JSON) instead of writing prose, so a machine can score it. |
| **Tool / function call** | You hand the intern a list of things it may ask you to do (`get_price`…); instead of answering it says "please run `get_price` for NVDA"; your laptop runs it and pastes the number back. |
| **Agent** | That intern, now given a Bloomberg terminal, a budget and permission to act. The brain did not change; the tools, the loop and the permissions did. |
| **Trace** | The printed `🔧 TOOL CALL …` lines — an audit log, *if* you build one. |
| **Look-ahead bias** | Testing a "prediction" on data the model had already read. The backtest sin. |
| **Cache** (STUDIO MODE) | The instructor's pre-run answers to the same prompts, replayed through the same code — so refusals, orders and numbers are still computed here. |

````python
#@title 🔁 Disconnected? Run this first (Lab 2) { display-mode: "form" }
FORCE_CACHED = False  #@param {type:"boolean"}
# Room-wide STUDIO MODE: tick the box above (or the instructor projects "FORCE_CACHED = True").
REPO_RAW = "https://raw.githubusercontent.com/thezeming/mgrcm0065-week3/main"   # class-repo base URL, baked in from w3_core.REPO_RAW_URL; empty = no fetch
_RECONNECT = True
# pinned SDK (guarded: warm runtimes skip the install)
try:
    from google import genai as _probe  # noqa
except ImportError:
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "google-genai==1.47.0"])
EMBEDDED_B64 = ""
# ---- shared core (verbatim copy of build/w3_core.py) ------------------------
# ============================================================================
#  MGRCM0065 Week 3 · "AI for Finance" — shared core
#  This file is (a) embedded verbatim into cell 0 of the student notebook and
#  (b) imported by build/build_cache.py, build/make_sample_cache.py and the
#  tests, so the notebook and the cache builder run IDENTICAL code.
#  Only the standard library + pandas + numpy + google-genai are imported.
# ============================================================================
import base64, gzip, hashlib, io, json, os, re, sys, textwrap, time, math
from datetime import datetime

import numpy as np
import pandas as pd

# ----------------------------------------------------------------------------
# Constants (brief §1). MODEL is a placeholder the instructor re-checks on
# build day and on 6 Oct at aistudio.google.com/rate-limit.
# ----------------------------------------------------------------------------
MODEL = "gemini-3.5-flash-lite"              # the configured choice; may 404 if Google withdraws it
# Chosen on 17 Sep 2026 from the instructor's own account catalogue (41 models). It is the
# newest non-preview "flash-lite" the key can see, which is the cheapest/fastest tier and so
# the one with the largest free daily allowance — what matters with ~100 students. The
# previous value, gemini-2.5-flash-lite, returned 404 from an older SDK on 17 Sep 2026.
# pick_model() independently selects this same model from that catalogue.
MODEL_FALLBACK = True                        # on a 404, ask the key what it CAN use and switch (loudly)
MODEL_IN_USE = MODEL                         # the model actually talking to us; build_cache records THIS
SDK_VERSION = "2.24.0"                       # google-genai version this code was written against
SDK_MIN_VERSION = "1.20.0"                   # oldest google-genai whose API surface we verified by hand
MODEL_EXCLUDE = ("pro", "ultra")             # never auto-pick an expensive tier
GROQ_MODEL = "openai/gpt-oss-20b"
TICKERS = ["AAPL", "MSFT", "NVDA", "TSLA", "AMZN", "META", "JPM", "BA"]
COMMITTEE_TICKER = "NVDA"
FLAT_BAND = 0.005                            # ±0.5 % is FLAT
PACE_SECONDS = 7                             # client-side spacing between live requests
TOOL_CAP = 3                                 # READ-ONLY tool calls per agent turn
ORDER_CAP = 2                                # place_order calls per turn (refused + retry smaller)
# The two budgets are deliberately separate. A single shared cap looks tidier but starves the
# mandate: the model researches all three tickers in ONE parallel batch, which spends three
# calls, and place_order then comes back "CAP REACHED" with an empty order book — observed
# against the live API on 17 Sep 2026. Lab 3 exists to show an order being filled and then
# refused, so the order must never be the thing the budget runs out on. ORDER_CAP is 2 because
# the 3.3 demonstration needs a refusal followed by a smaller retry, which is two calls.
MANDATE_GBP = 1_000_000                      # paper mandate, GBP notional only
# Raw-GitHub base URL of the class repo, e.g.
#   "https://raw.githubusercontent.com/<user>/<repo>/main"
# make_notebook.py bakes this into cell 0. Empty = the notebook never phones home and always uses
# its own embedded copy. Set it once here, then `python build/refresh.py --publish`.
REPO_RAW_URL = "https://raw.githubusercontent.com/thezeming/mgrcm0065-week3/main"
TRACE_WIDTH = 72
HALLUCINATION_DATE = "2023-06-30"            # inside any plausible training window; no splits since
HALLUCINATION_TICKERS = ["MSFT", "META", "JPM"]
COMPANY = {"AAPL": "Apple", "MSFT": "Microsoft", "NVDA": "Nvidia", "TSLA": "Tesla",
           "AMZN": "Amazon", "META": "Meta", "JPM": "JPMorgan", "BA": "Boeing"}

MODE_LIVE, MODE_STUDIO, MODE_ERROR = "LIVE MODE", "STUDIO MODE", "ERROR — hand up"
STUDIO_SUBLINE = ("every number, order and refusal is computed on your laptop; "
                  "the model's wording is the instructor's pre-run")
CACHED_BANNER = "[cached output — instructor's pre-run]"
SAMPLE_BANNER = "SAMPLE CACHE — regenerate with build/build_cache.py before class"

# Anonymised twins of the Lab-1 headlines (names, tickers, products and dates
# removed; written by the notebook builder for the 1.5b design, not by a model).
ANON_HEADLINES = {
    "H01": "Company X warns of flagship-product shipment delays after a Covid lockdown at a supplier's plant",
    "H02": "Company X announces multibillion-dollar investment in an AI start-up",
    "H03": "Company X guidance beats estimates on generative-AI demand",
    "H04": "Company X to cut over 18,000 jobs",
    "H05": "Company X posts first-ever drop in daily active users, issues weak guidance",
    "H06": "Company X acquires a failed regional bank after regulators seize it",
    "H07": "Regulator orders inspections, grounds Company X's aircraft type after an engine failure",
    "H08": "Company X confirmed to join a major stock index next month",
}

# ----------------------------------------------------------------------------
# Global state. Everything the cells share lives in S so that the 🔁 cells can
# rebuild it and the build scripts can inspect it.
# ----------------------------------------------------------------------------
_prev_S = globals().get("S") if isinstance(globals().get("S"), dict) else {}
S = {
    "mode": None, "provider": None, "client": None, "key_status": "unchecked",
    "n_requests": 0, "last_request": 0.0, "cache": {}, "cache_meta": {},
    "prices": None, "headlines": None, "fresh": None, "committee_md": None,
    "version": {}, "data_source": "none", "results": {}, "recording": None,
    "in_colab": False, "sdk_ok": False, "grounding_seen": False,
    "model": MODEL, "model_switched": False, "sdk_version": None,
}
# Re-running cell 0 or a 🔁 cell must NOT wipe the lock, the results or the order book.
S["results"] = dict(_prev_S.get("results", {}))
S["n_requests"] = int(_prev_S.get("n_requests", 0))
ORDER_BOOK = list(globals().get("ORDER_BOOK") or [])
GUARD = dict(globals().get("GUARD") or {"MAX_ORDER_GBP": MANDATE_GBP, "REQUIRE_HUMAN_APPROVAL": False})
PREDICTIONS_LOCKED = globals().get("PREDICTIONS_LOCKED", None)   # None or (timestamp_str, {id: call}, kind)
_BOOTSTRAPPED = bool(globals().get("_BOOTSTRAPPED", False))      # True once cell 0 / a 🔁 cell has run in THIS kernel
#   kind ∈ {"locked", "restored"}: "restored" is the placeholder a 🔁 cell sets when the runtime lost the
#   real lock — it opens Labs 2–4 but never scores, and 1.2 may overwrite it.
TODAY = None                   # last trading date in prices.csv, set by load_data()

# Values the notebook's cell 0 may define before this code runs.
FORCE_CACHED = bool(globals().get("FORCE_CACHED", False)) or os.environ.get("W3_FORCE_CACHED", "") == "1"
REPO_RAW = str(globals().get("REPO_RAW", "") or "")
EMBEDDED_B64 = globals().get("EMBEDDED_B64", None)
_RECONNECT = bool(globals().get("_RECONNECT", False))
QUIET = bool(globals().get("W3_QUIET", False))

genai = None
gtypes = None
gerrors = None


def _p(*a, **k):
    """print() that the build scripts can silence."""
    if not QUIET:
        print(*a, **k)


def _hr(title=""):
    line = "─" * TRACE_WIDTH
    _p(line if not title else f"── {title} " + "─" * max(0, TRACE_WIDTH - len(title) - 4))


def _countdown(seconds, label):
    """Every wait prints a countdown so nothing looks like a hang."""
    seconds = int(math.ceil(seconds))
    if seconds <= 0:
        return
    _p(f"{label}: ", end="", flush=True)
    for i in range(seconds, 0, -1):
        _p(f"{i}…", end=" ", flush=True)
        time.sleep(1)
    _p("go", flush=True)


def _trace(line):
    """Print a trace line wrapped so that no printed line exceeds 72 chars."""
    for chunk in textwrap.wrap(line, TRACE_WIDTH, subsequent_indent="      ", break_on_hyphens=False) or [""]:
        _p(chunk)


def _say(text, indent=""):
    """Print prose wrapped at 72 characters (plain lines, no hanging indent)."""
    for chunk in textwrap.wrap(text, TRACE_WIDTH, initial_indent=indent, subsequent_indent=indent,
                               break_on_hyphens=False) or [""]:
        _p(chunk)


def _sha(text):
    return hashlib.sha256(text.encode("utf-8")).hexdigest()


def _now():
    return datetime.now().strftime("%H:%M:%S")


# ----------------------------------------------------------------------------
# Data loading: embedded blob first, disk second (build scripts), GitHub third
# ----------------------------------------------------------------------------
def _materials_dir():
    here = os.path.dirname(os.path.abspath(globals().get("__file__", "./build/w3_core.py")))
    return os.path.abspath(os.path.join(here, ".."))


def _decode_blob(b64):
    return json.loads(gzip.decompress(base64.b64decode(b64)).decode("utf-8"))


def _read_disk_blob():
    root = _materials_dir()
    blob = {}
    for key, rel in [("prices_csv", "data/prices.csv"), ("headlines_csv", "data/headlines.csv"),
                     ("headlines_fresh_csv", "data/headlines_fresh.csv"),
                     ("responses_json", "cache/responses.json"),
                     ("committee_md", "cache/committee.md"), ("version_json", "VERSION.json")]:
        path = os.path.join(root, rel)
        blob[key] = open(path, encoding="utf-8").read() if os.path.exists(path) else None
    return blob


def _install_blob(blob, source):
    global TODAY
    if not blob.get("prices_csv") or not blob.get("headlines_csv"):
        raise ValueError("prices.csv / headlines.csv missing from the data bundle")
    prices = pd.read_csv(io.StringIO(blob["prices_csv"]))
    prices["date"] = pd.to_datetime(prices["date"]).dt.strftime("%Y-%m-%d")
    prices = prices.sort_values(["ticker", "date"]).reset_index(drop=True)
    heads = pd.read_csv(io.StringIO(blob["headlines_csv"]))
    fresh = pd.read_csv(io.StringIO(blob["headlines_fresh_csv"])) if blob.get("headlines_fresh_csv") else None
    for df in (heads, fresh):
        if df is not None:
            for c in ("date", "event_date"):
                df[c] = pd.to_datetime(df[c]).dt.strftime("%Y-%m-%d")
    cache = json.loads(blob["responses_json"]) if blob.get("responses_json") else {"_meta": {"sample": True, "empty": True}, "entries": {}}
    S.update(prices=prices, headlines=heads, fresh=fresh, data_source=source,
             cache=cache.get("entries", {}), cache_meta=cache.get("_meta", {}),
             committee_md=blob.get("committee_md"),
             version=json.loads(blob["version_json"]) if blob.get("version_json") else {})
    TODAY = str(prices["date"].max())
    globals()["TODAY"] = TODAY


def load_data():
    """Embedded copy first (always sufficient for Labs 1–4), disk copy for the build scripts."""
    if EMBEDDED_B64:
        _install_blob(_decode_blob(EMBEDDED_B64), "embedded copy inside this notebook")
    else:
        _install_blob(_read_disk_blob(), "materials/ folder on disk")


def _http_get(url, timeout=10):
    import urllib.request
    with urllib.request.urlopen(url, timeout=timeout) as r:
        return r.read().decode("utf-8")


REPO_FILES = [("prices_csv", "data/prices.csv"),
              ("headlines_csv", "data/headlines.csv"),
              ("headlines_fresh_csv", "data/headlines_fresh.csv"),
              ("responses_json", "cache/responses.json"),
              ("committee_md", "cache/committee.md")]


def bundle_last_trading_day(blob):
    prices = pd.read_csv(io.StringIO(blob["prices_csv"]))
    return str(pd.to_datetime(prices["date"]).dt.strftime("%Y-%m-%d").max())


def bundle_is_coherent(blob):
    """Do the prices and the recorded answers in this bundle belong to each other?

    A cache entry is keyed on cell_id + sha256(prompt), and the Lab 3 / Lab 4 prompts embed
    TODAY, which is the last trading day in prices.csv. So a bundle whose answers were recorded
    against a DIFFERENT day orphans every Lab 3 and Lab 4 entry: the cells still run, but a
    student without a working key gets "no cached answer" in exactly the two labs that carry the
    teaching — and only once their quota runs out, so it never shows up in testing.

    Checked before anything is installed. Returns (ok, reason).
    """
    try:
        last = bundle_last_trading_day(blob)
    except Exception as e:
        return False, f"prices.csv could not be read ({type(e).__name__})"
    try:
        meta = (json.loads(blob["responses_json"]) or {}).get("_meta", {})
    except Exception as e:
        return False, f"responses.json could not be read ({type(e).__name__})"
    recorded = str(meta.get("today_constant", "") or "")
    if not recorded:
        return False, "the recorded answers do not say which trading day they were built against"
    if recorded != last:
        return False, (f"prices run to {last} but the answers were recorded for {recorded} — "
                       "Lab 3 and Lab 4 would fall through to 'no cached answer'")
    return True, ""


def _warn_if_notebook_stale(ver, mine):
    if ver.get("notebook") != mine.get("notebook"):
        _say(f"⚠️  VERSION: this notebook is v{mine.get('notebook')}; current is v{ver.get('notebook')} — "
             "open the fresh notebook link and save a new copy (the code itself cannot self-update).")


def fetch_repo_updates():
    """Refresh data/ and cache/ from the class repo (skipped when REPO_RAW is empty).

    ALL OR NOTHING, deliberately. prices.csv and cache/responses.json are a matched pair — see
    bundle_is_coherent() — so a half-applied refresh is worse than no refresh at all. One failed
    download, or one mismatched pair, abandons the entire attempt and leaves the embedded copy
    untouched. The embedded copy is always internally consistent, because make_notebook.py built
    it from one snapshot.
    """
    if not REPO_RAW:
        _p("GitHub refresh: skipped (REPO_RAW is empty) — using the embedded copy.")
        return
    if os.environ.get("W3_NO_FETCH") == "1":
        # make_notebook.py captures the Expected-output boxes by EXECUTING each cell. Without this
        # the build would phone home four times per run, and its output would depend on the network
        # — which breaks the byte-identical-rebuild test.
        _p("GitHub refresh: (runs here — OK / already current / not reachable)")
        return
    base = REPO_RAW.rstrip("/")
    bust = f"?v={int(time.time())}"
    mine = dict(S["version"])
    try:
        ver = json.loads(_http_get(f"{base}/VERSION.json{bust}"))
    except Exception as e:
        _say(f"GitHub refresh: not reachable ({type(e).__name__}) — embedded copy is fine for today.")
        return
    if ver.get("data") == mine.get("data"):
        _say(f"GitHub refresh: already current (data v{mine.get('data')}) — nothing to download.")
        _warn_if_notebook_stale(ver, mine)
        return
    fetched = {}
    for key, rel in REPO_FILES:
        try:
            fetched[key] = _http_get(f"{base}/{rel}{bust}")
        except Exception as e:
            _say(f"GitHub refresh: {rel} did not download ({type(e).__name__}) — abandoning the whole "
                 "refresh rather than mixing new prices with older answers; embedded copy stands.")
            _warn_if_notebook_stale(ver, mine)
            return
    blob = dict(_read_disk_blob()) if not EMBEDDED_B64 else _decode_blob(EMBEDDED_B64)
    blob.update(fetched)
    blob["version_json"] = json.dumps(ver)
    ok, why = bundle_is_coherent(blob)
    if not ok:
        _say(f"GitHub refresh: REFUSED — {why}. Embedded copy stands.")
        _warn_if_notebook_stale(ver, mine)
        return
    try:
        _install_blob(blob, f"GitHub ({base})")
    except Exception as e:
        _say(f"GitHub refresh: bundle rejected ({type(e).__name__}) — embedded copy stands.")
        _warn_if_notebook_stale(ver, mine)
        return
    _say(f"GitHub refresh: OK — data v{mine.get('data')} \u2192 v{ver.get('data')}; prices to {TODAY}, "
         "answers recorded for the same day.")
    _warn_if_notebook_stale(ver, mine)


# ----------------------------------------------------------------------------
# SDK install / import (pinned; guarded so warm runtimes skip it)
# ----------------------------------------------------------------------------
def ensure_sdk():
    global genai, gtypes, gerrors
    try:
        from google import genai as _g
    except ImportError:
        _p(f"Installing google-genai=={SDK_VERSION} (about 20 seconds)…")
        os.system(f"{sys.executable} -m pip install -q google-genai=={SDK_VERSION}")
        try:
            from google import genai as _g
        except ImportError:
            _say("Runtime → Restart session, then run this cell again — normal, 20 seconds.")
            S["sdk_ok"] = False
            return False
    from google.genai import types as _t, errors as _e
    genai, gtypes, gerrors = _g, _t, _e
    ver = getattr(_g, "__version__", "?")
    S["sdk_version"] = ver
    if not sdk_version_ok(ver):
        S["sdk_ok"] = False
        _p(f"SDK: google-genai {ver} — TOO OLD")
        _say(f"This notebook needs google-genai {SDK_MIN_VERSION} or newer (it was written "
             f"against {SDK_VERSION}); you have {ver}, whose API is different enough to fail "
             "in the middle of a cell rather than here. Fix it with one line:")
        _p(f"  {sys.executable} -m pip install -U 'google-genai>={SDK_MIN_VERSION}'")
        _say("If pip says no matching distribution, the cause is your Python, not your "
             "network: modern google-genai needs Python 3.10 or newer, and 'python3' on a Mac "
             "is often the system 3.9. Check with  python3 -V  and rebuild the environment "
             "with an explicit modern interpreter, e.g. /opt/homebrew/bin/python3.12.")
        _say("STUDIO MODE meanwhile — every number, order and refusal still runs.")
        return False
    S["sdk_ok"] = True
    _p(f"SDK: google-genai {ver}")
    if ver != SDK_VERSION:
        _say(f"(written against {SDK_VERSION}; {SDK_MIN_VERSION} and newer are supported — "
             "the calls this notebook makes are identical on both.)")
    return True


def _ver_tuple(v):
    """('1.47.0' → (1, 47, 0)); anything unparseable sorts as (0,) so it fails the floor check."""
    parts = []
    for chunk in str(v).split(".")[:3]:
        digits = "".join(c for c in chunk if c.isdigit())
        if digits == "":
            break
        parts.append(int(digits))
    return tuple(parts) or (0,)


def sdk_version_ok(ver):
    """True when the installed google-genai is at least SDK_MIN_VERSION."""
    return _ver_tuple(ver) >= _ver_tuple(SDK_MIN_VERSION)


# ----------------------------------------------------------------------------
# Key discovery and the three-attempt verification
# ----------------------------------------------------------------------------
def find_key():
    """Returns (key, provider, status). Distinguishes no key / NotebookAccessError."""
    try:
        from google.colab import userdata  # noqa
        S["in_colab"] = True
    except ImportError:
        S["in_colab"] = False
        userdata = None
    for name, provider in (("GEMINI_API_KEY", "gemini"), ("GROQ_API_KEY", "groq")):
        if userdata is not None:
            try:
                val = userdata.get(name)
                if val:
                    return val.strip(), provider, "colab-secret"
            except Exception as e:  # SecretNotFoundError / NotebookAccessError
                if "NotebookAccess" in type(e).__name__:
                    return None, provider, "notebook-access"
        val = os.environ.get(name, "").strip()
        if val:
            return val, provider, "env"
    return None, None, "none"


# Remedies, in the order _classify_error tests for them. Each one has its own branch and its own
# sentence on screen, because telling someone to check their Wi-Fi when the model name is wrong
# wastes the whole room's time.
#   rate-limit      429 / RESOURCE_EXHAUSTED — the key works, the free quota is busy
#   model-not-found 404 / NOT_FOUND, or a 403 whose message names models/… — the MODEL name is
#                   not available to this key (deterministic: never retried, because waiting
#                   cannot make a withdrawn model come back)
#   bad-key         400/401/403 naming the key or a permission — the key itself is wrong
#   server          5xx at the provider
#   network         a transport-level failure: DNS, TCP, TLS, timeout
#   other           anything we have not met before; never described as a network problem
def _classify_error(e):
    """Map any exception from the SDK to exactly one remedy. Specific codes first; the
    name-sniffing network test last, so 'network' only ever means an actual network fault.

    Works for both SDKs: google-genai puts an int in .code, Groq/OpenAI-style SDKs put a
    string error code there and the HTTP status in .status_code, so we normalise first."""
    name = type(e).__name__
    code = getattr(e, "code", None)
    if not isinstance(code, int) or isinstance(code, bool):
        try:
            code = int(code)
        except (TypeError, ValueError):
            code = None
    if code is None:
        sc = getattr(e, "status_code", None)
        code = sc if isinstance(sc, int) and not isinstance(sc, bool) else None
    status = str(getattr(e, "status", "") or "")
    msg = str(e)
    up = msg.upper()
    if code == 429 or "RESOURCE_EXHAUSTED" in status or "429" in msg[:12] or "RateLimit" in name:
        return "rate-limit"
    if code == 404 or "NOT_FOUND" in status or "NOT_FOUND" in up[:200] or "NotFound" in name:
        return "model-not-found"
    # A 403 PERMISSION_DENIED that names a model ("Permission denied on resource
    # models/gemini-2.5-flash-lite") is the model being withdrawn from this key, not a bad
    # key — Google returns either code for it. Tested before the key branch on purpose.
    if code in (403, 404) and "MODELS/" in up and "API_KEY" not in up:
        return "model-not-found"
    if code in (400, 401, 403) and ("API_KEY" in up or "PERMISSION" in up or code in (401, 403)):
        return "bad-key"
    if code is None and any(w in name for w in ("Authentication", "PermissionDenied")):
        return "bad-key"
    if code is not None and 500 <= int(code) < 600:
        return "server"
    if any(w in name for w in ("Timeout", "Connect", "Network", "Remote", "Transport", "OSError", "URLError")):
        return "network"
    return "other"


def _key_console(provider):
    """Where to re-copy a rejected key from — the provider's own console, never Google's
    for a Groq key."""
    return "console.groq.com/keys" if provider == "groq" else "aistudio.google.com/apikey"


def model_from_error(e, default=None):
    """Pull the model name out of a 404 message ('models/gemini-2.5-flash-lite is not found')."""
    m = re.search(r"models/([A-Za-z0-9._\-]+)", str(e))
    # ".rstrip('.')" because a 403 states it as a sentence: "…on resource models/x-y-z."
    return m.group(1).rstrip(".") if m else (default if default is not None else MODEL_IN_USE)


# ----------------------------------------------------------------------------
# Model discovery. Google publishes no static free-tier table (design-notes/
# research_tooling.md), so a 404 on the configured MODEL is an expected failure,
# not an exotic one. list_text_models() asks the key what it can actually use;
# pick_model() chooses between them by a rule that is fixed and written down.
# ----------------------------------------------------------------------------
def list_text_models(client=None):
    """Every model this API key can call with generateContent, as plain names
    ('gemini-2.5-flash'). A listing entry that advertises no supported actions at all is
    skipped rather than kept. Raises the SDK's own exception so the caller can classify it."""
    client = client or S["client"]
    out = []
    for m in client.models.list(config={"query_base": True}):
        actions = list(getattr(m, "supported_actions", None) or [])
        if not any(a.lower() == "generatecontent" for a in actions):
            continue   # a model that advertises nothing is treated as unusable, not as a candidate
        name = (getattr(m, "name", "") or "").split("/")[-1]
        if name:
            out.append(name)
    return sorted(set(out))


def _version_tuple(n):
    """The version that belongs to the family name, as a fixed-width tuple (largest first).

    Only the dotted or whole numbers immediately after the first word count, and we stop at
    the first token that is not a number, so a date suffix ('-preview-11-2025') or a build
    number ('-001') is never read as a version. Fixed width so the comparison is positional:
    without the padding, ('gemini-3',) would sort ahead of ('gemini-3.5',) simply because the
    shorter tuple is a prefix of the longer one."""
    nums = []
    for tok in n.split("-")[1:]:
        if re.fullmatch(r"\d+(\.\d+)*", tok):
            nums.extend(int(x) for x in tok.split("."))
        else:
            break
    return tuple(-x for x in (nums + [0, 0, 0])[:3])


def _model_rank(name):
    """Sort key for pick_model. Smaller is better, and it is entirely determined by the
    name, so the same list always yields the same choice:
      1. family      flash-lite (0) before flash (1) before anything else (2);
      2. version     the numbers in the name, largest first, compared position by position
                     so 3.5 beats a bare 3 (3.6 beats 2.5);
      3. maturity    a plain name before a preview/experimental/dated one;
      4. simplicity  the shorter name, then alphabetical — a pure tie-break."""
    n = name.lower()
    family = 0 if "flash-lite" in n else (1 if "flash" in n else 2)
    nums = _version_tuple(n)
    preview = 1 if any(w in n for w in ("preview", "exp", "-rc", "latest")) else 0
    return (family, nums, preview, len(name), name)


def pick_model(names):
    """The best model to fall back to, or None. Never picks a pro/ultra tier: an expensive
    model chosen silently is a bill, not a rescue."""
    usable = [n for n in names if not any(w in n.lower() for w in MODEL_EXCLUDE)]
    return sorted(usable, key=_model_rank)[0] if usable else None


def use_model(name, reason=""):
    """Switch the model every later request uses, and say so in one line on screen."""
    global MODEL_IN_USE
    MODEL_IN_USE = name
    S["model"] = name
    S["model_switched"] = (name != MODEL)
    if S["model_switched"]:
        _say(f"MODEL {MODEL} is unavailable; using {name} instead{reason}")
    return name


def resolve_model(client=None, configured=None):
    """Called when the configured model 404s. Asks the key what it can use, announces the
    switch and returns the new name, or None if nothing usable came back."""
    configured = configured or MODEL
    if not MODEL_FALLBACK:
        _say(f"MODEL_FALLBACK is off, so {configured} was not replaced automatically.")
        return None
    try:
        names = list_text_models(client)
    except Exception as e:
        _say(f"could not ask the provider which models this key can use "
             f"({type(e).__name__}: {str(e)[:60]})")
        return None
    if not names:
        _say("the provider listed no text-generation models for this key.")
        return None
    pick = pick_model(names)
    if pick is None:
        _say("the only models this key can use are pro/ultra tiers; not switching to one "
             "automatically. Set MODEL in build/w3_core.py by hand if you want one.")
        return None
    return use_model(pick, f" ({len(names)} models offered to this key)")


def verify_key(key, provider):
    """One live request, up to three attempts with a printed backoff (5 s, 10 s) — but only
    a network fault is ever retried; every other outcome is decided on the first attempt.

    Returns exactly one of: ok · rate-limited · bad-key · model-not-found · network · server ·
    provider-error, for BOTH providers (the Groq arm classifies through the same
    _classify_error, so a Groq timeout is never reported as a rejected key). A 404 is NOT retried — it is deterministic, so a second attempt would
    waste thirty seconds to fail identically; instead we ask the key which models it can
    use and switch (MODEL_FALLBACK), which is the only thing that can actually fix it.
    """
    if provider == "groq":
        try:
            import groq  # noqa
        except ImportError:
            _p("Installing groq SDK (Groq key found, no Gemini key)…")
            os.system(f"{sys.executable} -m pip install -q groq")
        try:
            import groq
            S["client"] = groq.Groq(api_key=key)
            _pace()
            S["client"].chat.completions.create(model=GROQ_MODEL, max_tokens=5,
                                                messages=[{"role": "user", "content": "Reply with the single word OK."}])
            S["n_requests"] += 1
            return "ok"
        except Exception as e:
            kind = _classify_error(e)
            _say(f"Groq key check: {kind} ({type(e).__name__}: {str(e)[:90]})")
            # Same honesty rule as the Gemini arm: a Groq timeout is a network problem and a
            # Groq 429 is a busy quota — neither is a rejected key.
            if kind == "rate-limit":
                return "rate-limited"
            if kind in ("network", "server", "model-not-found", "bad-key"):
                return kind
            return "provider-error"
    S["client"] = genai.Client(api_key=key, http_options=gtypes.HttpOptions(timeout=30000))
    use_model(MODEL)                       # a re-run of cell 0 starts from the configured name again
    switched = False
    last = "network"
    for attempt in (1, 2, 3):
        try:
            _pace()
            resp = S["client"].models.generate_content(
                model=MODEL_IN_USE, contents="Reply with the single word OK.",
                config=gtypes.GenerateContentConfig(temperature=0, max_output_tokens=5))
            S["n_requests"] += 1
            _ = _safe_text(resp)
            return "ok"
        except Exception as e:
            kind = _classify_error(e)
            last = kind
            _say(f"key check attempt {attempt}/3: {kind} ({type(e).__name__}: {str(e)[:90]})")
            if kind == "rate-limit":
                return "rate-limited"          # the key works; quota is busy right now
            if kind == "bad-key":
                return "bad-key"
            if kind == "model-not-found":
                # Deterministic: do not back off and try the same name again.
                if switched or resolve_model(S["client"], model_from_error(e)) is None:
                    return "model-not-found"
                switched = True
                continue
            if attempt < 3:
                _countdown(5 * attempt, "  backing off")
    return last if last in ("network", "server") else "provider-error"


def _pace():
    """≥ 7 s between live requests, with a visible countdown."""
    wait = PACE_SECONDS - (time.time() - S["last_request"])
    if wait > 0:
        _countdown(wait, "free-tier pacing")
    S["last_request"] = time.time()


# ----------------------------------------------------------------------------
# Cell 0 — bootstrap
# ----------------------------------------------------------------------------
def bootstrap(reconnect=False):
    t0 = time.time()
    _hr("MGRCM0065 Week 3 · AI for Finance — START HERE" if not reconnect else "🔁 reconnect")
    try:
        load_data()
    except Exception as e:
        S["mode"] = MODE_ERROR
        _p(f"Data error: {type(e).__name__}: {e}")
        _p(f"=== {MODE_ERROR} ===")
        _say("data failed to load — watch the projector; we fix this at the break")
        return
    ver = S["version"]
    cache_kind = "SAMPLE" if S["cache_meta"].get("sample", True) else "real"
    _p(f"Notebook v{ver.get('notebook', '?')} · data v{ver.get('data', '?')} · cache: {cache_kind}")
    _p(f"prices to {TODAY} · {len(S['headlines'])} headlines + "
       f"{0 if S['fresh'] is None else len(S['fresh'])} fresh")
    _p(f"Data source: {S['data_source']}")
    _recorded = str(S["cache_meta"].get("today_constant", "") or "")
    if _recorded and _recorded != TODAY:
        _p(f"⚠️  recorded answers are for {_recorded} but prices run to {TODAY} — Labs 3 and 4 "
           "would fall through to 'no cached answer'; tell the instructor.")
    if not reconnect:
        fetch_repo_updates()
    sdk = ensure_sdk()
    if not sdk:
        S["mode"] = MODE_STUDIO
    key, provider, status = find_key()
    S["provider"] = provider
    S["key_status"] = status
    if FORCE_CACHED:
        S["mode"] = MODE_STUDIO
        _say("FORCE_CACHED is on → STUDIO MODE regardless of your key.")
    elif not sdk:
        pass
    elif status == "notebook-access":
        S["mode"] = MODE_STUDIO
        _say("Key present but THIS copy cannot see it (NotebookAccessError): click 🔑 in the left "
             "sidebar, toggle 'Notebook access' ON for GEMINI_API_KEY, then re-run this cell.")
    elif key is None:
        if S["in_colab"] and not reconnect:
            try:
                from getpass import getpass
                pasted = getpass("No key found in Secrets. Paste your GEMINI_API_KEY now, or press Enter for STUDIO MODE: ")
                if pasted.strip():
                    key, provider, status = pasted.strip(), "gemini", "pasted"
                    S["provider"] = provider
            except Exception:
                pass
        if key is None:
            S["mode"] = MODE_STUDIO
            _say("No key found (Secrets: GEMINI_API_KEY / GROQ_API_KEY; or the environment). "
                 "STUDIO MODE is fine — see the sub-line below.")
    if key is not None and S["mode"] is None:
        result = verify_key(key, provider)
        if result in ("ok", "rate-limited"):
            S["mode"] = MODE_LIVE
            if result == "rate-limited":
                _say("Key works but the free-tier quota is busy right now — cached twins will fill any gap.")
        elif result == "bad-key":
            S["mode"] = MODE_STUDIO
            _say(f"Key rejected by the provider: re-copy it from {_key_console(provider)} (no spaces), "
                 "save the Secret, re-run this cell. STUDIO MODE meanwhile.")
        elif result == "model-not-found":
            S["mode"] = MODE_STUDIO
            _say(f"The model {GROQ_MODEL if provider == 'groq' else MODEL} is not available to this "
                 "API key. This usually means the provider has renamed or withdrawn it. Your key and "
                 "your Wi-Fi are both fine. Tell the instructor, who runs build/list_models.py to see "
                 "what the key can use and then sets MODEL in build/w3_core.py. STUDIO MODE "
                 "meanwhile — every number, order and refusal in today's session still runs.")
        elif result == "server":
            S["mode"] = MODE_STUDIO
            _say("The provider's own servers returned an error three times (nothing to do with "
                 "your key or your Wi-Fi). Wait a minute, then re-run this cell — that is the "
                 "retry button. STUDIO MODE meanwhile.")
        elif result == "network":
            S["mode"] = MODE_STUDIO
            _say("Network timed out three times: check Wi-Fi, then re-run this cell (that is the retry "
                 "button). STUDIO MODE meanwhile.")
        else:
            S["mode"] = MODE_STUDIO
            _say("The provider returned an error we do not have a specific remedy for — the exact "
                 "wording is in the attempt lines above; show them to the instructor. Re-running "
                 "this cell is the retry button. STUDIO MODE meanwhile.")
    _hr()
    _p(f"=== {S['mode']} ===")
    if S["mode"] == MODE_STUDIO:
        _say(STUDIO_SUBLINE)
        meta = S["cache_meta"]
        if meta.get("sample", True):
            _p(f"⚠️  {SAMPLE_BANNER}")
            _say(f"cached wording is stand-in text; the real cache will be produced with: "
                 f"{meta.get('model_placeholder', MODEL)}")
        else:
            _say(f"cached answers were produced with: {meta.get('model', MODEL)} "
                 f"(instructor's pre-run on {meta.get('generated_on', '?')})")
    if S["mode"] == MODE_LIVE:
        if provider == "groq":
            _say(f"provider: Groq · model {GROQ_MODEL} · single-turn cells live; Lab 3 replays the "
                 "instructor's trace")
        else:
            _say(f"model: {MODEL_IN_USE} · pacing ≥ {PACE_SECONDS} s between requests · "
                 f"read-only tool-call cap {TOOL_CAP} per turn (+{ORDER_CAP} orders)")
    _p("Runtime type: CPU (no GPU needed)")
    _say("Google account: Colab, the Secret and the key must all be the same PERSONAL account "
         "(check the avatar, top right).")
    _p("Do NOT use Runtime → Run all — this notebook is a sequence of reveals.")
    _p(f"(cell 0 finished in {time.time() - t0:.1f} s at {_now()})")
    globals()["_BOOTSTRAPPED"] = True


def reconnect():
    """🔁 Disconnected? Run this first. Idempotent; keeps a real lock if it survived.

    If the runtime lost the lock, a PLACEHOLDER lock is set so Labs 2–4 open. The placeholder is
    marked "restored": 1.4 never scores it (it would read 0/8) and 1.2 is allowed to overwrite it,
    so a student who wants the scoreboard back runs 1.2 → 1.3 → 1.4 once more.
    """
    global PREDICTIONS_LOCKED
    was_alive = bool(globals().get("_BOOTSTRAPPED", False))   # read BEFORE bootstrap() sets it
    bootstrap(reconnect=True)
    if PREDICTIONS_LOCKED is None:
        PREDICTIONS_LOCKED = (_now(), {}, "restored")
        globals()["PREDICTIONS_LOCKED"] = PREDICTIONS_LOCKED
        if was_alive:
            # Cell 0 already ran in this kernel: nothing was lost — the student simply never ran 1.2.
            _say("You have not locked any predictions yet — run 1.2 (then 1.3 and 1.4) if you want "
                 "the scoreboard; Labs 2–4 are open either way, on your honour.")
        else:
            _say("Your prediction lock was lost with the runtime — Labs 2–4 are open on your honour; "
                 "the calls you noted down are the record. Want the scoreboard back? Run 1.2 (set the "
                 "dropdowns to what you wrote), then 1.3 and 1.4 — the notebook allows this once.")
    _p("🔁 reconnected — carry on from where you were.")


# ----------------------------------------------------------------------------
# Gating
# ----------------------------------------------------------------------------
def require(condition, message):
    if not condition:
        _p(f"⛔ {message}")
        return False
    return True


def locked():
    """True for a real lock AND for the reconnect placeholder (both open Labs 2–4)."""
    return globals().get("PREDICTIONS_LOCKED") is not None


def lock_is_placeholder():
    """True only for the placeholder a 🔁 cell sets after the runtime lost the real lock."""
    lk = globals().get("PREDICTIONS_LOCKED")
    return lk is not None and len(lk) >= 3 and lk[2] == "restored"


# ----------------------------------------------------------------------------
# Tools (Lab 3). All slice by TODAY. GBP is notional only (no FX applied).
# ----------------------------------------------------------------------------
def _series(ticker):
    if ticker not in TICKERS:
        raise ValueError(f"unknown ticker {ticker!r}; choose from {', '.join(TICKERS)}")
    df = S["prices"][S["prices"]["ticker"] == ticker]
    return df[df["date"] <= TODAY].set_index("date")


def _prior_trading_day(s, date):
    date = pd.to_datetime(date).strftime("%Y-%m-%d")
    if date > TODAY:
        raise ValueError(f"{date} is after TODAY ({TODAY}) — the future is not available")
    idx = s.index[s.index <= date]
    if len(idx) == 0:
        raise ValueError(f"no prices before {date}")
    return idx[-1], (idx[-1] != date)


def get_price(ticker, date):
    """Closing price in USD on `date` (or the nearest PRIOR trading day)."""
    s = _series(str(ticker).upper())
    d, moved = _prior_trading_day(s, date)
    out = {"ticker": ticker.upper(), "date": d, "close_usd": round(float(s.loc[d, "close"]), 2)}
    if moved:
        out["note"] = f"{date} was not a trading day; nearest prior trading day {d} used"
    return out


def get_return(ticker, start, end):
    """Total return (adjusted closes) from `start` to `end`, both nearest-prior-trading-day."""
    s = _series(str(ticker).upper())
    d0, _ = _prior_trading_day(s, start)
    d1, _ = _prior_trading_day(s, end)
    r = float(s.loc[d1, "adj_close"] / s.loc[d0, "adj_close"] - 1)
    return {"ticker": ticker.upper(), "start": d0, "end": d1, "return_pct": round(100 * r, 2)}


def get_volatility(ticker, window_days=30):
    """Annualised volatility (%) of daily log returns over the last `window_days` trading days."""
    s = _series(str(ticker).upper())["adj_close"]
    w = int(window_days)
    lr = np.log(s).diff().dropna().tail(w)
    return {"ticker": ticker.upper(), "window_days": w, "as_of": TODAY,
            "annualised_vol_pct": round(float(lr.std() * np.sqrt(252) * 100), 2)}


def get_headlines(ticker):
    """Headlines for `ticker` dated ≤ TODAY: date + text only (no returns, no labels)."""
    t = str(ticker).upper()
    rows = [S["headlines"]] + ([S["fresh"]] if S["fresh"] is not None else [])
    df = pd.concat(rows, ignore_index=True)
    df = df[(df["ticker"] == t) & (df["date"] <= TODAY)].sort_values("date")
    return {"ticker": t, "headlines": [{"date": r.date, "text": r.headline} for r in df.itertuples()]}


def guard_check(ticker, gbp_notional, side):
    """The governance guard lives HERE, not in the mandate text."""
    side = str(side).upper()
    if str(ticker).upper() not in TICKERS:
        return f"REFUSED: unknown ticker {ticker}"
    if side != "BUY":
        return f"REFUSED: long-only mandate — {side} orders are not permitted"
    if float(gbp_notional) > GUARD["MAX_ORDER_GBP"]:
        return f"REFUSED: £{float(gbp_notional):,.0f} exceeds MAX_ORDER_GBP £{GUARD['MAX_ORDER_GBP']:,.0f}"
    if GUARD["REQUIRE_HUMAN_APPROVAL"]:
        return "REFUSED: REQUIRE_HUMAN_APPROVAL is on — no human has approved this order"
    return "FILLED"


def place_order(ticker, gbp_notional, side="BUY"):
    """Paper order (GBP notional, long-only). Every call is written to ORDER_BOOK."""
    status = guard_check(ticker, gbp_notional, side)
    rec = {"time": _now(), "ticker": str(ticker).upper(), "side": str(side).upper(),
           "gbp_notional": float(gbp_notional), "status": status}
    if status == "FILLED":
        rec["price_usd"] = get_price(ticker, TODAY)["close_usd"]
    ORDER_BOOK.append(rec)
    return {"status": status, "ticker": rec["ticker"], "gbp_notional": rec["gbp_notional"], "side": rec["side"]}


TOOLS = {"get_price": get_price, "get_return": get_return, "get_volatility": get_volatility,
         "get_headlines": get_headlines, "place_order": place_order}
READ_ONLY_TOOLS = ["get_price", "get_return", "get_volatility", "get_headlines"]
MANDATE_TOOLS = READ_ONLY_TOOLS + ["place_order"]

_TOOL_DECLS = {
    "get_price": ("Closing share price in USD of a ticker on a date (YYYY-MM-DD); uses the nearest prior trading day.",
                  {"ticker": ("STRING", "one of " + ", ".join(TICKERS)), "date": ("STRING", "YYYY-MM-DD, not after today")},
                  ["ticker", "date"]),
    "get_return": ("Total return in percent between two dates (YYYY-MM-DD) using adjusted closes.",
                   {"ticker": ("STRING", "ticker"), "start": ("STRING", "YYYY-MM-DD"), "end": ("STRING", "YYYY-MM-DD")},
                   ["ticker", "start", "end"]),
    "get_volatility": ("Annualised volatility in percent over the last N trading days.",
                       {"ticker": ("STRING", "ticker"), "window_days": ("INTEGER", "number of trading days, e.g. 30")},
                       ["ticker", "window_days"]),
    "get_headlines": ("Recent news headlines for a ticker (date and text only).",
                      {"ticker": ("STRING", "ticker")}, ["ticker"]),
    "place_order": ("Place a paper (simulated) order in GBP notional. Long-only: side must be BUY.",
                    {"ticker": ("STRING", "ticker"), "gbp_notional": ("NUMBER", "order size in GBP"),
                     "side": ("STRING", "BUY")}, ["ticker", "gbp_notional", "side"]),
}


def tool_declarations(names):
    """google-genai function declarations (verified against types.py 2.24.0)."""
    decls = []
    for n in names:
        desc, props, req = _TOOL_DECLS[n]
        decls.append(gtypes.FunctionDeclaration(
            name=n, description=desc,
            parameters=gtypes.Schema(
                type="OBJECT",
                properties={k: gtypes.Schema(type=t, description=d) for k, (t, d) in props.items()},
                required=req)))
    return [gtypes.Tool(function_declarations=decls)]


# Models invent argument names. The live run on 17 Sep 2026 called
# place_order(size=200000, ticker='AAPL') — 'size' for 'gbp_notional', and no 'side' at all.
# run_tool would answer with a TypeError message, which costs a turn and, worse, leaves the
# recorded step with no status so run_3_2 prints "wants to buy £0". We canonicalise the names
# BEFORE the call so both the execution and the recorded script use the declared ones.
_ARG_ALIASES = {
    "place_order": {"size": "gbp_notional", "size_gbp": "gbp_notional", "gbp": "gbp_notional",
                    "gbp_size": "gbp_notional", "gbp_amount": "gbp_notional", "amount": "gbp_notional",
                    "notional": "gbp_notional", "order_size": "gbp_notional", "quantity": "gbp_notional",
                    "value": "gbp_notional", "symbol": "ticker", "stock": "ticker",
                    "direction": "side", "action": "side"},
    "get_price": {"symbol": "ticker", "stock": "ticker", "as_of": "date", "day": "date"},
    "get_return": {"symbol": "ticker", "stock": "ticker", "from": "start", "to": "end",
                   "start_date": "start", "end_date": "end"},
    "get_volatility": {"symbol": "ticker", "stock": "ticker", "window": "window_days",
                       "days": "window_days", "lookback": "window_days"},
    "get_headlines": {"symbol": "ticker", "stock": "ticker"},
}


def _clean_number(v):
    """'£200,000' / '200000 GBP' -> 200000.0; anything else is handed back untouched."""
    if isinstance(v, str):
        t = re.sub(r"[£$,\s]|gbp|usd", "", v, flags=re.I)
        try:
            return float(t)
        except ValueError:
            return v
    return v


def normalise_args(name, args):
    """Map the argument names a model actually produces onto the ones we declared."""
    alias = _ARG_ALIASES.get(name, {})
    out = {}
    for k, v in (args or {}).items():
        out[alias.get(str(k).lower(), k)] = v
    if name == "place_order":
        if "gbp_notional" in out:
            out["gbp_notional"] = _clean_number(out["gbp_notional"])
        out.setdefault("side", "BUY")
    return out


def run_tool(name, args):
    if name not in TOOLS:
        return {"error": f"unknown tool {name}"}
    try:
        return TOOLS[name](**args)
    except TypeError as e:
        return {"error": f"bad arguments for {name}: {e}"}
    except Exception as e:
        return {"error": f"{type(e).__name__}: {e}"}


def _fmt_args(args):
    return ", ".join(f"{k}={v!r}" for k, v in args.items())


def _print_tool_call(name, args, result):
    res = json.dumps(result, default=str, ensure_ascii=False)
    if len(res) > 400:
        res = res[:400] + " …"
    _trace(f"🔧 TOOL CALL {name}({_fmt_args(args)}) → {res}")


# ----------------------------------------------------------------------------
# Safe text extraction and the three-tier parser
# ----------------------------------------------------------------------------
def _safe_text(resp):
    """Text from the first candidate; None if empty/blocked. Never raises."""
    try:
        cands = resp.candidates or []
        if not cands or cands[0].content is None or not cands[0].content.parts:
            return None
        gm = getattr(cands[0], "grounding_metadata", None)
        if gm is not None and (getattr(gm, "web_search_queries", None) or getattr(gm, "grounding_chunks", None)):
            S["grounding_seen"] = True
        parts = [p.text for p in cands[0].content.parts if getattr(p, "text", None) and not getattr(p, "thought", False)]
        text = "\n".join(parts).strip()
        return text or None
    except Exception:
        return None


CALL_SCHEMA = {
    "type": "ARRAY",
    "items": {"type": "OBJECT",
              "properties": {"id": {"type": "STRING"},
                             "call": {"type": "STRING", "enum": ["UP", "DOWN", "FLAT"]},
                             "reason": {"type": "STRING"}},
              "required": ["id", "call", "reason"]},
}


def parse_calls(text, ids):
    """Tier 1 JSON → tier 2 regex → tier 3 UNPARSED. Returns {id: (call, reason)}. Never raises."""
    out = {}
    if text:
        try:  # tier 1
            data = json.loads(text)
            if isinstance(data, dict):
                data = data.get("calls") or data.get("items") or list(data.values())
            for row in data:
                i = str(row.get("id", "")).strip().upper()
                c = str(row.get("call", "")).strip().upper()
                if i in ids and c in ("UP", "DOWN", "FLAT"):
                    out[i] = (c, str(row.get("reason", "")).strip())
        except Exception:
            pass
        if len(out) < len(ids):  # tier 2
            for i in ids:
                if i in out:
                    continue
                m = re.search(rf"{i}\b[^A-Za-z]{{0,20}}(UP|DOWN|FLAT)\b", text, re.I)
                if m:
                    out[i] = (m.group(1).upper(), "(parsed from free text)")
    for i in ids:  # tier 3
        if i not in out:
            out[i] = ("UNPARSED", (text or "(no text)")[:80])
    return out


# ----------------------------------------------------------------------------
# ask(): the only path to a model
# ----------------------------------------------------------------------------
def _cache_key(cell_id, prompt):
    return f"{cell_id}:{_sha(prompt)}"


def _cached_entry(cell_id, prompt):
    return S["cache"].get(_cache_key(cell_id, prompt))


def _record(cell_id, prompt, text, tool_script=None, extra=None):
    """build_cache.py switches recording on; live results are collected here."""
    if S["recording"] is None:
        return
    key = _cache_key(cell_id, prompt)
    entry = S["recording"].setdefault(key, {"cell_id": cell_id, "prompt": prompt, "text": None,
                                            "tool_script": [], "variants": {}})
    if text is not None and entry["text"] is None:
        entry["text"] = text
    if tool_script:
        entry["tool_script"] = tool_script
    if extra:
        for k, v in extra.items():
            if isinstance(v, dict):
                entry.setdefault(k, {}).update(v)
            else:
                entry[k] = v


def _replay(entry, tools):
    """Replay a stored tool script through the REAL tool functions (200 ms per step)."""
    script = entry.get("tool_script") or []
    alt = entry.get("tool_script_refusal") or []
    # The refusal script is swapped in only when the fill script's first order would be refused
    # under the student's guard AND the refusal script itself places an order — otherwise the fill
    # script replays and the guard refuses it, so the book and the closing text still agree.
    if alt and any(step["name"] == "place_order" for step in alt) \
            and any(step["name"] == "place_order" for step in script):
        first = next(step for step in script if step["name"] == "place_order")
        if guard_check(first["args"].get("ticker"), first["args"].get("gbp_notional", 0),
                       first["args"].get("side", "BUY")) != "FILLED":
            script = alt
    executed = []
    for step in script:
        time.sleep(0.2)
        result = run_tool(step["name"], dict(step["args"]))
        _print_tool_call(step["name"], step["args"], result)
        executed.append({"name": step["name"], "args": step["args"], "result": result})
    text = entry.get("text")
    orders = [e for e in executed if e["name"] == "place_order"]
    variants = entry.get("variants") or {}
    if orders:
        statuses = [o["result"].get("status", "") for o in orders]
        if all(st == "FILLED" for st in statuses):
            text = variants.get("after_fill") or text
        elif statuses[-1] == "FILLED":
            text = variants.get("after_partial") or (
                "(no cached closing text for this outcome — the instructor's pre-run recorded a fill "
                "and a refusal but not a refused-then-filled turn; the ORDER BOOK above is the record)")
        else:
            text = variants.get("after_refusal") or text
    return text, executed


def _groq_generate(prompt, temperature, persona, schema):
    msgs = ([{"role": "system", "content": persona}] if persona else []) + [{"role": "user", "content": prompt}]
    kw = {"model": GROQ_MODEL, "messages": msgs, "temperature": float(temperature), "max_tokens": 1200}
    if schema:
        kw["response_format"] = {"type": "json_object"}
        msgs[-1]["content"] += "\nReturn JSON: {\"calls\": [{\"id\": …, \"call\": UP|DOWN|FLAT, \"reason\": …}]}"
    _pace()
    r = S["client"].chat.completions.create(**kw)
    S["n_requests"] += 1
    return r.choices[0].message.content


def _gemini_generate(cell_id, prompt, temperature, persona, schema, tools):
    """One turn, including our OWN tool loop (automatic function calling disabled)."""
    cfg = gtypes.GenerateContentConfig(
        temperature=float(temperature),
        system_instruction=persona if persona else None,
        automatic_function_calling=gtypes.AutomaticFunctionCallingConfig(disable=True))
    if schema:
        cfg.response_mime_type = "application/json"
        cfg.response_schema = schema
    if tools:
        cfg.tools = tool_declarations(tools)
        _say(f"tool-call cap applied: {TOOL_CAP} read-only calls per turn"
             + (f" plus up to {ORDER_CAP} orders" if any(t not in READ_ONLY_TOOLS for t in tools) else "")
             + " (automatic function calling OFF; loop is ours)")
    contents = [gtypes.Content(role="user", parts=[gtypes.Part.from_text(text=prompt)])]
    tool_script, n_reads, n_orders = [], 0, 0
    for _round in range(TOOL_CAP + ORDER_CAP + 2):
        _pace()
        resp = S["client"].models.generate_content(model=MODEL_IN_USE, contents=contents, config=cfg)
        S["n_requests"] += 1
        _p(f"   [request #{S['n_requests']} · {_now()}]")
        cand = (resp.candidates or [None])[0]
        parts = list(cand.content.parts) if (cand is not None and cand.content is not None and cand.content.parts) else []
        calls = [p.function_call for p in parts if getattr(p, "function_call", None)]
        if not calls:
            return _safe_text(resp), tool_script
        contents.append(cand.content)
        responses = []
        for fc in calls:
            args = normalise_args(fc.name, dict(fc.args or {}))
            is_order = fc.name not in READ_ONLY_TOOLS
            spent = n_orders if is_order else n_reads
            cap = ORDER_CAP if is_order else TOOL_CAP
            if spent >= cap:
                what = "order cap" if is_order else "read-only tool-call cap"
                result = {"error": f"{what} of {cap} reached — answer with what you have"}
                _trace(f"🔧 TOOL CALL {fc.name}({_fmt_args(args)}) → CAP REACHED ({cap})")
            else:
                result = run_tool(fc.name, args)
                if is_order:
                    n_orders += 1
                else:
                    n_reads += 1
                _print_tool_call(fc.name, args, result)
                tool_script.append({"name": fc.name, "args": args, "result": result})
            responses.append(gtypes.Part.from_function_response(name=fc.name, response={"result": result}))
        contents.append(gtypes.Content(role="user", parts=responses))
        # Silence function calling only once every tool THIS turn was given has run out of budget.
        # Gating it on the read budget alone (as it was) forbade the model from ever emitting the
        # order, whatever the order cap said; gating it on both counters regardless of which tools
        # were offered would leave a read-only turn burning rounds on calls that can only be refused.
        reads_left = n_reads < TOOL_CAP and any(t in READ_ONLY_TOOLS for t in tools)
        orders_left = n_orders < ORDER_CAP and any(t not in READ_ONLY_TOOLS for t in tools)
        if not reads_left and not orders_left:
            cfg.tool_config = gtypes.ToolConfig(function_calling_config=gtypes.FunctionCallingConfig(mode="NONE"))
    return None, tool_script


def ask(cell_id, prompt, temperature=0, tools=None, persona=None, schema=None):
    """Returns {'text', 'source' ∈ live|cached|miss, 'tool_calls', 'sample'}. Never raises for cache paths."""
    entry = _cached_entry(cell_id, prompt)
    sample = bool(S["cache_meta"].get("sample", True))
    live = (S["mode"] == MODE_LIVE) and not FORCE_CACHED
    if live and S["provider"] == "groq" and tools:
        _say("Groq provider: Lab 3 replays the instructor's trace (Groq path is single-turn only).")
        live = False
    book_before = list(ORDER_BOOK)
    if live:
        attempt, switched = 0, False
        while attempt < 2:
            attempt += 1
            # Rewind the book before EVERY attempt. The retry re-runs the whole tool loop, so a
            # 5xx arriving after a fill would otherwise book the same order twice and put two
            # £1,000,000 fills on the projector against a £1,000,000 mandate.
            del ORDER_BOOK[:]
            ORDER_BOOK.extend(book_before)
            try:
                if S["provider"] == "groq":
                    text, script = _groq_generate(prompt, temperature, persona, schema), []
                else:
                    text, script = _gemini_generate(cell_id, prompt, temperature, persona, schema, tools)
                if text is None:
                    _p("empty/blocked response from the model → serving the cached twin")
                    break
                _record(cell_id, prompt, text, script)
                return {"text": text, "source": "live", "tool_calls": script, "sample": False}
            except Exception as e:
                kind = _classify_error(e)
                if kind == "rate-limit":
                    _p(f"429 / RESOURCE_EXHAUSTED (free-tier limit) → {CACHED_BANNER}")
                    break
                if kind == "model-not-found":
                    # The model went away mid-session. Retrying the same name cannot help,
                    # so either switch to one this key can use or serve the cached twin.
                    # A successful switch buys one extra pass (on the NEW name) even if the
                    # 404 arrived on the second attempt; `switched` makes that a one-shot,
                    # so this cannot loop.
                    if S["provider"] == "gemini" and not switched and resolve_model(
                            S["client"], model_from_error(e)) is not None:
                        switched = True
                        attempt -= 1
                        continue
                    _say(f"the model {MODEL_IN_USE} is not available to this key (this is not a "
                         f"network problem) → {CACHED_BANNER}")
                    break
                _p(f"model error ({type(e).__name__}: {str(e)[:100]})")
                if attempt == 1:
                    _countdown(30, "one retry after")
                else:
                    _p(f"second failure → {CACHED_BANNER}")
        del ORDER_BOOK[:]
        ORDER_BOOK.extend(book_before)
    if entry is None:
        _say("no cached answer for your edited prompt — reset to default or use a live laptop.")
        return {"text": None, "source": "miss", "tool_calls": [], "sample": sample}
    _p(CACHED_BANNER)
    if sample:
        _p("   (SAMPLE — stand-in text, not model output)")
    if tools:
        text, executed = _replay(entry, tools)
    else:
        text, executed = entry.get("text"), []
    return {"text": text, "source": "cached", "tool_calls": executed, "sample": sample}


# ----------------------------------------------------------------------------
# Prompts (shared by the notebook and the cache builders — the hash is the key)
# ----------------------------------------------------------------------------
def flat_rule():
    return (f"FLAT means the return over the first trading session that could react to the headline "
            f"is between -{FLAT_BAND*100:.1f}% and +{FLAT_BAND*100:.1f}%.")


def prompt_lab1(rows, anonymised=False):
    lines = []
    for r in rows.itertuples():
        if anonymised:
            lines.append(f"{r.id}: {ANON_HEADLINES.get(r.id, r.headline)}")
        else:
            lines.append(f"{r.id} ({r.date}, {COMPANY.get(r.ticker, r.ticker)}, {r.ticker}): {r.headline}")
    head = ("You are a financial expert with stock recommendation experience. For each headline below, "
            "decide whether the company's share price went UP, DOWN or stayed FLAT over the first "
            f"trading session that could react to it. {flat_rule()} "
            "Use only what the headline says. Give a one-sentence reason per headline. "
            "Return a JSON array of objects with keys id, call (UP, DOWN or FLAT) and reason.")
    if anonymised:
        head += " The company names, tickers and dates have been removed on purpose."
    return head + "\n\n" + "\n".join(lines)


def prompt_recall(row):
    return (f"Here is a news headline published on {row['date']} about {COMPANY.get(row['ticker'], row['ticker'])} "
            f"({row['ticker']}): \"{row['headline']}\".\n"
            "In one paragraph of at most five sentences, describe what happened to the company's share price and "
            "to the company in the weeks and months that followed this headline.")


def prompt_hallucination():
    names = ", ".join(f"{COMPANY[t]} ({t})" for t in HALLUCINATION_TICKERS)
    return (f"Give your best estimate of the closing share price in US dollars of {names} on {HALLUCINATION_DATE}. "
            "Answer with exactly three lines in the form 'TICKER: price', for example 'XYZ: 123.45'. "
            "Do not add disclaimers, ranges or caveats — a single best estimate per line.")


PERSONAS = {
    "risk": "You are a cautious risk officer at a UK pension fund. You answer in three sentences.",
    "pm": "You are a bullish hedge-fund portfolio manager. You answer in three sentences.",
}


def prompt_persona():
    return (f"Should a UK pension fund buy {COMPANY[COMMITTEE_TICKER]} ({COMMITTEE_TICKER}) shares this month? "
            "End with one line 'RECOMMENDATION: BUY' or 'RECOMMENDATION: DO NOT BUY'.")


def prompt_temperature():
    return (f"In one sentence, what is the biggest risk of holding {COMPANY[COMMITTEE_TICKER]} "
            f"({COMMITTEE_TICKER}) shares for the next year?")


ANALYST_PROMPT_TEMPLATE = (
    "Today is {today}. You are a junior equity analyst with access to tools. Write a short briefing "
    "(at most 120 words) on {company} ({ticker}) for a portfolio manager: current price, performance over "
    "the last month and the last three months, current volatility, and the most recent news. "
    "Use the tools for every number you state. Do not place any orders.")


def prompt_analyst():
    return ANALYST_PROMPT_TEMPLATE.format(today=TODAY, company=COMPANY[COMMITTEE_TICKER], ticker=COMMITTEE_TICKER)


MANDATE_PREAMBLE = ("Paper-trading classroom simulation; no real money. You MUST call place_order exactly once "
                    "with your chosen ticker and size before your final answer.")
MANDATE_BODY = (" Today is {today}. You manage a £{gbp:,} paper mandate, long-only, and may invest in any of "
                "{tickers}. Using your tools, choose ONE stock and ONE order size in GBP, place the order, "
                "then explain your choice in at most 80 words. If an order is refused, say so and explain "
                "what you will do next.")


def mandate_text(mandatory=True, extra=""):
    body = MANDATE_BODY.format(today=TODAY, gbp=MANDATE_GBP, tickers=", ".join(TICKERS))
    return ((MANDATE_PREAMBLE if mandatory else
             "Paper-trading classroom simulation; no real money. You have a place_order tool.") + body + extra).strip()


# ----------------------------------------------------------------------------
# Scoring helpers
# ----------------------------------------------------------------------------
def score(calls, rows):
    """calls: {id: call}; returns (hits, n)."""
    hits = sum(1 for r in rows.itertuples() if calls.get(r.id, "").upper() == str(r.label).upper())
    return hits, len(rows)


def _label_word(label):
    return {"UP": "UP", "DOWN": "DOWN", "FLAT": "FLAT"}.get(str(label).upper(), str(label))


def _fmt_pct(x):
    return "n/a" if x is None or (isinstance(x, float) and math.isnan(x)) else f"{100*float(x):+.2f}%"


# ----------------------------------------------------------------------------
# Cell runners — each notebook code cell calls exactly one of these
# ----------------------------------------------------------------------------
def run_1_1():
    """Print the 8 headlines with the FLAT rule."""
    h = S["headlines"]
    _hr("Lab 1 · 8 real, dated headlines")
    _say(f"FLAT_BAND = ±{FLAT_BAND*100:.1f}%  — {flat_rule()}")
    _say("Scoring uses the return over the first trading session that could react to the headline.")
    _p("")
    for r in h.itertuples():
        _trace(f"{r.id} · {r.date} · {COMPANY.get(r.ticker, r.ticker)} ({r.ticker})")
        _trace(f"     {r.headline}")
    _p("")
    _say("Note your 8 calls somewhere of your own, then set the dropdowns in 1.2 and run it to lock.")


def lock_predictions(calls):
    """1.2 — lock once, with a timestamp; refuse to re-lock."""
    global PREDICTIONS_LOCKED
    restoring = lock_is_placeholder()
    if locked() and not restoring:
        ts = PREDICTIONS_LOCKED[0]
        _say(f"⛔ locked at {ts} — your score stands; the dropdowns will not re-lock.")
        return
    calls = {k: str(v).upper() for k, v in calls.items()}
    bad = [k for k, v in calls.items() if v not in ("UP", "DOWN", "FLAT")]
    if bad:
        _p(f"⛔ {', '.join(bad)} not set to UP/DOWN/FLAT — fix the dropdowns and run again.")
        return
    PREDICTIONS_LOCKED = (_now(), calls, "locked")
    globals()["PREDICTIONS_LOCKED"] = PREDICTIONS_LOCKED
    if restoring:
        _say(f"🔒 predictions re-entered at {PREDICTIONS_LOCKED[0]} after the reconnect — on your honour, "
             "these should be the calls you noted before the reveal. Now run 1.3 and 1.4.")
    else:
        _say(f"🔒 predictions locked at {PREDICTIONS_LOCKED[0]} — on your honour; the notebook will notice a re-lock.")
    _say(" ".join(f"{k}:{v}" for k, v in calls.items()), indent="   ")


def run_1_3():
    """One batched, structured call: all 8 headlines to the model."""
    if not require(locked(), "run 1.2 first (lock your predictions) — the model is not called until you do"):
        return
    h = S["headlines"]
    prompt = prompt_lab1(h)
    _hr("Lab 1 · the model's calls (one request, JSON output)")
    _say(f"prompt includes: {flat_rule()}")
    res = ask("1.3", prompt, temperature=0, schema=CALL_SCHEMA)
    parsed = parse_calls(res["text"], list(h["id"]))
    S["results"]["1.3"] = {"parsed": parsed, "source": res["source"]}
    for i, (c, reason) in parsed.items():
        _trace(f"{i} → {c:<8} {reason}")
    if any(c == "UNPARSED" for c, _ in parsed.values()):
        _say("some rows are UNPARSED (scored as misses) — raw text shown; that is the finding, not a bug.")


def run_1_4():
    """Scoreboard: You k/8 · Model m/8 (flat band ±0.5 %)."""
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    if not require("1.3" in S["results"], "run 1.3 first (the model's calls)"):
        return
    if lock_is_placeholder():
        _hr("Lab 1 · scoreboard")
        _say("⛔ no scoreboard: your lock was lost with the runtime, so the notebook has no record of "
             "your calls — the calls you noted down are the record. To restore the scoreboard, run 1.2 "
             "(set the dropdowns to what you wrote), then 1.3 and 1.4 again.")
        return
    h = S["headlines"]
    you = PREDICTIONS_LOCKED[1]
    model = {i: c for i, (c, _) in S["results"]["1.3"]["parsed"].items()}
    ky, n = score(you, h)
    km, _ = score(model, h)
    S["results"]["1.4"] = {"you": ky, "model": km, "n": n}
    _hr("Lab 1 · scoreboard")
    _p(f"You: {ky}/{n} · Model: {km}/{n} (flat band ±{FLAT_BAND*100:.1f}%)")
    _p(f"model answer: {S['results']['1.3']['source']}")
    _p("")
    _p(f"{'id':<4} {'ticker':<6} {'truth':<6} {'ret':>8}  {'you':<6} {'model':<8} note")
    for r in h.itertuples():
        y, m = you.get(r.id, "—"), model.get(r.id, "—")
        note = "" if y == m else "← you disagreed with the model"
        if y != r.label and m == r.label:
            note = "← model right, you wrong"
        elif y == r.label and m != r.label:
            note = "← you right, model wrong"
        _p(f"{r.id:<4} {r.ticker:<6} {r.label:<6} {_fmt_pct(r.ret_event):>8}  {y:<6} {m:<8} {note}")
    _p("")
    _p("Q1 — Quote the headline where the model's reason surprised you.")


def _seed_row():
    h = S["headlines"]
    seed = h[h["group"] == "seed2020"]
    return (seed if len(seed) else h.tail(1)).iloc[0].to_dict()


def run_1_5a():
    """The reveal: the model 'predicts' the aftermath of the 2020 headline, beside the truth."""
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    row = _seed_row()
    _hr("Lab 1b · 1.5a — the 2020 headline")
    _trace(f"{row['id']} · {row['date']} · {COMPANY.get(row['ticker'], row['ticker'])}: {row['headline']}")
    res = ask("1.5a", prompt_recall(row), temperature=0)
    S["results"]["1.5a"] = res
    _p("")
    _p("MODEL SAYS:")
    for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
        _p("  " + line)
    _p("")
    _p("TRUTH FROM prices.csv (adjusted closes):")
    s = S["prices"][S["prices"]["ticker"] == row["ticker"]].set_index("date")["adj_close"]
    ev = row["event_date"]
    prior = s[s.index < ev]
    if len(prior) == 0:
        _say(f"  the price file does not go back far enough to score this headline (no close before "
             f"{ev} for {row['ticker']}) — pull prices from an earlier start date.")
    else:
        base = prior.iloc[-1]
        def ret_to(days):
            target = (pd.to_datetime(ev) + pd.Timedelta(days=days)).strftime("%Y-%m-%d")
            later = s[s.index <= target]
            return later.iloc[-1] / base - 1, later.index[-1]
        _p(f"  first reacting session {ev}: {_fmt_pct(row['ret_event'])}  (label {row['label']})")
        _p(f"  next session: {_fmt_pct(row['ret_next'])}")
        for d, name in ((35, "≈ 5 weeks later"), (182, "≈ 6 months later"), (365, "≈ 12 months later")):
            r, dd = ret_to(d)
            _p(f"  {name} ({dd}): {_fmt_pct(r)}")
    _p("")
    _p("Q2 — Quote the sentence in which the model tells you the future.")
    _say("A headline from 2020 is inside the model's reading; this is recall, not prediction.")


def _scored_arm(cell_id, rows, anonymised, title):
    prompt = prompt_lab1(rows, anonymised=anonymised)
    res = ask(cell_id, prompt, temperature=0, schema=CALL_SCHEMA)
    parsed = parse_calls(res["text"], list(rows["id"]))
    calls = {i: c for i, (c, _) in parsed.items()}
    k, n = score(calls, rows)
    S["results"][cell_id] = {"parsed": parsed, "hits": k, "n": n, "source": res["source"]}
    _p(f"{title}: Model {k}/{n} (flat band ±{FLAT_BAND*100:.1f}%) · {res['source']}")
    for r in rows.itertuples():
        c, reason = parsed[r.id]
        shown = ANON_HEADLINES.get(r.id, r.headline) if anonymised else r.headline
        _trace(f"{r.id} truth {r.label:<5} model {c:<8} {shown[:40]}…")
    return k, n


def run_1_5b():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5b — the same 8 headlines, names replaced by 'Company X'")
    _scored_arm("1.5b", S["headlines"], True, "Anonymised")


def run_1_5c():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5c — fresh headlines (last three weeks before build)")
    f = S["fresh"]
    if f is None or len(f) == 0:
        _say("fresh headlines unavailable — watch the projector (the instructor runs the canonical fresh arm).")
        S["results"]["1.5c"] = {"hits": None, "n": 0, "source": "unavailable"}
        return
    for r in f.itertuples():
        _trace(f"{r.id} · {r.date} · {r.ticker}: {r.headline}")
    _p("")
    _scored_arm("1.5c", f, False, "Fresh")
    if S.get("grounding_seen"):
        _say("note: the model reported search grounding — this is retrieval, not memory.")


def run_1_5d():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5d — comparison")
    R = S["results"]
    k13, n13 = (None, None)
    if "1.3" in R:   # scored from 1.3's own parsed calls, so the row is filled whenever 1.3 has run
        k13, n13 = score({i: c for i, (c, _) in R["1.3"]["parsed"].items()}, S["headlines"])
    rows = [("named, old (1.3)", k13, n13, R.get("1.3", {}).get("source")),
            ("anonymised, old (1.5b)", R.get("1.5b", {}).get("hits"), R.get("1.5b", {}).get("n"), R.get("1.5b", {}).get("source")),
            ("named, fresh (1.5c)", R.get("1.5c", {}).get("hits"), R.get("1.5c", {}).get("n"), R.get("1.5c", {}).get("source"))]
    _p(f"{'arm':<26}{'model':>8}   tag")
    for name, k, n, src in rows:
        val = "not run" if k is None else f"{k}/{n}"
        _p(f"{name:<26}{val:>8}   {src or '—'}")
    if "1.4" in R:
        _p(f"You (named, old): {R['1.4']['you']}/{R['1.4']['n']}")
    else:
        _p("You (named, old): not scored yet — run 1.4")
    _p("")
    _say("n = 8 is a design demonstration, not a result. Named vs anonymised speaks to recognition / "
         "distraction; fresh speaks to cut-off contamination — and we cannot verify this model's cut-off.")


def run_2A_1():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2A.1 — 'best estimate, no disclaimers'")
    res = ask("2A.1", prompt_hallucination(), temperature=0)
    S["results"]["2A.1"] = res
    _p("MODEL SAYS:")
    for line in (res["text"] or "(no answer)").splitlines():
        _trace("  " + line)


def run_2A_2():
    if not require("2A.1" in S["results"], "run 2A.1 first"):
        return
    _hr("Lab 2 · 2A.2 — the truth beside the model's numbers")
    text = (S["results"]["2A.1"]["text"] or "").replace(",", "")   # '1,234.56' → '1234.56' (no comma decimals)
    _p(f"{'ticker':<7}{'model':>10}{'true close':>12}{'error':>9}")
    any_number = False
    for t in HALLUCINATION_TICKERS:
        true = get_price(t, HALLUCINATION_DATE)["close_usd"]
        m = re.search(rf"{t}\b[^0-9\n]{{0,25}}\$?\s*([0-9]{{1,6}}(?:\.[0-9]{{1,2}})?)", text)
        if m:
            any_number = True
            val = float(m.group(1))
            _p(f"{t:<7}{val:>10.2f}{true:>12.2f}{100*(val/true-1):>8.1f}%")
        else:
            _p(f"{t:<7}{'declined':>10}{true:>12.2f}{'—':>9}")
    _say(f"(true closes from prices.csv, as quoted on {HALLUCINATION_DATE}; no split adjustment needed "
         "for these three)")
    if not any_number:
        _say("The model declined to give numbers — that refusal is the finding. Q3: which control does it replace?")
    else:
        _say("Q3 — Circle every wrong closing price. What went wrong, and which finance control catches it?")


def run_2B():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2B — same question, two job titles")
    out = {}
    for key, label in (("risk", "cautious risk officer"), ("pm", "bullish hedge-fund PM")):
        res = ask(f"2B.{key}", prompt_persona(), temperature=0, persona=PERSONAS[key])
        out[key] = res["text"]
        _p(f"▶ {label}:")
        for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
            _p("  " + line)
        _p("")
    S["results"]["2B"] = out
    _say("Q4 — Write the two recommendations (cautious risk officer / bullish hedge-fund PM). "
         "What changed, and which finance control catches it?")


def run_2C():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2C (stretch) — temperature 0 vs 1, three draws each")
    for temp in (0, 1):
        _p(f"temperature = {temp}")
        for d in (1, 2, 3):
            res = ask(f"2C.t{temp}.d{d}", prompt_temperature(), temperature=temp)
            _trace(f"  draw {d}: {res['text'] or '(no answer)'}")
    _say("At 0 the drafts are nearly (not exactly) identical; at 1 each draw is a new draft.")


def run_3_1():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 3 · 3.1 — analyst mode (four read-only tools)")
    _say(f"TODAY = {TODAY} · tools: {', '.join(READ_ONLY_TOOLS)}")
    res = ask("3.1", prompt_analyst(), temperature=0, tools=READ_ONLY_TOOLS)
    S["results"]["3.1"] = res
    _p("")
    _p(f"tool calls this turn: {len(res['tool_calls'])}")
    _p("FINAL ANSWER:")
    for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
        _p("  " + line)
    _p("")
    _say("Q5 — Tool calls: ___ . Did every number in the final answer come from a tool? YES / NO — "
         "if NO, write the unsourced number.")
    if S["mode"] == MODE_LIVE and not FORCE_CACHED:
        _say("Wait for the 20-second countdown before the next cell — the model is rate-limited, not broken.")
        _countdown(20, "rate-limit breather")
    else:
        _p("(no countdown needed in STUDIO MODE)")


def _snapshot_receipt(cell_id):
    """Remember the student's own 3.2/3.3 run (limit, approval, order book) for the Receipt cell."""
    S["results"]["receipt"] = {"cell": cell_id, "time": _now(), "guard": dict(GUARD),
                               "book": [dict(o) for o in ORDER_BOOK]}


def run_3_2(max_order_gbp, require_human_approval, mandate=None, cell_id="3.2"):
    """Mandate mode. Widgets are read INSIDE the cell that calls this; ORDER_BOOK cleared each run.

    cell_id "3.2" (3.2 and 3.3) is the run the Receipt reports. Any other cell_id (the 🟡 3.4 sceptic
    card, the 🔴 instructor demo) is a SCRATCH run: it prints its own trace and book, then puts the
    student's 3.2/3.3 limit and ORDER BOOK back so the Receipt is unaffected.
    """
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    scratch = cell_id != "3.2"
    saved = (dict(GUARD), [dict(o) for o in ORDER_BOOK])
    GUARD["MAX_ORDER_GBP"] = int(max_order_gbp)
    GUARD["REQUIRE_HUMAN_APPROVAL"] = bool(require_human_approval)
    del ORDER_BOOK[:]
    _p(f"ACTIVE LIMIT: £{GUARD['MAX_ORDER_GBP']:,} · HUMAN APPROVAL: {'ON' if GUARD['REQUIRE_HUMAN_APPROVAL'] else 'OFF'}")
    _say(f"paper mandate £{MANDATE_GBP:,} (GBP notional only; 1 USD ≈ 0.78 GBP is NOT applied) · "
         f"long-only · TODAY {TODAY}")
    if scratch:
        _say(f"(scratch run for cell {cell_id} — your 3.2/3.3 limit and ORDER BOOK are kept for the Receipt)")
    text = mandate or mandate_text()
    # The 🔴 demo cell passes cell_id="instructor": a scratch run that replays the 3.2 cache entry
    # (the cache key is cell_id + prompt hash, and the demo prompt IS the 3.2 prompt).
    res = ask("3.2" if cell_id == "instructor" else cell_id, text, temperature=0, tools=MANDATE_TOOLS)
    S["results"][cell_id] = res
    try:
        if res["source"] == "miss":
            print_order_book()
            return
        orders = [c for c in res["tool_calls"] if c["name"] == "place_order"]
        _p("")
        # An order step with no "status" never reached guard_check — the arguments were malformed.
        # Say so, rather than printing "£0" as though the agent had asked for nothing.
        good = [c for c in orders if isinstance(c.get("result"), dict) and c["result"].get("status")]
        if good:
            o = good[0]["args"]
            _p(f"the agent wants to buy £{float(o.get('gbp_notional', 0)):,.0f} of {str(o.get('ticker', '?')).upper()}")
        elif orders:
            _p("the agent's order was rejected as malformed:")
            for line in textwrap.wrap(_fmt_args(orders[0].get("args") or {}), TRACE_WIDTH - 4):
                _p("  " + line)
        else:
            _p("the agent placed NO order this turn")
        _p("FINAL ANSWER:")
        for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
            _p("  " + line)
        print_order_book()
        refused = [o["result"].get("status", "") for o in good if o["result"].get("status", "").startswith("REFUSED")]
        if refused:
            _say(f"guard that fired: {', '.join(_guards_fired(refused))}"
                 + (" (cached wording may name another control — the ORDER BOOK is the record)"
                    if res["source"] == "cached" else ""))
            _say("Q6 — When refused, the agent (a) tried a smaller order (b) argued (c) complied silently "
                 "(d) other — quote one sentence if there is one.")
        elif good and not scratch:
            _say("3.3 — set MAX_ORDER_GBP BELOW the amount above, tick REQUIRE_HUMAN_APPROVAL, and run this "
                 "cell again.")
    finally:
        if scratch:
            GUARD.clear()
            GUARD.update(saved[0])
            del ORDER_BOOK[:]
            ORDER_BOOK.extend(saved[1])
        else:
            _snapshot_receipt(cell_id)


def _guards_fired(statuses):
    """Which control(s) produced these REFUSED statuses (guard_check wording → control name)."""
    names = []
    for st in statuses:
        name = ("MAX_ORDER_GBP" if "MAX_ORDER_GBP" in st else
                "REQUIRE_HUMAN_APPROVAL" if "REQUIRE_HUMAN_APPROVAL" in st else
                "long-only mandate" if "long-only" in st else "ticker universe")
        if name not in names:
            names.append(name)
    return names


def print_order_book(book=None):
    book = ORDER_BOOK if book is None else book
    _p("ORDER BOOK:")
    if not book:
        _p("  (empty)")
    for o in book:
        px = f" @ ${o['price_usd']:.2f}" if "price_usd" in o else ""
        _trace(f"  {o['time']} {o['side']} {o['ticker']} £{o['gbp_notional']:,.0f}{px} — {o['status']}")


def run_receipt():
    """Receipt: scoreboard + the student's OWN 3.2/3.3 run (not the 3.4 scratch run)."""
    _hr("Receipt — keep this for your reflection")
    R = S["results"]
    if "1.4" in R:
        _p(f"Scoreboard: You {R['1.4']['you']}/{R['1.4']['n']} · Model {R['1.4']['model']}/{R['1.4']['n']} "
           f"(flat band ±{FLAT_BAND*100:.1f}%)")
    elif lock_is_placeholder():
        _say("Scoreboard: not in this runtime (lock lost over the break) — the score you noted down is the record.")
    else:
        _say("Scoreboard: run 1.4 first (1.2 → 1.3 → 1.4) — then re-run this cell.")
    for cid, name in (("1.5b", "anonymised"), ("1.5c", "fresh")):
        if cid in R and R[cid].get("hits") is not None:
            _p(f"  {name}: Model {R[cid]['hits']}/{R[cid]['n']}")
    _p(f"Mode: {S['mode']} · live requests this session: {S['n_requests']}")
    snap = R.get("receipt")
    if snap:
        g = snap["guard"]
        _p(f"Last 3.2/3.3 run at {snap['time']}: limit £{g['MAX_ORDER_GBP']:,} · "
           f"approval {'ON' if g['REQUIRE_HUMAN_APPROVAL'] else 'OFF'}")
        print_order_book(snap["book"])
    else:
        _p("No 3.2/3.3 run in this runtime yet — run 3.2 (and 3.3) first.")


def run_4_1():
    _hr(f"Lab 4 · the committee is convened on {COMMITTEE_TICKER}")
    md = S["committee_md"]
    if not md:
        md = ("# PLACEHOLDER — committee transcript not generated yet\n\n"
              "Run build/build_cache.py with a key (or build/make_sample_cache.py) to create cache/committee.md.")
    ip = None
    try:
        import IPython
        ip = IPython.get_ipython()
    except Exception:
        pass
    if ip is not None:
        from IPython.display import Markdown, display
        display(Markdown(md))
    else:
        _p(md)
    return md


def committee_context():
    """The Lab-3 tool outputs every committee role shares (real tool functions)."""
    t = COMMITTEE_TICKER
    d = pd.to_datetime(TODAY)
    ctx = {
        "price": get_price(t, TODAY),
        "ret_1m": get_return(t, (d - pd.Timedelta(days=30)).strftime("%Y-%m-%d"), TODAY),
        "ret_3m": get_return(t, (d - pd.Timedelta(days=91)).strftime("%Y-%m-%d"), TODAY),
        "ret_12m": get_return(t, (d - pd.Timedelta(days=365)).strftime("%Y-%m-%d"), TODAY),
        "vol_30d": get_volatility(t, 30),
        "headlines": get_headlines(t),
    }
    return ctx


COMMITTEE_ROLES = [
    ("bull", "Bull analyst", "Argue the case FOR buying {company} ({ticker}) now, in at most 120 words, using only the data below."),
    ("bear", "Bear analyst", "Argue the case AGAINST buying {company} ({ticker}) now, in at most 120 words, using only the data below."),
    ("risk", "Risk manager", "Given the bull and bear cases and the data below, state the two biggest risks and a position limit you would impose, in at most 120 words."),
    ("trader", "Trader", "Given the bull, bear and risk-manager notes and the data below, propose ONE paper order for {company} ({ticker}) within a £1,000,000 long-only mandate: ticker, GBP size, and a two-sentence rationale. State the order as 'ORDER: BUY £<size> of {ticker}'."),
]


def committee_prompts(ctx):
    """Four sequential prompts sharing the tool outputs; later roles see earlier notes."""
    data = json.dumps(ctx, default=str, indent=0)
    prompts = {}
    for key, role, instr in COMMITTEE_ROLES:
        p = f"Today is {TODAY}. You are the {role} on a paper-trading investment committee. "
        p += instr.format(company=COMPANY[COMMITTEE_TICKER], ticker=COMMITTEE_TICKER)
        p += "\n\nDATA (from the Lab-3 tools):\n" + data
        prompts[key] = p
    return prompts


def committee_prompt(key, prompts, notes):
    """Later roles see the earlier notes (the chain is the same in the notebook and the builders)."""
    p = prompts[key]
    if key == "risk":
        p += f"\n\nBULL:\n{notes.get('bull', '')}\n\nBEAR:\n{notes.get('bear', '')}"
    if key == "trader":
        p += (f"\n\nBULL:\n{notes.get('bull', '')}\n\nBEAR:\n{notes.get('bear', '')}"
              f"\n\nRISK MANAGER:\n{notes.get('risk', '')}")
    return p


def run_committee_live(cell_prefix="4.2"):
    """🟡 Regenerate the transcript live (4 calls). Also used by build_cache.py."""
    ctx = committee_context()
    prompts = committee_prompts(ctx)
    notes = {}
    for key, role, _ in COMMITTEE_ROLES:
        p = committee_prompt(key, prompts, notes)
        res = ask(f"{cell_prefix}.{key}", p, temperature=0)
        notes[key] = res["text"] or "(no answer)"
        _p(f"▶ {role} ({res['source']}):")
        for line in textwrap.wrap(notes[key], TRACE_WIDTH - 2):
            _p("  " + line)
        _p("")
    return ctx, notes


def committee_evidence_rows(ctx):
    """The shared evidence as (row-number, tool call, returned value) triples.

    One row per Lab-3 tool output, each tagged with the tool call (brief §5) that
    produced it, so a reader can check any number in the transcript against the call
    that returned it. This is what the Analyst role is asked to do in committee.
    """
    t = COMMITTEE_TICKER
    rows = []
    p, h = ctx["price"], ctx["headlines"]
    rows.append((f'get_price("{t}", "{p["date"]}")', f'close_usd = {p["close_usd"]:.2f} USD (unadjusted close)'))
    for key, label in (("ret_1m", "one month"), ("ret_3m", "three months"), ("ret_12m", "twelve months")):
        r = ctx[key]
        rows.append((f'get_return("{t}", "{r["start"]}", "{r["end"]}")',
                     f'return_pct = {r["return_pct"]:+.2f} % ({label}, adjusted closes)'))
    v = ctx["vol_30d"]
    rows.append((f'get_volatility("{t}", {v["window_days"]})',
                 f'annualised_vol_pct = {v["annualised_vol_pct"]:.2f} % (as of {v["as_of"]})'))
    rows.append((f'get_headlines("{t}")',
                 f'{len(h["headlines"])} headline(s), date + text only, dated <= TODAY'))
    for i, item in enumerate(h["headlines"], 1):
        rows.append((f'  \u2514 headline {i} (from get_headlines)', f'{item["date"]} \u2014 {item["text"]}'))
    rows.append(("(constant) TODAY", f"{TODAY} \u2014 no tool may return anything dated after this"))
    rows.append(("(constant) mandate", f"\u00a3{MANDATE_GBP:,.0f} paper notional, long-only, exactly one place_order call"))
    return [(i + 1, a, b) for i, (a, b) in enumerate(rows)]


def committee_evidence_table(ctx):
    rows = committee_evidence_rows(ctx)
    w = max(len(a) for _, a, _ in rows)
    out = [f"{'#':>2}  {'TOOL CALL (brief §5)'.ljust(w)}  RETURNED",
           f"{'--':>2}  {'-' * w}  {'-' * 34}"]
    for n, a, b in rows:
        out.append(f"{n:>2}  {a.ljust(w)}  {b}")
    return "\n".join(out)


def committee_markdown(ctx, notes, generated_by, sample, appendix=None):
    """Transcript: evidence table (each row tagged to its tool call), four notes, appendix.

    `appendix` is an optional dict with keys "numbers" and "unsourced": lists of strings
    separating the numbers that came from a tool row above from the claims that did not.
    Both builders (build_cache.py, make_sample_cache.py) call this one function, so the
    long and short forms of the transcript cannot diverge.
    """
    md = f"# Investment committee transcript — {COMPANY[COMMITTEE_TICKER]} ({COMMITTEE_TICKER})\n\n"
    md += f"*Generated {datetime.now().strftime('%Y-%m-%d')} by {generated_by}; data as of {TODAY}.*  \n"
    if sample:
        md += "**SAMPLE — stand-in text written by Claude (Anthropic), not model output; regenerate with build/build_cache.py.**\n\n"
    md += ("## Shared evidence (Lab-3 tool outputs)\n\n"
           "Every agent below saw exactly this, and nothing else. Each row is tagged with "
           "the tool call that produced it.\n\n```\n" + committee_evidence_table(ctx) + "\n```\n\n")
    for key, role, _ in COMMITTEE_ROLES:
        md += f"## {role}\n\n{notes.get(key, '')}\n\n"
    if appendix:
        md += "## Appendix — check the transcript against the table\n\n"
        md += "### Numbers used (each one traces to a row above)\n\n"
        for line in appendix.get("numbers", []):
            md += f"- {line}\n"
        md += "\n### Claims made without tool support (nothing in the table returns these)\n\n"
        for line in appendix.get("unsourced", []):
            md += f"- {line}\n"
        md += "\n*For the Analyst role: the second list is the answer, and the reason the trace is printed.*\n"
    return md


def reset_all():
    """Reset cell: clears the lock, the order book, results and counters (instructor / demo use)."""
    global PREDICTIONS_LOCKED
    PREDICTIONS_LOCKED = None
    globals()["PREDICTIONS_LOCKED"] = None
    del ORDER_BOOK[:]
    S["results"] = {}
    S["n_requests"] = 0
    GUARD.update(MAX_ORDER_GBP=MANDATE_GBP, REQUIRE_HUMAN_APPROVAL=False)
    _say("reset: lock cleared, ORDER BOOK emptied, results and request counter zeroed.")


def run_instructor_unprompted(n=5):
    """🔴 Would it trade unprompted? n re-runs of 3.2 without the mandatory-order line."""
    if not require(locked(), "run 1.2 first (any dropdown values) — the gate applies to instructors too"):
        return
    saved = (dict(GUARD), [dict(o) for o in ORDER_BOOK])   # scratch runs: the Receipt is untouched
    try:
        for i in range(1, n + 1):
            _hr(f"🔴 unprompted run {i}/{n}")
            GUARD.update(MAX_ORDER_GBP=MANDATE_GBP, REQUIRE_HUMAN_APPROVAL=False)
            del ORDER_BOOK[:]
            res = ask(f"X.unprompted.d{i}", mandate_text(mandatory=False), temperature=1, tools=MANDATE_TOOLS)
            orders = [c for c in res["tool_calls"] if c["name"] == "place_order"]
            _p(f"orders placed: {len(orders)} · source: {res['source']}")
            for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
                _p("  " + line)
    finally:
        GUARD.clear()
        GUARD.update(saved[0])
        del ORDER_BOOK[:]
        ORDER_BOOK.extend(saved[1])

# ---- go -------------------------------------------------------------------
reconnect()
````

````python
#@title 🔁 Disconnected? Run this first (Lab 3) { display-mode: "form" }
FORCE_CACHED = False  #@param {type:"boolean"}
# Room-wide STUDIO MODE: tick the box above (or the instructor projects "FORCE_CACHED = True").
REPO_RAW = "https://raw.githubusercontent.com/thezeming/mgrcm0065-week3/main"   # class-repo base URL, baked in from w3_core.REPO_RAW_URL; empty = no fetch
_RECONNECT = True
# pinned SDK (guarded: warm runtimes skip the install)
try:
    from google import genai as _probe  # noqa
except ImportError:
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "google-genai==1.47.0"])
EMBEDDED_B64 = ""
# ---- shared core (verbatim copy of build/w3_core.py) ------------------------
# ============================================================================
#  MGRCM0065 Week 3 · "AI for Finance" — shared core
#  This file is (a) embedded verbatim into cell 0 of the student notebook and
#  (b) imported by build/build_cache.py, build/make_sample_cache.py and the
#  tests, so the notebook and the cache builder run IDENTICAL code.
#  Only the standard library + pandas + numpy + google-genai are imported.
# ============================================================================
import base64, gzip, hashlib, io, json, os, re, sys, textwrap, time, math
from datetime import datetime

import numpy as np
import pandas as pd

# ----------------------------------------------------------------------------
# Constants (brief §1). MODEL is a placeholder the instructor re-checks on
# build day and on 6 Oct at aistudio.google.com/rate-limit.
# ----------------------------------------------------------------------------
MODEL = "gemini-3.5-flash-lite"              # the configured choice; may 404 if Google withdraws it
# Chosen on 17 Sep 2026 from the instructor's own account catalogue (41 models). It is the
# newest non-preview "flash-lite" the key can see, which is the cheapest/fastest tier and so
# the one with the largest free daily allowance — what matters with ~100 students. The
# previous value, gemini-2.5-flash-lite, returned 404 from an older SDK on 17 Sep 2026.
# pick_model() independently selects this same model from that catalogue.
MODEL_FALLBACK = True                        # on a 404, ask the key what it CAN use and switch (loudly)
MODEL_IN_USE = MODEL                         # the model actually talking to us; build_cache records THIS
SDK_VERSION = "2.24.0"                       # google-genai version this code was written against
SDK_MIN_VERSION = "1.20.0"                   # oldest google-genai whose API surface we verified by hand
MODEL_EXCLUDE = ("pro", "ultra")             # never auto-pick an expensive tier
GROQ_MODEL = "openai/gpt-oss-20b"
TICKERS = ["AAPL", "MSFT", "NVDA", "TSLA", "AMZN", "META", "JPM", "BA"]
COMMITTEE_TICKER = "NVDA"
FLAT_BAND = 0.005                            # ±0.5 % is FLAT
PACE_SECONDS = 7                             # client-side spacing between live requests
TOOL_CAP = 3                                 # READ-ONLY tool calls per agent turn
ORDER_CAP = 2                                # place_order calls per turn (refused + retry smaller)
# The two budgets are deliberately separate. A single shared cap looks tidier but starves the
# mandate: the model researches all three tickers in ONE parallel batch, which spends three
# calls, and place_order then comes back "CAP REACHED" with an empty order book — observed
# against the live API on 17 Sep 2026. Lab 3 exists to show an order being filled and then
# refused, so the order must never be the thing the budget runs out on. ORDER_CAP is 2 because
# the 3.3 demonstration needs a refusal followed by a smaller retry, which is two calls.
MANDATE_GBP = 1_000_000                      # paper mandate, GBP notional only
# Raw-GitHub base URL of the class repo, e.g.
#   "https://raw.githubusercontent.com/<user>/<repo>/main"
# make_notebook.py bakes this into cell 0. Empty = the notebook never phones home and always uses
# its own embedded copy. Set it once here, then `python build/refresh.py --publish`.
REPO_RAW_URL = "https://raw.githubusercontent.com/thezeming/mgrcm0065-week3/main"
TRACE_WIDTH = 72
HALLUCINATION_DATE = "2023-06-30"            # inside any plausible training window; no splits since
HALLUCINATION_TICKERS = ["MSFT", "META", "JPM"]
COMPANY = {"AAPL": "Apple", "MSFT": "Microsoft", "NVDA": "Nvidia", "TSLA": "Tesla",
           "AMZN": "Amazon", "META": "Meta", "JPM": "JPMorgan", "BA": "Boeing"}

MODE_LIVE, MODE_STUDIO, MODE_ERROR = "LIVE MODE", "STUDIO MODE", "ERROR — hand up"
STUDIO_SUBLINE = ("every number, order and refusal is computed on your laptop; "
                  "the model's wording is the instructor's pre-run")
CACHED_BANNER = "[cached output — instructor's pre-run]"
SAMPLE_BANNER = "SAMPLE CACHE — regenerate with build/build_cache.py before class"

# Anonymised twins of the Lab-1 headlines (names, tickers, products and dates
# removed; written by the notebook builder for the 1.5b design, not by a model).
ANON_HEADLINES = {
    "H01": "Company X warns of flagship-product shipment delays after a Covid lockdown at a supplier's plant",
    "H02": "Company X announces multibillion-dollar investment in an AI start-up",
    "H03": "Company X guidance beats estimates on generative-AI demand",
    "H04": "Company X to cut over 18,000 jobs",
    "H05": "Company X posts first-ever drop in daily active users, issues weak guidance",
    "H06": "Company X acquires a failed regional bank after regulators seize it",
    "H07": "Regulator orders inspections, grounds Company X's aircraft type after an engine failure",
    "H08": "Company X confirmed to join a major stock index next month",
}

# ----------------------------------------------------------------------------
# Global state. Everything the cells share lives in S so that the 🔁 cells can
# rebuild it and the build scripts can inspect it.
# ----------------------------------------------------------------------------
_prev_S = globals().get("S") if isinstance(globals().get("S"), dict) else {}
S = {
    "mode": None, "provider": None, "client": None, "key_status": "unchecked",
    "n_requests": 0, "last_request": 0.0, "cache": {}, "cache_meta": {},
    "prices": None, "headlines": None, "fresh": None, "committee_md": None,
    "version": {}, "data_source": "none", "results": {}, "recording": None,
    "in_colab": False, "sdk_ok": False, "grounding_seen": False,
    "model": MODEL, "model_switched": False, "sdk_version": None,
}
# Re-running cell 0 or a 🔁 cell must NOT wipe the lock, the results or the order book.
S["results"] = dict(_prev_S.get("results", {}))
S["n_requests"] = int(_prev_S.get("n_requests", 0))
ORDER_BOOK = list(globals().get("ORDER_BOOK") or [])
GUARD = dict(globals().get("GUARD") or {"MAX_ORDER_GBP": MANDATE_GBP, "REQUIRE_HUMAN_APPROVAL": False})
PREDICTIONS_LOCKED = globals().get("PREDICTIONS_LOCKED", None)   # None or (timestamp_str, {id: call}, kind)
_BOOTSTRAPPED = bool(globals().get("_BOOTSTRAPPED", False))      # True once cell 0 / a 🔁 cell has run in THIS kernel
#   kind ∈ {"locked", "restored"}: "restored" is the placeholder a 🔁 cell sets when the runtime lost the
#   real lock — it opens Labs 2–4 but never scores, and 1.2 may overwrite it.
TODAY = None                   # last trading date in prices.csv, set by load_data()

# Values the notebook's cell 0 may define before this code runs.
FORCE_CACHED = bool(globals().get("FORCE_CACHED", False)) or os.environ.get("W3_FORCE_CACHED", "") == "1"
REPO_RAW = str(globals().get("REPO_RAW", "") or "")
EMBEDDED_B64 = globals().get("EMBEDDED_B64", None)
_RECONNECT = bool(globals().get("_RECONNECT", False))
QUIET = bool(globals().get("W3_QUIET", False))

genai = None
gtypes = None
gerrors = None


def _p(*a, **k):
    """print() that the build scripts can silence."""
    if not QUIET:
        print(*a, **k)


def _hr(title=""):
    line = "─" * TRACE_WIDTH
    _p(line if not title else f"── {title} " + "─" * max(0, TRACE_WIDTH - len(title) - 4))


def _countdown(seconds, label):
    """Every wait prints a countdown so nothing looks like a hang."""
    seconds = int(math.ceil(seconds))
    if seconds <= 0:
        return
    _p(f"{label}: ", end="", flush=True)
    for i in range(seconds, 0, -1):
        _p(f"{i}…", end=" ", flush=True)
        time.sleep(1)
    _p("go", flush=True)


def _trace(line):
    """Print a trace line wrapped so that no printed line exceeds 72 chars."""
    for chunk in textwrap.wrap(line, TRACE_WIDTH, subsequent_indent="      ", break_on_hyphens=False) or [""]:
        _p(chunk)


def _say(text, indent=""):
    """Print prose wrapped at 72 characters (plain lines, no hanging indent)."""
    for chunk in textwrap.wrap(text, TRACE_WIDTH, initial_indent=indent, subsequent_indent=indent,
                               break_on_hyphens=False) or [""]:
        _p(chunk)


def _sha(text):
    return hashlib.sha256(text.encode("utf-8")).hexdigest()


def _now():
    return datetime.now().strftime("%H:%M:%S")


# ----------------------------------------------------------------------------
# Data loading: embedded blob first, disk second (build scripts), GitHub third
# ----------------------------------------------------------------------------
def _materials_dir():
    here = os.path.dirname(os.path.abspath(globals().get("__file__", "./build/w3_core.py")))
    return os.path.abspath(os.path.join(here, ".."))


def _decode_blob(b64):
    return json.loads(gzip.decompress(base64.b64decode(b64)).decode("utf-8"))


def _read_disk_blob():
    root = _materials_dir()
    blob = {}
    for key, rel in [("prices_csv", "data/prices.csv"), ("headlines_csv", "data/headlines.csv"),
                     ("headlines_fresh_csv", "data/headlines_fresh.csv"),
                     ("responses_json", "cache/responses.json"),
                     ("committee_md", "cache/committee.md"), ("version_json", "VERSION.json")]:
        path = os.path.join(root, rel)
        blob[key] = open(path, encoding="utf-8").read() if os.path.exists(path) else None
    return blob


def _install_blob(blob, source):
    global TODAY
    if not blob.get("prices_csv") or not blob.get("headlines_csv"):
        raise ValueError("prices.csv / headlines.csv missing from the data bundle")
    prices = pd.read_csv(io.StringIO(blob["prices_csv"]))
    prices["date"] = pd.to_datetime(prices["date"]).dt.strftime("%Y-%m-%d")
    prices = prices.sort_values(["ticker", "date"]).reset_index(drop=True)
    heads = pd.read_csv(io.StringIO(blob["headlines_csv"]))
    fresh = pd.read_csv(io.StringIO(blob["headlines_fresh_csv"])) if blob.get("headlines_fresh_csv") else None
    for df in (heads, fresh):
        if df is not None:
            for c in ("date", "event_date"):
                df[c] = pd.to_datetime(df[c]).dt.strftime("%Y-%m-%d")
    cache = json.loads(blob["responses_json"]) if blob.get("responses_json") else {"_meta": {"sample": True, "empty": True}, "entries": {}}
    S.update(prices=prices, headlines=heads, fresh=fresh, data_source=source,
             cache=cache.get("entries", {}), cache_meta=cache.get("_meta", {}),
             committee_md=blob.get("committee_md"),
             version=json.loads(blob["version_json"]) if blob.get("version_json") else {})
    TODAY = str(prices["date"].max())
    globals()["TODAY"] = TODAY


def load_data():
    """Embedded copy first (always sufficient for Labs 1–4), disk copy for the build scripts."""
    if EMBEDDED_B64:
        _install_blob(_decode_blob(EMBEDDED_B64), "embedded copy inside this notebook")
    else:
        _install_blob(_read_disk_blob(), "materials/ folder on disk")


def _http_get(url, timeout=10):
    import urllib.request
    with urllib.request.urlopen(url, timeout=timeout) as r:
        return r.read().decode("utf-8")


REPO_FILES = [("prices_csv", "data/prices.csv"),
              ("headlines_csv", "data/headlines.csv"),
              ("headlines_fresh_csv", "data/headlines_fresh.csv"),
              ("responses_json", "cache/responses.json"),
              ("committee_md", "cache/committee.md")]


def bundle_last_trading_day(blob):
    prices = pd.read_csv(io.StringIO(blob["prices_csv"]))
    return str(pd.to_datetime(prices["date"]).dt.strftime("%Y-%m-%d").max())


def bundle_is_coherent(blob):
    """Do the prices and the recorded answers in this bundle belong to each other?

    A cache entry is keyed on cell_id + sha256(prompt), and the Lab 3 / Lab 4 prompts embed
    TODAY, which is the last trading day in prices.csv. So a bundle whose answers were recorded
    against a DIFFERENT day orphans every Lab 3 and Lab 4 entry: the cells still run, but a
    student without a working key gets "no cached answer" in exactly the two labs that carry the
    teaching — and only once their quota runs out, so it never shows up in testing.

    Checked before anything is installed. Returns (ok, reason).
    """
    try:
        last = bundle_last_trading_day(blob)
    except Exception as e:
        return False, f"prices.csv could not be read ({type(e).__name__})"
    try:
        meta = (json.loads(blob["responses_json"]) or {}).get("_meta", {})
    except Exception as e:
        return False, f"responses.json could not be read ({type(e).__name__})"
    recorded = str(meta.get("today_constant", "") or "")
    if not recorded:
        return False, "the recorded answers do not say which trading day they were built against"
    if recorded != last:
        return False, (f"prices run to {last} but the answers were recorded for {recorded} — "
                       "Lab 3 and Lab 4 would fall through to 'no cached answer'")
    return True, ""


def _warn_if_notebook_stale(ver, mine):
    if ver.get("notebook") != mine.get("notebook"):
        _say(f"⚠️  VERSION: this notebook is v{mine.get('notebook')}; current is v{ver.get('notebook')} — "
             "open the fresh notebook link and save a new copy (the code itself cannot self-update).")


def fetch_repo_updates():
    """Refresh data/ and cache/ from the class repo (skipped when REPO_RAW is empty).

    ALL OR NOTHING, deliberately. prices.csv and cache/responses.json are a matched pair — see
    bundle_is_coherent() — so a half-applied refresh is worse than no refresh at all. One failed
    download, or one mismatched pair, abandons the entire attempt and leaves the embedded copy
    untouched. The embedded copy is always internally consistent, because make_notebook.py built
    it from one snapshot.
    """
    if not REPO_RAW:
        _p("GitHub refresh: skipped (REPO_RAW is empty) — using the embedded copy.")
        return
    if os.environ.get("W3_NO_FETCH") == "1":
        # make_notebook.py captures the Expected-output boxes by EXECUTING each cell. Without this
        # the build would phone home four times per run, and its output would depend on the network
        # — which breaks the byte-identical-rebuild test.
        _p("GitHub refresh: (runs here — OK / already current / not reachable)")
        return
    base = REPO_RAW.rstrip("/")
    bust = f"?v={int(time.time())}"
    mine = dict(S["version"])
    try:
        ver = json.loads(_http_get(f"{base}/VERSION.json{bust}"))
    except Exception as e:
        _say(f"GitHub refresh: not reachable ({type(e).__name__}) — embedded copy is fine for today.")
        return
    if ver.get("data") == mine.get("data"):
        _say(f"GitHub refresh: already current (data v{mine.get('data')}) — nothing to download.")
        _warn_if_notebook_stale(ver, mine)
        return
    fetched = {}
    for key, rel in REPO_FILES:
        try:
            fetched[key] = _http_get(f"{base}/{rel}{bust}")
        except Exception as e:
            _say(f"GitHub refresh: {rel} did not download ({type(e).__name__}) — abandoning the whole "
                 "refresh rather than mixing new prices with older answers; embedded copy stands.")
            _warn_if_notebook_stale(ver, mine)
            return
    blob = dict(_read_disk_blob()) if not EMBEDDED_B64 else _decode_blob(EMBEDDED_B64)
    blob.update(fetched)
    blob["version_json"] = json.dumps(ver)
    ok, why = bundle_is_coherent(blob)
    if not ok:
        _say(f"GitHub refresh: REFUSED — {why}. Embedded copy stands.")
        _warn_if_notebook_stale(ver, mine)
        return
    try:
        _install_blob(blob, f"GitHub ({base})")
    except Exception as e:
        _say(f"GitHub refresh: bundle rejected ({type(e).__name__}) — embedded copy stands.")
        _warn_if_notebook_stale(ver, mine)
        return
    _say(f"GitHub refresh: OK — data v{mine.get('data')} \u2192 v{ver.get('data')}; prices to {TODAY}, "
         "answers recorded for the same day.")
    _warn_if_notebook_stale(ver, mine)


# ----------------------------------------------------------------------------
# SDK install / import (pinned; guarded so warm runtimes skip it)
# ----------------------------------------------------------------------------
def ensure_sdk():
    global genai, gtypes, gerrors
    try:
        from google import genai as _g
    except ImportError:
        _p(f"Installing google-genai=={SDK_VERSION} (about 20 seconds)…")
        os.system(f"{sys.executable} -m pip install -q google-genai=={SDK_VERSION}")
        try:
            from google import genai as _g
        except ImportError:
            _say("Runtime → Restart session, then run this cell again — normal, 20 seconds.")
            S["sdk_ok"] = False
            return False
    from google.genai import types as _t, errors as _e
    genai, gtypes, gerrors = _g, _t, _e
    ver = getattr(_g, "__version__", "?")
    S["sdk_version"] = ver
    if not sdk_version_ok(ver):
        S["sdk_ok"] = False
        _p(f"SDK: google-genai {ver} — TOO OLD")
        _say(f"This notebook needs google-genai {SDK_MIN_VERSION} or newer (it was written "
             f"against {SDK_VERSION}); you have {ver}, whose API is different enough to fail "
             "in the middle of a cell rather than here. Fix it with one line:")
        _p(f"  {sys.executable} -m pip install -U 'google-genai>={SDK_MIN_VERSION}'")
        _say("If pip says no matching distribution, the cause is your Python, not your "
             "network: modern google-genai needs Python 3.10 or newer, and 'python3' on a Mac "
             "is often the system 3.9. Check with  python3 -V  and rebuild the environment "
             "with an explicit modern interpreter, e.g. /opt/homebrew/bin/python3.12.")
        _say("STUDIO MODE meanwhile — every number, order and refusal still runs.")
        return False
    S["sdk_ok"] = True
    _p(f"SDK: google-genai {ver}")
    if ver != SDK_VERSION:
        _say(f"(written against {SDK_VERSION}; {SDK_MIN_VERSION} and newer are supported — "
             "the calls this notebook makes are identical on both.)")
    return True


def _ver_tuple(v):
    """('1.47.0' → (1, 47, 0)); anything unparseable sorts as (0,) so it fails the floor check."""
    parts = []
    for chunk in str(v).split(".")[:3]:
        digits = "".join(c for c in chunk if c.isdigit())
        if digits == "":
            break
        parts.append(int(digits))
    return tuple(parts) or (0,)


def sdk_version_ok(ver):
    """True when the installed google-genai is at least SDK_MIN_VERSION."""
    return _ver_tuple(ver) >= _ver_tuple(SDK_MIN_VERSION)


# ----------------------------------------------------------------------------
# Key discovery and the three-attempt verification
# ----------------------------------------------------------------------------
def find_key():
    """Returns (key, provider, status). Distinguishes no key / NotebookAccessError."""
    try:
        from google.colab import userdata  # noqa
        S["in_colab"] = True
    except ImportError:
        S["in_colab"] = False
        userdata = None
    for name, provider in (("GEMINI_API_KEY", "gemini"), ("GROQ_API_KEY", "groq")):
        if userdata is not None:
            try:
                val = userdata.get(name)
                if val:
                    return val.strip(), provider, "colab-secret"
            except Exception as e:  # SecretNotFoundError / NotebookAccessError
                if "NotebookAccess" in type(e).__name__:
                    return None, provider, "notebook-access"
        val = os.environ.get(name, "").strip()
        if val:
            return val, provider, "env"
    return None, None, "none"


# Remedies, in the order _classify_error tests for them. Each one has its own branch and its own
# sentence on screen, because telling someone to check their Wi-Fi when the model name is wrong
# wastes the whole room's time.
#   rate-limit      429 / RESOURCE_EXHAUSTED — the key works, the free quota is busy
#   model-not-found 404 / NOT_FOUND, or a 403 whose message names models/… — the MODEL name is
#                   not available to this key (deterministic: never retried, because waiting
#                   cannot make a withdrawn model come back)
#   bad-key         400/401/403 naming the key or a permission — the key itself is wrong
#   server          5xx at the provider
#   network         a transport-level failure: DNS, TCP, TLS, timeout
#   other           anything we have not met before; never described as a network problem
def _classify_error(e):
    """Map any exception from the SDK to exactly one remedy. Specific codes first; the
    name-sniffing network test last, so 'network' only ever means an actual network fault.

    Works for both SDKs: google-genai puts an int in .code, Groq/OpenAI-style SDKs put a
    string error code there and the HTTP status in .status_code, so we normalise first."""
    name = type(e).__name__
    code = getattr(e, "code", None)
    if not isinstance(code, int) or isinstance(code, bool):
        try:
            code = int(code)
        except (TypeError, ValueError):
            code = None
    if code is None:
        sc = getattr(e, "status_code", None)
        code = sc if isinstance(sc, int) and not isinstance(sc, bool) else None
    status = str(getattr(e, "status", "") or "")
    msg = str(e)
    up = msg.upper()
    if code == 429 or "RESOURCE_EXHAUSTED" in status or "429" in msg[:12] or "RateLimit" in name:
        return "rate-limit"
    if code == 404 or "NOT_FOUND" in status or "NOT_FOUND" in up[:200] or "NotFound" in name:
        return "model-not-found"
    # A 403 PERMISSION_DENIED that names a model ("Permission denied on resource
    # models/gemini-2.5-flash-lite") is the model being withdrawn from this key, not a bad
    # key — Google returns either code for it. Tested before the key branch on purpose.
    if code in (403, 404) and "MODELS/" in up and "API_KEY" not in up:
        return "model-not-found"
    if code in (400, 401, 403) and ("API_KEY" in up or "PERMISSION" in up or code in (401, 403)):
        return "bad-key"
    if code is None and any(w in name for w in ("Authentication", "PermissionDenied")):
        return "bad-key"
    if code is not None and 500 <= int(code) < 600:
        return "server"
    if any(w in name for w in ("Timeout", "Connect", "Network", "Remote", "Transport", "OSError", "URLError")):
        return "network"
    return "other"


def _key_console(provider):
    """Where to re-copy a rejected key from — the provider's own console, never Google's
    for a Groq key."""
    return "console.groq.com/keys" if provider == "groq" else "aistudio.google.com/apikey"


def model_from_error(e, default=None):
    """Pull the model name out of a 404 message ('models/gemini-2.5-flash-lite is not found')."""
    m = re.search(r"models/([A-Za-z0-9._\-]+)", str(e))
    # ".rstrip('.')" because a 403 states it as a sentence: "…on resource models/x-y-z."
    return m.group(1).rstrip(".") if m else (default if default is not None else MODEL_IN_USE)


# ----------------------------------------------------------------------------
# Model discovery. Google publishes no static free-tier table (design-notes/
# research_tooling.md), so a 404 on the configured MODEL is an expected failure,
# not an exotic one. list_text_models() asks the key what it can actually use;
# pick_model() chooses between them by a rule that is fixed and written down.
# ----------------------------------------------------------------------------
def list_text_models(client=None):
    """Every model this API key can call with generateContent, as plain names
    ('gemini-2.5-flash'). A listing entry that advertises no supported actions at all is
    skipped rather than kept. Raises the SDK's own exception so the caller can classify it."""
    client = client or S["client"]
    out = []
    for m in client.models.list(config={"query_base": True}):
        actions = list(getattr(m, "supported_actions", None) or [])
        if not any(a.lower() == "generatecontent" for a in actions):
            continue   # a model that advertises nothing is treated as unusable, not as a candidate
        name = (getattr(m, "name", "") or "").split("/")[-1]
        if name:
            out.append(name)
    return sorted(set(out))


def _version_tuple(n):
    """The version that belongs to the family name, as a fixed-width tuple (largest first).

    Only the dotted or whole numbers immediately after the first word count, and we stop at
    the first token that is not a number, so a date suffix ('-preview-11-2025') or a build
    number ('-001') is never read as a version. Fixed width so the comparison is positional:
    without the padding, ('gemini-3',) would sort ahead of ('gemini-3.5',) simply because the
    shorter tuple is a prefix of the longer one."""
    nums = []
    for tok in n.split("-")[1:]:
        if re.fullmatch(r"\d+(\.\d+)*", tok):
            nums.extend(int(x) for x in tok.split("."))
        else:
            break
    return tuple(-x for x in (nums + [0, 0, 0])[:3])


def _model_rank(name):
    """Sort key for pick_model. Smaller is better, and it is entirely determined by the
    name, so the same list always yields the same choice:
      1. family      flash-lite (0) before flash (1) before anything else (2);
      2. version     the numbers in the name, largest first, compared position by position
                     so 3.5 beats a bare 3 (3.6 beats 2.5);
      3. maturity    a plain name before a preview/experimental/dated one;
      4. simplicity  the shorter name, then alphabetical — a pure tie-break."""
    n = name.lower()
    family = 0 if "flash-lite" in n else (1 if "flash" in n else 2)
    nums = _version_tuple(n)
    preview = 1 if any(w in n for w in ("preview", "exp", "-rc", "latest")) else 0
    return (family, nums, preview, len(name), name)


def pick_model(names):
    """The best model to fall back to, or None. Never picks a pro/ultra tier: an expensive
    model chosen silently is a bill, not a rescue."""
    usable = [n for n in names if not any(w in n.lower() for w in MODEL_EXCLUDE)]
    return sorted(usable, key=_model_rank)[0] if usable else None


def use_model(name, reason=""):
    """Switch the model every later request uses, and say so in one line on screen."""
    global MODEL_IN_USE
    MODEL_IN_USE = name
    S["model"] = name
    S["model_switched"] = (name != MODEL)
    if S["model_switched"]:
        _say(f"MODEL {MODEL} is unavailable; using {name} instead{reason}")
    return name


def resolve_model(client=None, configured=None):
    """Called when the configured model 404s. Asks the key what it can use, announces the
    switch and returns the new name, or None if nothing usable came back."""
    configured = configured or MODEL
    if not MODEL_FALLBACK:
        _say(f"MODEL_FALLBACK is off, so {configured} was not replaced automatically.")
        return None
    try:
        names = list_text_models(client)
    except Exception as e:
        _say(f"could not ask the provider which models this key can use "
             f"({type(e).__name__}: {str(e)[:60]})")
        return None
    if not names:
        _say("the provider listed no text-generation models for this key.")
        return None
    pick = pick_model(names)
    if pick is None:
        _say("the only models this key can use are pro/ultra tiers; not switching to one "
             "automatically. Set MODEL in build/w3_core.py by hand if you want one.")
        return None
    return use_model(pick, f" ({len(names)} models offered to this key)")


def verify_key(key, provider):
    """One live request, up to three attempts with a printed backoff (5 s, 10 s) — but only
    a network fault is ever retried; every other outcome is decided on the first attempt.

    Returns exactly one of: ok · rate-limited · bad-key · model-not-found · network · server ·
    provider-error, for BOTH providers (the Groq arm classifies through the same
    _classify_error, so a Groq timeout is never reported as a rejected key). A 404 is NOT retried — it is deterministic, so a second attempt would
    waste thirty seconds to fail identically; instead we ask the key which models it can
    use and switch (MODEL_FALLBACK), which is the only thing that can actually fix it.
    """
    if provider == "groq":
        try:
            import groq  # noqa
        except ImportError:
            _p("Installing groq SDK (Groq key found, no Gemini key)…")
            os.system(f"{sys.executable} -m pip install -q groq")
        try:
            import groq
            S["client"] = groq.Groq(api_key=key)
            _pace()
            S["client"].chat.completions.create(model=GROQ_MODEL, max_tokens=5,
                                                messages=[{"role": "user", "content": "Reply with the single word OK."}])
            S["n_requests"] += 1
            return "ok"
        except Exception as e:
            kind = _classify_error(e)
            _say(f"Groq key check: {kind} ({type(e).__name__}: {str(e)[:90]})")
            # Same honesty rule as the Gemini arm: a Groq timeout is a network problem and a
            # Groq 429 is a busy quota — neither is a rejected key.
            if kind == "rate-limit":
                return "rate-limited"
            if kind in ("network", "server", "model-not-found", "bad-key"):
                return kind
            return "provider-error"
    S["client"] = genai.Client(api_key=key, http_options=gtypes.HttpOptions(timeout=30000))
    use_model(MODEL)                       # a re-run of cell 0 starts from the configured name again
    switched = False
    last = "network"
    for attempt in (1, 2, 3):
        try:
            _pace()
            resp = S["client"].models.generate_content(
                model=MODEL_IN_USE, contents="Reply with the single word OK.",
                config=gtypes.GenerateContentConfig(temperature=0, max_output_tokens=5))
            S["n_requests"] += 1
            _ = _safe_text(resp)
            return "ok"
        except Exception as e:
            kind = _classify_error(e)
            last = kind
            _say(f"key check attempt {attempt}/3: {kind} ({type(e).__name__}: {str(e)[:90]})")
            if kind == "rate-limit":
                return "rate-limited"          # the key works; quota is busy right now
            if kind == "bad-key":
                return "bad-key"
            if kind == "model-not-found":
                # Deterministic: do not back off and try the same name again.
                if switched or resolve_model(S["client"], model_from_error(e)) is None:
                    return "model-not-found"
                switched = True
                continue
            if attempt < 3:
                _countdown(5 * attempt, "  backing off")
    return last if last in ("network", "server") else "provider-error"


def _pace():
    """≥ 7 s between live requests, with a visible countdown."""
    wait = PACE_SECONDS - (time.time() - S["last_request"])
    if wait > 0:
        _countdown(wait, "free-tier pacing")
    S["last_request"] = time.time()


# ----------------------------------------------------------------------------
# Cell 0 — bootstrap
# ----------------------------------------------------------------------------
def bootstrap(reconnect=False):
    t0 = time.time()
    _hr("MGRCM0065 Week 3 · AI for Finance — START HERE" if not reconnect else "🔁 reconnect")
    try:
        load_data()
    except Exception as e:
        S["mode"] = MODE_ERROR
        _p(f"Data error: {type(e).__name__}: {e}")
        _p(f"=== {MODE_ERROR} ===")
        _say("data failed to load — watch the projector; we fix this at the break")
        return
    ver = S["version"]
    cache_kind = "SAMPLE" if S["cache_meta"].get("sample", True) else "real"
    _p(f"Notebook v{ver.get('notebook', '?')} · data v{ver.get('data', '?')} · cache: {cache_kind}")
    _p(f"prices to {TODAY} · {len(S['headlines'])} headlines + "
       f"{0 if S['fresh'] is None else len(S['fresh'])} fresh")
    _p(f"Data source: {S['data_source']}")
    _recorded = str(S["cache_meta"].get("today_constant", "") or "")
    if _recorded and _recorded != TODAY:
        _p(f"⚠️  recorded answers are for {_recorded} but prices run to {TODAY} — Labs 3 and 4 "
           "would fall through to 'no cached answer'; tell the instructor.")
    if not reconnect:
        fetch_repo_updates()
    sdk = ensure_sdk()
    if not sdk:
        S["mode"] = MODE_STUDIO
    key, provider, status = find_key()
    S["provider"] = provider
    S["key_status"] = status
    if FORCE_CACHED:
        S["mode"] = MODE_STUDIO
        _say("FORCE_CACHED is on → STUDIO MODE regardless of your key.")
    elif not sdk:
        pass
    elif status == "notebook-access":
        S["mode"] = MODE_STUDIO
        _say("Key present but THIS copy cannot see it (NotebookAccessError): click 🔑 in the left "
             "sidebar, toggle 'Notebook access' ON for GEMINI_API_KEY, then re-run this cell.")
    elif key is None:
        if S["in_colab"] and not reconnect:
            try:
                from getpass import getpass
                pasted = getpass("No key found in Secrets. Paste your GEMINI_API_KEY now, or press Enter for STUDIO MODE: ")
                if pasted.strip():
                    key, provider, status = pasted.strip(), "gemini", "pasted"
                    S["provider"] = provider
            except Exception:
                pass
        if key is None:
            S["mode"] = MODE_STUDIO
            _say("No key found (Secrets: GEMINI_API_KEY / GROQ_API_KEY; or the environment). "
                 "STUDIO MODE is fine — see the sub-line below.")
    if key is not None and S["mode"] is None:
        result = verify_key(key, provider)
        if result in ("ok", "rate-limited"):
            S["mode"] = MODE_LIVE
            if result == "rate-limited":
                _say("Key works but the free-tier quota is busy right now — cached twins will fill any gap.")
        elif result == "bad-key":
            S["mode"] = MODE_STUDIO
            _say(f"Key rejected by the provider: re-copy it from {_key_console(provider)} (no spaces), "
                 "save the Secret, re-run this cell. STUDIO MODE meanwhile.")
        elif result == "model-not-found":
            S["mode"] = MODE_STUDIO
            _say(f"The model {GROQ_MODEL if provider == 'groq' else MODEL} is not available to this "
                 "API key. This usually means the provider has renamed or withdrawn it. Your key and "
                 "your Wi-Fi are both fine. Tell the instructor, who runs build/list_models.py to see "
                 "what the key can use and then sets MODEL in build/w3_core.py. STUDIO MODE "
                 "meanwhile — every number, order and refusal in today's session still runs.")
        elif result == "server":
            S["mode"] = MODE_STUDIO
            _say("The provider's own servers returned an error three times (nothing to do with "
                 "your key or your Wi-Fi). Wait a minute, then re-run this cell — that is the "
                 "retry button. STUDIO MODE meanwhile.")
        elif result == "network":
            S["mode"] = MODE_STUDIO
            _say("Network timed out three times: check Wi-Fi, then re-run this cell (that is the retry "
                 "button). STUDIO MODE meanwhile.")
        else:
            S["mode"] = MODE_STUDIO
            _say("The provider returned an error we do not have a specific remedy for — the exact "
                 "wording is in the attempt lines above; show them to the instructor. Re-running "
                 "this cell is the retry button. STUDIO MODE meanwhile.")
    _hr()
    _p(f"=== {S['mode']} ===")
    if S["mode"] == MODE_STUDIO:
        _say(STUDIO_SUBLINE)
        meta = S["cache_meta"]
        if meta.get("sample", True):
            _p(f"⚠️  {SAMPLE_BANNER}")
            _say(f"cached wording is stand-in text; the real cache will be produced with: "
                 f"{meta.get('model_placeholder', MODEL)}")
        else:
            _say(f"cached answers were produced with: {meta.get('model', MODEL)} "
                 f"(instructor's pre-run on {meta.get('generated_on', '?')})")
    if S["mode"] == MODE_LIVE:
        if provider == "groq":
            _say(f"provider: Groq · model {GROQ_MODEL} · single-turn cells live; Lab 3 replays the "
                 "instructor's trace")
        else:
            _say(f"model: {MODEL_IN_USE} · pacing ≥ {PACE_SECONDS} s between requests · "
                 f"read-only tool-call cap {TOOL_CAP} per turn (+{ORDER_CAP} orders)")
    _p("Runtime type: CPU (no GPU needed)")
    _say("Google account: Colab, the Secret and the key must all be the same PERSONAL account "
         "(check the avatar, top right).")
    _p("Do NOT use Runtime → Run all — this notebook is a sequence of reveals.")
    _p(f"(cell 0 finished in {time.time() - t0:.1f} s at {_now()})")
    globals()["_BOOTSTRAPPED"] = True


def reconnect():
    """🔁 Disconnected? Run this first. Idempotent; keeps a real lock if it survived.

    If the runtime lost the lock, a PLACEHOLDER lock is set so Labs 2–4 open. The placeholder is
    marked "restored": 1.4 never scores it (it would read 0/8) and 1.2 is allowed to overwrite it,
    so a student who wants the scoreboard back runs 1.2 → 1.3 → 1.4 once more.
    """
    global PREDICTIONS_LOCKED
    was_alive = bool(globals().get("_BOOTSTRAPPED", False))   # read BEFORE bootstrap() sets it
    bootstrap(reconnect=True)
    if PREDICTIONS_LOCKED is None:
        PREDICTIONS_LOCKED = (_now(), {}, "restored")
        globals()["PREDICTIONS_LOCKED"] = PREDICTIONS_LOCKED
        if was_alive:
            # Cell 0 already ran in this kernel: nothing was lost — the student simply never ran 1.2.
            _say("You have not locked any predictions yet — run 1.2 (then 1.3 and 1.4) if you want "
                 "the scoreboard; Labs 2–4 are open either way, on your honour.")
        else:
            _say("Your prediction lock was lost with the runtime — Labs 2–4 are open on your honour; "
                 "the calls you noted down are the record. Want the scoreboard back? Run 1.2 (set the "
                 "dropdowns to what you wrote), then 1.3 and 1.4 — the notebook allows this once.")
    _p("🔁 reconnected — carry on from where you were.")


# ----------------------------------------------------------------------------
# Gating
# ----------------------------------------------------------------------------
def require(condition, message):
    if not condition:
        _p(f"⛔ {message}")
        return False
    return True


def locked():
    """True for a real lock AND for the reconnect placeholder (both open Labs 2–4)."""
    return globals().get("PREDICTIONS_LOCKED") is not None


def lock_is_placeholder():
    """True only for the placeholder a 🔁 cell sets after the runtime lost the real lock."""
    lk = globals().get("PREDICTIONS_LOCKED")
    return lk is not None and len(lk) >= 3 and lk[2] == "restored"


# ----------------------------------------------------------------------------
# Tools (Lab 3). All slice by TODAY. GBP is notional only (no FX applied).
# ----------------------------------------------------------------------------
def _series(ticker):
    if ticker not in TICKERS:
        raise ValueError(f"unknown ticker {ticker!r}; choose from {', '.join(TICKERS)}")
    df = S["prices"][S["prices"]["ticker"] == ticker]
    return df[df["date"] <= TODAY].set_index("date")


def _prior_trading_day(s, date):
    date = pd.to_datetime(date).strftime("%Y-%m-%d")
    if date > TODAY:
        raise ValueError(f"{date} is after TODAY ({TODAY}) — the future is not available")
    idx = s.index[s.index <= date]
    if len(idx) == 0:
        raise ValueError(f"no prices before {date}")
    return idx[-1], (idx[-1] != date)


def get_price(ticker, date):
    """Closing price in USD on `date` (or the nearest PRIOR trading day)."""
    s = _series(str(ticker).upper())
    d, moved = _prior_trading_day(s, date)
    out = {"ticker": ticker.upper(), "date": d, "close_usd": round(float(s.loc[d, "close"]), 2)}
    if moved:
        out["note"] = f"{date} was not a trading day; nearest prior trading day {d} used"
    return out


def get_return(ticker, start, end):
    """Total return (adjusted closes) from `start` to `end`, both nearest-prior-trading-day."""
    s = _series(str(ticker).upper())
    d0, _ = _prior_trading_day(s, start)
    d1, _ = _prior_trading_day(s, end)
    r = float(s.loc[d1, "adj_close"] / s.loc[d0, "adj_close"] - 1)
    return {"ticker": ticker.upper(), "start": d0, "end": d1, "return_pct": round(100 * r, 2)}


def get_volatility(ticker, window_days=30):
    """Annualised volatility (%) of daily log returns over the last `window_days` trading days."""
    s = _series(str(ticker).upper())["adj_close"]
    w = int(window_days)
    lr = np.log(s).diff().dropna().tail(w)
    return {"ticker": ticker.upper(), "window_days": w, "as_of": TODAY,
            "annualised_vol_pct": round(float(lr.std() * np.sqrt(252) * 100), 2)}


def get_headlines(ticker):
    """Headlines for `ticker` dated ≤ TODAY: date + text only (no returns, no labels)."""
    t = str(ticker).upper()
    rows = [S["headlines"]] + ([S["fresh"]] if S["fresh"] is not None else [])
    df = pd.concat(rows, ignore_index=True)
    df = df[(df["ticker"] == t) & (df["date"] <= TODAY)].sort_values("date")
    return {"ticker": t, "headlines": [{"date": r.date, "text": r.headline} for r in df.itertuples()]}


def guard_check(ticker, gbp_notional, side):
    """The governance guard lives HERE, not in the mandate text."""
    side = str(side).upper()
    if str(ticker).upper() not in TICKERS:
        return f"REFUSED: unknown ticker {ticker}"
    if side != "BUY":
        return f"REFUSED: long-only mandate — {side} orders are not permitted"
    if float(gbp_notional) > GUARD["MAX_ORDER_GBP"]:
        return f"REFUSED: £{float(gbp_notional):,.0f} exceeds MAX_ORDER_GBP £{GUARD['MAX_ORDER_GBP']:,.0f}"
    if GUARD["REQUIRE_HUMAN_APPROVAL"]:
        return "REFUSED: REQUIRE_HUMAN_APPROVAL is on — no human has approved this order"
    return "FILLED"


def place_order(ticker, gbp_notional, side="BUY"):
    """Paper order (GBP notional, long-only). Every call is written to ORDER_BOOK."""
    status = guard_check(ticker, gbp_notional, side)
    rec = {"time": _now(), "ticker": str(ticker).upper(), "side": str(side).upper(),
           "gbp_notional": float(gbp_notional), "status": status}
    if status == "FILLED":
        rec["price_usd"] = get_price(ticker, TODAY)["close_usd"]
    ORDER_BOOK.append(rec)
    return {"status": status, "ticker": rec["ticker"], "gbp_notional": rec["gbp_notional"], "side": rec["side"]}


TOOLS = {"get_price": get_price, "get_return": get_return, "get_volatility": get_volatility,
         "get_headlines": get_headlines, "place_order": place_order}
READ_ONLY_TOOLS = ["get_price", "get_return", "get_volatility", "get_headlines"]
MANDATE_TOOLS = READ_ONLY_TOOLS + ["place_order"]

_TOOL_DECLS = {
    "get_price": ("Closing share price in USD of a ticker on a date (YYYY-MM-DD); uses the nearest prior trading day.",
                  {"ticker": ("STRING", "one of " + ", ".join(TICKERS)), "date": ("STRING", "YYYY-MM-DD, not after today")},
                  ["ticker", "date"]),
    "get_return": ("Total return in percent between two dates (YYYY-MM-DD) using adjusted closes.",
                   {"ticker": ("STRING", "ticker"), "start": ("STRING", "YYYY-MM-DD"), "end": ("STRING", "YYYY-MM-DD")},
                   ["ticker", "start", "end"]),
    "get_volatility": ("Annualised volatility in percent over the last N trading days.",
                       {"ticker": ("STRING", "ticker"), "window_days": ("INTEGER", "number of trading days, e.g. 30")},
                       ["ticker", "window_days"]),
    "get_headlines": ("Recent news headlines for a ticker (date and text only).",
                      {"ticker": ("STRING", "ticker")}, ["ticker"]),
    "place_order": ("Place a paper (simulated) order in GBP notional. Long-only: side must be BUY.",
                    {"ticker": ("STRING", "ticker"), "gbp_notional": ("NUMBER", "order size in GBP"),
                     "side": ("STRING", "BUY")}, ["ticker", "gbp_notional", "side"]),
}


def tool_declarations(names):
    """google-genai function declarations (verified against types.py 2.24.0)."""
    decls = []
    for n in names:
        desc, props, req = _TOOL_DECLS[n]
        decls.append(gtypes.FunctionDeclaration(
            name=n, description=desc,
            parameters=gtypes.Schema(
                type="OBJECT",
                properties={k: gtypes.Schema(type=t, description=d) for k, (t, d) in props.items()},
                required=req)))
    return [gtypes.Tool(function_declarations=decls)]


# Models invent argument names. The live run on 17 Sep 2026 called
# place_order(size=200000, ticker='AAPL') — 'size' for 'gbp_notional', and no 'side' at all.
# run_tool would answer with a TypeError message, which costs a turn and, worse, leaves the
# recorded step with no status so run_3_2 prints "wants to buy £0". We canonicalise the names
# BEFORE the call so both the execution and the recorded script use the declared ones.
_ARG_ALIASES = {
    "place_order": {"size": "gbp_notional", "size_gbp": "gbp_notional", "gbp": "gbp_notional",
                    "gbp_size": "gbp_notional", "gbp_amount": "gbp_notional", "amount": "gbp_notional",
                    "notional": "gbp_notional", "order_size": "gbp_notional", "quantity": "gbp_notional",
                    "value": "gbp_notional", "symbol": "ticker", "stock": "ticker",
                    "direction": "side", "action": "side"},
    "get_price": {"symbol": "ticker", "stock": "ticker", "as_of": "date", "day": "date"},
    "get_return": {"symbol": "ticker", "stock": "ticker", "from": "start", "to": "end",
                   "start_date": "start", "end_date": "end"},
    "get_volatility": {"symbol": "ticker", "stock": "ticker", "window": "window_days",
                       "days": "window_days", "lookback": "window_days"},
    "get_headlines": {"symbol": "ticker", "stock": "ticker"},
}


def _clean_number(v):
    """'£200,000' / '200000 GBP' -> 200000.0; anything else is handed back untouched."""
    if isinstance(v, str):
        t = re.sub(r"[£$,\s]|gbp|usd", "", v, flags=re.I)
        try:
            return float(t)
        except ValueError:
            return v
    return v


def normalise_args(name, args):
    """Map the argument names a model actually produces onto the ones we declared."""
    alias = _ARG_ALIASES.get(name, {})
    out = {}
    for k, v in (args or {}).items():
        out[alias.get(str(k).lower(), k)] = v
    if name == "place_order":
        if "gbp_notional" in out:
            out["gbp_notional"] = _clean_number(out["gbp_notional"])
        out.setdefault("side", "BUY")
    return out


def run_tool(name, args):
    if name not in TOOLS:
        return {"error": f"unknown tool {name}"}
    try:
        return TOOLS[name](**args)
    except TypeError as e:
        return {"error": f"bad arguments for {name}: {e}"}
    except Exception as e:
        return {"error": f"{type(e).__name__}: {e}"}


def _fmt_args(args):
    return ", ".join(f"{k}={v!r}" for k, v in args.items())


def _print_tool_call(name, args, result):
    res = json.dumps(result, default=str, ensure_ascii=False)
    if len(res) > 400:
        res = res[:400] + " …"
    _trace(f"🔧 TOOL CALL {name}({_fmt_args(args)}) → {res}")


# ----------------------------------------------------------------------------
# Safe text extraction and the three-tier parser
# ----------------------------------------------------------------------------
def _safe_text(resp):
    """Text from the first candidate; None if empty/blocked. Never raises."""
    try:
        cands = resp.candidates or []
        if not cands or cands[0].content is None or not cands[0].content.parts:
            return None
        gm = getattr(cands[0], "grounding_metadata", None)
        if gm is not None and (getattr(gm, "web_search_queries", None) or getattr(gm, "grounding_chunks", None)):
            S["grounding_seen"] = True
        parts = [p.text for p in cands[0].content.parts if getattr(p, "text", None) and not getattr(p, "thought", False)]
        text = "\n".join(parts).strip()
        return text or None
    except Exception:
        return None


CALL_SCHEMA = {
    "type": "ARRAY",
    "items": {"type": "OBJECT",
              "properties": {"id": {"type": "STRING"},
                             "call": {"type": "STRING", "enum": ["UP", "DOWN", "FLAT"]},
                             "reason": {"type": "STRING"}},
              "required": ["id", "call", "reason"]},
}


def parse_calls(text, ids):
    """Tier 1 JSON → tier 2 regex → tier 3 UNPARSED. Returns {id: (call, reason)}. Never raises."""
    out = {}
    if text:
        try:  # tier 1
            data = json.loads(text)
            if isinstance(data, dict):
                data = data.get("calls") or data.get("items") or list(data.values())
            for row in data:
                i = str(row.get("id", "")).strip().upper()
                c = str(row.get("call", "")).strip().upper()
                if i in ids and c in ("UP", "DOWN", "FLAT"):
                    out[i] = (c, str(row.get("reason", "")).strip())
        except Exception:
            pass
        if len(out) < len(ids):  # tier 2
            for i in ids:
                if i in out:
                    continue
                m = re.search(rf"{i}\b[^A-Za-z]{{0,20}}(UP|DOWN|FLAT)\b", text, re.I)
                if m:
                    out[i] = (m.group(1).upper(), "(parsed from free text)")
    for i in ids:  # tier 3
        if i not in out:
            out[i] = ("UNPARSED", (text or "(no text)")[:80])
    return out


# ----------------------------------------------------------------------------
# ask(): the only path to a model
# ----------------------------------------------------------------------------
def _cache_key(cell_id, prompt):
    return f"{cell_id}:{_sha(prompt)}"


def _cached_entry(cell_id, prompt):
    return S["cache"].get(_cache_key(cell_id, prompt))


def _record(cell_id, prompt, text, tool_script=None, extra=None):
    """build_cache.py switches recording on; live results are collected here."""
    if S["recording"] is None:
        return
    key = _cache_key(cell_id, prompt)
    entry = S["recording"].setdefault(key, {"cell_id": cell_id, "prompt": prompt, "text": None,
                                            "tool_script": [], "variants": {}})
    if text is not None and entry["text"] is None:
        entry["text"] = text
    if tool_script:
        entry["tool_script"] = tool_script
    if extra:
        for k, v in extra.items():
            if isinstance(v, dict):
                entry.setdefault(k, {}).update(v)
            else:
                entry[k] = v


def _replay(entry, tools):
    """Replay a stored tool script through the REAL tool functions (200 ms per step)."""
    script = entry.get("tool_script") or []
    alt = entry.get("tool_script_refusal") or []
    # The refusal script is swapped in only when the fill script's first order would be refused
    # under the student's guard AND the refusal script itself places an order — otherwise the fill
    # script replays and the guard refuses it, so the book and the closing text still agree.
    if alt and any(step["name"] == "place_order" for step in alt) \
            and any(step["name"] == "place_order" for step in script):
        first = next(step for step in script if step["name"] == "place_order")
        if guard_check(first["args"].get("ticker"), first["args"].get("gbp_notional", 0),
                       first["args"].get("side", "BUY")) != "FILLED":
            script = alt
    executed = []
    for step in script:
        time.sleep(0.2)
        result = run_tool(step["name"], dict(step["args"]))
        _print_tool_call(step["name"], step["args"], result)
        executed.append({"name": step["name"], "args": step["args"], "result": result})
    text = entry.get("text")
    orders = [e for e in executed if e["name"] == "place_order"]
    variants = entry.get("variants") or {}
    if orders:
        statuses = [o["result"].get("status", "") for o in orders]
        if all(st == "FILLED" for st in statuses):
            text = variants.get("after_fill") or text
        elif statuses[-1] == "FILLED":
            text = variants.get("after_partial") or (
                "(no cached closing text for this outcome — the instructor's pre-run recorded a fill "
                "and a refusal but not a refused-then-filled turn; the ORDER BOOK above is the record)")
        else:
            text = variants.get("after_refusal") or text
    return text, executed


def _groq_generate(prompt, temperature, persona, schema):
    msgs = ([{"role": "system", "content": persona}] if persona else []) + [{"role": "user", "content": prompt}]
    kw = {"model": GROQ_MODEL, "messages": msgs, "temperature": float(temperature), "max_tokens": 1200}
    if schema:
        kw["response_format"] = {"type": "json_object"}
        msgs[-1]["content"] += "\nReturn JSON: {\"calls\": [{\"id\": …, \"call\": UP|DOWN|FLAT, \"reason\": …}]}"
    _pace()
    r = S["client"].chat.completions.create(**kw)
    S["n_requests"] += 1
    return r.choices[0].message.content


def _gemini_generate(cell_id, prompt, temperature, persona, schema, tools):
    """One turn, including our OWN tool loop (automatic function calling disabled)."""
    cfg = gtypes.GenerateContentConfig(
        temperature=float(temperature),
        system_instruction=persona if persona else None,
        automatic_function_calling=gtypes.AutomaticFunctionCallingConfig(disable=True))
    if schema:
        cfg.response_mime_type = "application/json"
        cfg.response_schema = schema
    if tools:
        cfg.tools = tool_declarations(tools)
        _say(f"tool-call cap applied: {TOOL_CAP} read-only calls per turn"
             + (f" plus up to {ORDER_CAP} orders" if any(t not in READ_ONLY_TOOLS for t in tools) else "")
             + " (automatic function calling OFF; loop is ours)")
    contents = [gtypes.Content(role="user", parts=[gtypes.Part.from_text(text=prompt)])]
    tool_script, n_reads, n_orders = [], 0, 0
    for _round in range(TOOL_CAP + ORDER_CAP + 2):
        _pace()
        resp = S["client"].models.generate_content(model=MODEL_IN_USE, contents=contents, config=cfg)
        S["n_requests"] += 1
        _p(f"   [request #{S['n_requests']} · {_now()}]")
        cand = (resp.candidates or [None])[0]
        parts = list(cand.content.parts) if (cand is not None and cand.content is not None and cand.content.parts) else []
        calls = [p.function_call for p in parts if getattr(p, "function_call", None)]
        if not calls:
            return _safe_text(resp), tool_script
        contents.append(cand.content)
        responses = []
        for fc in calls:
            args = normalise_args(fc.name, dict(fc.args or {}))
            is_order = fc.name not in READ_ONLY_TOOLS
            spent = n_orders if is_order else n_reads
            cap = ORDER_CAP if is_order else TOOL_CAP
            if spent >= cap:
                what = "order cap" if is_order else "read-only tool-call cap"
                result = {"error": f"{what} of {cap} reached — answer with what you have"}
                _trace(f"🔧 TOOL CALL {fc.name}({_fmt_args(args)}) → CAP REACHED ({cap})")
            else:
                result = run_tool(fc.name, args)
                if is_order:
                    n_orders += 1
                else:
                    n_reads += 1
                _print_tool_call(fc.name, args, result)
                tool_script.append({"name": fc.name, "args": args, "result": result})
            responses.append(gtypes.Part.from_function_response(name=fc.name, response={"result": result}))
        contents.append(gtypes.Content(role="user", parts=responses))
        # Silence function calling only once every tool THIS turn was given has run out of budget.
        # Gating it on the read budget alone (as it was) forbade the model from ever emitting the
        # order, whatever the order cap said; gating it on both counters regardless of which tools
        # were offered would leave a read-only turn burning rounds on calls that can only be refused.
        reads_left = n_reads < TOOL_CAP and any(t in READ_ONLY_TOOLS for t in tools)
        orders_left = n_orders < ORDER_CAP and any(t not in READ_ONLY_TOOLS for t in tools)
        if not reads_left and not orders_left:
            cfg.tool_config = gtypes.ToolConfig(function_calling_config=gtypes.FunctionCallingConfig(mode="NONE"))
    return None, tool_script


def ask(cell_id, prompt, temperature=0, tools=None, persona=None, schema=None):
    """Returns {'text', 'source' ∈ live|cached|miss, 'tool_calls', 'sample'}. Never raises for cache paths."""
    entry = _cached_entry(cell_id, prompt)
    sample = bool(S["cache_meta"].get("sample", True))
    live = (S["mode"] == MODE_LIVE) and not FORCE_CACHED
    if live and S["provider"] == "groq" and tools:
        _say("Groq provider: Lab 3 replays the instructor's trace (Groq path is single-turn only).")
        live = False
    book_before = list(ORDER_BOOK)
    if live:
        attempt, switched = 0, False
        while attempt < 2:
            attempt += 1
            # Rewind the book before EVERY attempt. The retry re-runs the whole tool loop, so a
            # 5xx arriving after a fill would otherwise book the same order twice and put two
            # £1,000,000 fills on the projector against a £1,000,000 mandate.
            del ORDER_BOOK[:]
            ORDER_BOOK.extend(book_before)
            try:
                if S["provider"] == "groq":
                    text, script = _groq_generate(prompt, temperature, persona, schema), []
                else:
                    text, script = _gemini_generate(cell_id, prompt, temperature, persona, schema, tools)
                if text is None:
                    _p("empty/blocked response from the model → serving the cached twin")
                    break
                _record(cell_id, prompt, text, script)
                return {"text": text, "source": "live", "tool_calls": script, "sample": False}
            except Exception as e:
                kind = _classify_error(e)
                if kind == "rate-limit":
                    _p(f"429 / RESOURCE_EXHAUSTED (free-tier limit) → {CACHED_BANNER}")
                    break
                if kind == "model-not-found":
                    # The model went away mid-session. Retrying the same name cannot help,
                    # so either switch to one this key can use or serve the cached twin.
                    # A successful switch buys one extra pass (on the NEW name) even if the
                    # 404 arrived on the second attempt; `switched` makes that a one-shot,
                    # so this cannot loop.
                    if S["provider"] == "gemini" and not switched and resolve_model(
                            S["client"], model_from_error(e)) is not None:
                        switched = True
                        attempt -= 1
                        continue
                    _say(f"the model {MODEL_IN_USE} is not available to this key (this is not a "
                         f"network problem) → {CACHED_BANNER}")
                    break
                _p(f"model error ({type(e).__name__}: {str(e)[:100]})")
                if attempt == 1:
                    _countdown(30, "one retry after")
                else:
                    _p(f"second failure → {CACHED_BANNER}")
        del ORDER_BOOK[:]
        ORDER_BOOK.extend(book_before)
    if entry is None:
        _say("no cached answer for your edited prompt — reset to default or use a live laptop.")
        return {"text": None, "source": "miss", "tool_calls": [], "sample": sample}
    _p(CACHED_BANNER)
    if sample:
        _p("   (SAMPLE — stand-in text, not model output)")
    if tools:
        text, executed = _replay(entry, tools)
    else:
        text, executed = entry.get("text"), []
    return {"text": text, "source": "cached", "tool_calls": executed, "sample": sample}


# ----------------------------------------------------------------------------
# Prompts (shared by the notebook and the cache builders — the hash is the key)
# ----------------------------------------------------------------------------
def flat_rule():
    return (f"FLAT means the return over the first trading session that could react to the headline "
            f"is between -{FLAT_BAND*100:.1f}% and +{FLAT_BAND*100:.1f}%.")


def prompt_lab1(rows, anonymised=False):
    lines = []
    for r in rows.itertuples():
        if anonymised:
            lines.append(f"{r.id}: {ANON_HEADLINES.get(r.id, r.headline)}")
        else:
            lines.append(f"{r.id} ({r.date}, {COMPANY.get(r.ticker, r.ticker)}, {r.ticker}): {r.headline}")
    head = ("You are a financial expert with stock recommendation experience. For each headline below, "
            "decide whether the company's share price went UP, DOWN or stayed FLAT over the first "
            f"trading session that could react to it. {flat_rule()} "
            "Use only what the headline says. Give a one-sentence reason per headline. "
            "Return a JSON array of objects with keys id, call (UP, DOWN or FLAT) and reason.")
    if anonymised:
        head += " The company names, tickers and dates have been removed on purpose."
    return head + "\n\n" + "\n".join(lines)


def prompt_recall(row):
    return (f"Here is a news headline published on {row['date']} about {COMPANY.get(row['ticker'], row['ticker'])} "
            f"({row['ticker']}): \"{row['headline']}\".\n"
            "In one paragraph of at most five sentences, describe what happened to the company's share price and "
            "to the company in the weeks and months that followed this headline.")


def prompt_hallucination():
    names = ", ".join(f"{COMPANY[t]} ({t})" for t in HALLUCINATION_TICKERS)
    return (f"Give your best estimate of the closing share price in US dollars of {names} on {HALLUCINATION_DATE}. "
            "Answer with exactly three lines in the form 'TICKER: price', for example 'XYZ: 123.45'. "
            "Do not add disclaimers, ranges or caveats — a single best estimate per line.")


PERSONAS = {
    "risk": "You are a cautious risk officer at a UK pension fund. You answer in three sentences.",
    "pm": "You are a bullish hedge-fund portfolio manager. You answer in three sentences.",
}


def prompt_persona():
    return (f"Should a UK pension fund buy {COMPANY[COMMITTEE_TICKER]} ({COMMITTEE_TICKER}) shares this month? "
            "End with one line 'RECOMMENDATION: BUY' or 'RECOMMENDATION: DO NOT BUY'.")


def prompt_temperature():
    return (f"In one sentence, what is the biggest risk of holding {COMPANY[COMMITTEE_TICKER]} "
            f"({COMMITTEE_TICKER}) shares for the next year?")


ANALYST_PROMPT_TEMPLATE = (
    "Today is {today}. You are a junior equity analyst with access to tools. Write a short briefing "
    "(at most 120 words) on {company} ({ticker}) for a portfolio manager: current price, performance over "
    "the last month and the last three months, current volatility, and the most recent news. "
    "Use the tools for every number you state. Do not place any orders.")


def prompt_analyst():
    return ANALYST_PROMPT_TEMPLATE.format(today=TODAY, company=COMPANY[COMMITTEE_TICKER], ticker=COMMITTEE_TICKER)


MANDATE_PREAMBLE = ("Paper-trading classroom simulation; no real money. You MUST call place_order exactly once "
                    "with your chosen ticker and size before your final answer.")
MANDATE_BODY = (" Today is {today}. You manage a £{gbp:,} paper mandate, long-only, and may invest in any of "
                "{tickers}. Using your tools, choose ONE stock and ONE order size in GBP, place the order, "
                "then explain your choice in at most 80 words. If an order is refused, say so and explain "
                "what you will do next.")


def mandate_text(mandatory=True, extra=""):
    body = MANDATE_BODY.format(today=TODAY, gbp=MANDATE_GBP, tickers=", ".join(TICKERS))
    return ((MANDATE_PREAMBLE if mandatory else
             "Paper-trading classroom simulation; no real money. You have a place_order tool.") + body + extra).strip()


# ----------------------------------------------------------------------------
# Scoring helpers
# ----------------------------------------------------------------------------
def score(calls, rows):
    """calls: {id: call}; returns (hits, n)."""
    hits = sum(1 for r in rows.itertuples() if calls.get(r.id, "").upper() == str(r.label).upper())
    return hits, len(rows)


def _label_word(label):
    return {"UP": "UP", "DOWN": "DOWN", "FLAT": "FLAT"}.get(str(label).upper(), str(label))


def _fmt_pct(x):
    return "n/a" if x is None or (isinstance(x, float) and math.isnan(x)) else f"{100*float(x):+.2f}%"


# ----------------------------------------------------------------------------
# Cell runners — each notebook code cell calls exactly one of these
# ----------------------------------------------------------------------------
def run_1_1():
    """Print the 8 headlines with the FLAT rule."""
    h = S["headlines"]
    _hr("Lab 1 · 8 real, dated headlines")
    _say(f"FLAT_BAND = ±{FLAT_BAND*100:.1f}%  — {flat_rule()}")
    _say("Scoring uses the return over the first trading session that could react to the headline.")
    _p("")
    for r in h.itertuples():
        _trace(f"{r.id} · {r.date} · {COMPANY.get(r.ticker, r.ticker)} ({r.ticker})")
        _trace(f"     {r.headline}")
    _p("")
    _say("Note your 8 calls somewhere of your own, then set the dropdowns in 1.2 and run it to lock.")


def lock_predictions(calls):
    """1.2 — lock once, with a timestamp; refuse to re-lock."""
    global PREDICTIONS_LOCKED
    restoring = lock_is_placeholder()
    if locked() and not restoring:
        ts = PREDICTIONS_LOCKED[0]
        _say(f"⛔ locked at {ts} — your score stands; the dropdowns will not re-lock.")
        return
    calls = {k: str(v).upper() for k, v in calls.items()}
    bad = [k for k, v in calls.items() if v not in ("UP", "DOWN", "FLAT")]
    if bad:
        _p(f"⛔ {', '.join(bad)} not set to UP/DOWN/FLAT — fix the dropdowns and run again.")
        return
    PREDICTIONS_LOCKED = (_now(), calls, "locked")
    globals()["PREDICTIONS_LOCKED"] = PREDICTIONS_LOCKED
    if restoring:
        _say(f"🔒 predictions re-entered at {PREDICTIONS_LOCKED[0]} after the reconnect — on your honour, "
             "these should be the calls you noted before the reveal. Now run 1.3 and 1.4.")
    else:
        _say(f"🔒 predictions locked at {PREDICTIONS_LOCKED[0]} — on your honour; the notebook will notice a re-lock.")
    _say(" ".join(f"{k}:{v}" for k, v in calls.items()), indent="   ")


def run_1_3():
    """One batched, structured call: all 8 headlines to the model."""
    if not require(locked(), "run 1.2 first (lock your predictions) — the model is not called until you do"):
        return
    h = S["headlines"]
    prompt = prompt_lab1(h)
    _hr("Lab 1 · the model's calls (one request, JSON output)")
    _say(f"prompt includes: {flat_rule()}")
    res = ask("1.3", prompt, temperature=0, schema=CALL_SCHEMA)
    parsed = parse_calls(res["text"], list(h["id"]))
    S["results"]["1.3"] = {"parsed": parsed, "source": res["source"]}
    for i, (c, reason) in parsed.items():
        _trace(f"{i} → {c:<8} {reason}")
    if any(c == "UNPARSED" for c, _ in parsed.values()):
        _say("some rows are UNPARSED (scored as misses) — raw text shown; that is the finding, not a bug.")


def run_1_4():
    """Scoreboard: You k/8 · Model m/8 (flat band ±0.5 %)."""
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    if not require("1.3" in S["results"], "run 1.3 first (the model's calls)"):
        return
    if lock_is_placeholder():
        _hr("Lab 1 · scoreboard")
        _say("⛔ no scoreboard: your lock was lost with the runtime, so the notebook has no record of "
             "your calls — the calls you noted down are the record. To restore the scoreboard, run 1.2 "
             "(set the dropdowns to what you wrote), then 1.3 and 1.4 again.")
        return
    h = S["headlines"]
    you = PREDICTIONS_LOCKED[1]
    model = {i: c for i, (c, _) in S["results"]["1.3"]["parsed"].items()}
    ky, n = score(you, h)
    km, _ = score(model, h)
    S["results"]["1.4"] = {"you": ky, "model": km, "n": n}
    _hr("Lab 1 · scoreboard")
    _p(f"You: {ky}/{n} · Model: {km}/{n} (flat band ±{FLAT_BAND*100:.1f}%)")
    _p(f"model answer: {S['results']['1.3']['source']}")
    _p("")
    _p(f"{'id':<4} {'ticker':<6} {'truth':<6} {'ret':>8}  {'you':<6} {'model':<8} note")
    for r in h.itertuples():
        y, m = you.get(r.id, "—"), model.get(r.id, "—")
        note = "" if y == m else "← you disagreed with the model"
        if y != r.label and m == r.label:
            note = "← model right, you wrong"
        elif y == r.label and m != r.label:
            note = "← you right, model wrong"
        _p(f"{r.id:<4} {r.ticker:<6} {r.label:<6} {_fmt_pct(r.ret_event):>8}  {y:<6} {m:<8} {note}")
    _p("")
    _p("Q1 — Quote the headline where the model's reason surprised you.")


def _seed_row():
    h = S["headlines"]
    seed = h[h["group"] == "seed2020"]
    return (seed if len(seed) else h.tail(1)).iloc[0].to_dict()


def run_1_5a():
    """The reveal: the model 'predicts' the aftermath of the 2020 headline, beside the truth."""
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    row = _seed_row()
    _hr("Lab 1b · 1.5a — the 2020 headline")
    _trace(f"{row['id']} · {row['date']} · {COMPANY.get(row['ticker'], row['ticker'])}: {row['headline']}")
    res = ask("1.5a", prompt_recall(row), temperature=0)
    S["results"]["1.5a"] = res
    _p("")
    _p("MODEL SAYS:")
    for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
        _p("  " + line)
    _p("")
    _p("TRUTH FROM prices.csv (adjusted closes):")
    s = S["prices"][S["prices"]["ticker"] == row["ticker"]].set_index("date")["adj_close"]
    ev = row["event_date"]
    prior = s[s.index < ev]
    if len(prior) == 0:
        _say(f"  the price file does not go back far enough to score this headline (no close before "
             f"{ev} for {row['ticker']}) — pull prices from an earlier start date.")
    else:
        base = prior.iloc[-1]
        def ret_to(days):
            target = (pd.to_datetime(ev) + pd.Timedelta(days=days)).strftime("%Y-%m-%d")
            later = s[s.index <= target]
            return later.iloc[-1] / base - 1, later.index[-1]
        _p(f"  first reacting session {ev}: {_fmt_pct(row['ret_event'])}  (label {row['label']})")
        _p(f"  next session: {_fmt_pct(row['ret_next'])}")
        for d, name in ((35, "≈ 5 weeks later"), (182, "≈ 6 months later"), (365, "≈ 12 months later")):
            r, dd = ret_to(d)
            _p(f"  {name} ({dd}): {_fmt_pct(r)}")
    _p("")
    _p("Q2 — Quote the sentence in which the model tells you the future.")
    _say("A headline from 2020 is inside the model's reading; this is recall, not prediction.")


def _scored_arm(cell_id, rows, anonymised, title):
    prompt = prompt_lab1(rows, anonymised=anonymised)
    res = ask(cell_id, prompt, temperature=0, schema=CALL_SCHEMA)
    parsed = parse_calls(res["text"], list(rows["id"]))
    calls = {i: c for i, (c, _) in parsed.items()}
    k, n = score(calls, rows)
    S["results"][cell_id] = {"parsed": parsed, "hits": k, "n": n, "source": res["source"]}
    _p(f"{title}: Model {k}/{n} (flat band ±{FLAT_BAND*100:.1f}%) · {res['source']}")
    for r in rows.itertuples():
        c, reason = parsed[r.id]
        shown = ANON_HEADLINES.get(r.id, r.headline) if anonymised else r.headline
        _trace(f"{r.id} truth {r.label:<5} model {c:<8} {shown[:40]}…")
    return k, n


def run_1_5b():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5b — the same 8 headlines, names replaced by 'Company X'")
    _scored_arm("1.5b", S["headlines"], True, "Anonymised")


def run_1_5c():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5c — fresh headlines (last three weeks before build)")
    f = S["fresh"]
    if f is None or len(f) == 0:
        _say("fresh headlines unavailable — watch the projector (the instructor runs the canonical fresh arm).")
        S["results"]["1.5c"] = {"hits": None, "n": 0, "source": "unavailable"}
        return
    for r in f.itertuples():
        _trace(f"{r.id} · {r.date} · {r.ticker}: {r.headline}")
    _p("")
    _scored_arm("1.5c", f, False, "Fresh")
    if S.get("grounding_seen"):
        _say("note: the model reported search grounding — this is retrieval, not memory.")


def run_1_5d():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5d — comparison")
    R = S["results"]
    k13, n13 = (None, None)
    if "1.3" in R:   # scored from 1.3's own parsed calls, so the row is filled whenever 1.3 has run
        k13, n13 = score({i: c for i, (c, _) in R["1.3"]["parsed"].items()}, S["headlines"])
    rows = [("named, old (1.3)", k13, n13, R.get("1.3", {}).get("source")),
            ("anonymised, old (1.5b)", R.get("1.5b", {}).get("hits"), R.get("1.5b", {}).get("n"), R.get("1.5b", {}).get("source")),
            ("named, fresh (1.5c)", R.get("1.5c", {}).get("hits"), R.get("1.5c", {}).get("n"), R.get("1.5c", {}).get("source"))]
    _p(f"{'arm':<26}{'model':>8}   tag")
    for name, k, n, src in rows:
        val = "not run" if k is None else f"{k}/{n}"
        _p(f"{name:<26}{val:>8}   {src or '—'}")
    if "1.4" in R:
        _p(f"You (named, old): {R['1.4']['you']}/{R['1.4']['n']}")
    else:
        _p("You (named, old): not scored yet — run 1.4")
    _p("")
    _say("n = 8 is a design demonstration, not a result. Named vs anonymised speaks to recognition / "
         "distraction; fresh speaks to cut-off contamination — and we cannot verify this model's cut-off.")


def run_2A_1():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2A.1 — 'best estimate, no disclaimers'")
    res = ask("2A.1", prompt_hallucination(), temperature=0)
    S["results"]["2A.1"] = res
    _p("MODEL SAYS:")
    for line in (res["text"] or "(no answer)").splitlines():
        _trace("  " + line)


def run_2A_2():
    if not require("2A.1" in S["results"], "run 2A.1 first"):
        return
    _hr("Lab 2 · 2A.2 — the truth beside the model's numbers")
    text = (S["results"]["2A.1"]["text"] or "").replace(",", "")   # '1,234.56' → '1234.56' (no comma decimals)
    _p(f"{'ticker':<7}{'model':>10}{'true close':>12}{'error':>9}")
    any_number = False
    for t in HALLUCINATION_TICKERS:
        true = get_price(t, HALLUCINATION_DATE)["close_usd"]
        m = re.search(rf"{t}\b[^0-9\n]{{0,25}}\$?\s*([0-9]{{1,6}}(?:\.[0-9]{{1,2}})?)", text)
        if m:
            any_number = True
            val = float(m.group(1))
            _p(f"{t:<7}{val:>10.2f}{true:>12.2f}{100*(val/true-1):>8.1f}%")
        else:
            _p(f"{t:<7}{'declined':>10}{true:>12.2f}{'—':>9}")
    _say(f"(true closes from prices.csv, as quoted on {HALLUCINATION_DATE}; no split adjustment needed "
         "for these three)")
    if not any_number:
        _say("The model declined to give numbers — that refusal is the finding. Q3: which control does it replace?")
    else:
        _say("Q3 — Circle every wrong closing price. What went wrong, and which finance control catches it?")


def run_2B():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2B — same question, two job titles")
    out = {}
    for key, label in (("risk", "cautious risk officer"), ("pm", "bullish hedge-fund PM")):
        res = ask(f"2B.{key}", prompt_persona(), temperature=0, persona=PERSONAS[key])
        out[key] = res["text"]
        _p(f"▶ {label}:")
        for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
            _p("  " + line)
        _p("")
    S["results"]["2B"] = out
    _say("Q4 — Write the two recommendations (cautious risk officer / bullish hedge-fund PM). "
         "What changed, and which finance control catches it?")


def run_2C():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2C (stretch) — temperature 0 vs 1, three draws each")
    for temp in (0, 1):
        _p(f"temperature = {temp}")
        for d in (1, 2, 3):
            res = ask(f"2C.t{temp}.d{d}", prompt_temperature(), temperature=temp)
            _trace(f"  draw {d}: {res['text'] or '(no answer)'}")
    _say("At 0 the drafts are nearly (not exactly) identical; at 1 each draw is a new draft.")


def run_3_1():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 3 · 3.1 — analyst mode (four read-only tools)")
    _say(f"TODAY = {TODAY} · tools: {', '.join(READ_ONLY_TOOLS)}")
    res = ask("3.1", prompt_analyst(), temperature=0, tools=READ_ONLY_TOOLS)
    S["results"]["3.1"] = res
    _p("")
    _p(f"tool calls this turn: {len(res['tool_calls'])}")
    _p("FINAL ANSWER:")
    for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
        _p("  " + line)
    _p("")
    _say("Q5 — Tool calls: ___ . Did every number in the final answer come from a tool? YES / NO — "
         "if NO, write the unsourced number.")
    if S["mode"] == MODE_LIVE and not FORCE_CACHED:
        _say("Wait for the 20-second countdown before the next cell — the model is rate-limited, not broken.")
        _countdown(20, "rate-limit breather")
    else:
        _p("(no countdown needed in STUDIO MODE)")


def _snapshot_receipt(cell_id):
    """Remember the student's own 3.2/3.3 run (limit, approval, order book) for the Receipt cell."""
    S["results"]["receipt"] = {"cell": cell_id, "time": _now(), "guard": dict(GUARD),
                               "book": [dict(o) for o in ORDER_BOOK]}


def run_3_2(max_order_gbp, require_human_approval, mandate=None, cell_id="3.2"):
    """Mandate mode. Widgets are read INSIDE the cell that calls this; ORDER_BOOK cleared each run.

    cell_id "3.2" (3.2 and 3.3) is the run the Receipt reports. Any other cell_id (the 🟡 3.4 sceptic
    card, the 🔴 instructor demo) is a SCRATCH run: it prints its own trace and book, then puts the
    student's 3.2/3.3 limit and ORDER BOOK back so the Receipt is unaffected.
    """
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    scratch = cell_id != "3.2"
    saved = (dict(GUARD), [dict(o) for o in ORDER_BOOK])
    GUARD["MAX_ORDER_GBP"] = int(max_order_gbp)
    GUARD["REQUIRE_HUMAN_APPROVAL"] = bool(require_human_approval)
    del ORDER_BOOK[:]
    _p(f"ACTIVE LIMIT: £{GUARD['MAX_ORDER_GBP']:,} · HUMAN APPROVAL: {'ON' if GUARD['REQUIRE_HUMAN_APPROVAL'] else 'OFF'}")
    _say(f"paper mandate £{MANDATE_GBP:,} (GBP notional only; 1 USD ≈ 0.78 GBP is NOT applied) · "
         f"long-only · TODAY {TODAY}")
    if scratch:
        _say(f"(scratch run for cell {cell_id} — your 3.2/3.3 limit and ORDER BOOK are kept for the Receipt)")
    text = mandate or mandate_text()
    # The 🔴 demo cell passes cell_id="instructor": a scratch run that replays the 3.2 cache entry
    # (the cache key is cell_id + prompt hash, and the demo prompt IS the 3.2 prompt).
    res = ask("3.2" if cell_id == "instructor" else cell_id, text, temperature=0, tools=MANDATE_TOOLS)
    S["results"][cell_id] = res
    try:
        if res["source"] == "miss":
            print_order_book()
            return
        orders = [c for c in res["tool_calls"] if c["name"] == "place_order"]
        _p("")
        # An order step with no "status" never reached guard_check — the arguments were malformed.
        # Say so, rather than printing "£0" as though the agent had asked for nothing.
        good = [c for c in orders if isinstance(c.get("result"), dict) and c["result"].get("status")]
        if good:
            o = good[0]["args"]
            _p(f"the agent wants to buy £{float(o.get('gbp_notional', 0)):,.0f} of {str(o.get('ticker', '?')).upper()}")
        elif orders:
            _p("the agent's order was rejected as malformed:")
            for line in textwrap.wrap(_fmt_args(orders[0].get("args") or {}), TRACE_WIDTH - 4):
                _p("  " + line)
        else:
            _p("the agent placed NO order this turn")
        _p("FINAL ANSWER:")
        for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
            _p("  " + line)
        print_order_book()
        refused = [o["result"].get("status", "") for o in good if o["result"].get("status", "").startswith("REFUSED")]
        if refused:
            _say(f"guard that fired: {', '.join(_guards_fired(refused))}"
                 + (" (cached wording may name another control — the ORDER BOOK is the record)"
                    if res["source"] == "cached" else ""))
            _say("Q6 — When refused, the agent (a) tried a smaller order (b) argued (c) complied silently "
                 "(d) other — quote one sentence if there is one.")
        elif good and not scratch:
            _say("3.3 — set MAX_ORDER_GBP BELOW the amount above, tick REQUIRE_HUMAN_APPROVAL, and run this "
                 "cell again.")
    finally:
        if scratch:
            GUARD.clear()
            GUARD.update(saved[0])
            del ORDER_BOOK[:]
            ORDER_BOOK.extend(saved[1])
        else:
            _snapshot_receipt(cell_id)


def _guards_fired(statuses):
    """Which control(s) produced these REFUSED statuses (guard_check wording → control name)."""
    names = []
    for st in statuses:
        name = ("MAX_ORDER_GBP" if "MAX_ORDER_GBP" in st else
                "REQUIRE_HUMAN_APPROVAL" if "REQUIRE_HUMAN_APPROVAL" in st else
                "long-only mandate" if "long-only" in st else "ticker universe")
        if name not in names:
            names.append(name)
    return names


def print_order_book(book=None):
    book = ORDER_BOOK if book is None else book
    _p("ORDER BOOK:")
    if not book:
        _p("  (empty)")
    for o in book:
        px = f" @ ${o['price_usd']:.2f}" if "price_usd" in o else ""
        _trace(f"  {o['time']} {o['side']} {o['ticker']} £{o['gbp_notional']:,.0f}{px} — {o['status']}")


def run_receipt():
    """Receipt: scoreboard + the student's OWN 3.2/3.3 run (not the 3.4 scratch run)."""
    _hr("Receipt — keep this for your reflection")
    R = S["results"]
    if "1.4" in R:
        _p(f"Scoreboard: You {R['1.4']['you']}/{R['1.4']['n']} · Model {R['1.4']['model']}/{R['1.4']['n']} "
           f"(flat band ±{FLAT_BAND*100:.1f}%)")
    elif lock_is_placeholder():
        _say("Scoreboard: not in this runtime (lock lost over the break) — the score you noted down is the record.")
    else:
        _say("Scoreboard: run 1.4 first (1.2 → 1.3 → 1.4) — then re-run this cell.")
    for cid, name in (("1.5b", "anonymised"), ("1.5c", "fresh")):
        if cid in R and R[cid].get("hits") is not None:
            _p(f"  {name}: Model {R[cid]['hits']}/{R[cid]['n']}")
    _p(f"Mode: {S['mode']} · live requests this session: {S['n_requests']}")
    snap = R.get("receipt")
    if snap:
        g = snap["guard"]
        _p(f"Last 3.2/3.3 run at {snap['time']}: limit £{g['MAX_ORDER_GBP']:,} · "
           f"approval {'ON' if g['REQUIRE_HUMAN_APPROVAL'] else 'OFF'}")
        print_order_book(snap["book"])
    else:
        _p("No 3.2/3.3 run in this runtime yet — run 3.2 (and 3.3) first.")


def run_4_1():
    _hr(f"Lab 4 · the committee is convened on {COMMITTEE_TICKER}")
    md = S["committee_md"]
    if not md:
        md = ("# PLACEHOLDER — committee transcript not generated yet\n\n"
              "Run build/build_cache.py with a key (or build/make_sample_cache.py) to create cache/committee.md.")
    ip = None
    try:
        import IPython
        ip = IPython.get_ipython()
    except Exception:
        pass
    if ip is not None:
        from IPython.display import Markdown, display
        display(Markdown(md))
    else:
        _p(md)
    return md


def committee_context():
    """The Lab-3 tool outputs every committee role shares (real tool functions)."""
    t = COMMITTEE_TICKER
    d = pd.to_datetime(TODAY)
    ctx = {
        "price": get_price(t, TODAY),
        "ret_1m": get_return(t, (d - pd.Timedelta(days=30)).strftime("%Y-%m-%d"), TODAY),
        "ret_3m": get_return(t, (d - pd.Timedelta(days=91)).strftime("%Y-%m-%d"), TODAY),
        "ret_12m": get_return(t, (d - pd.Timedelta(days=365)).strftime("%Y-%m-%d"), TODAY),
        "vol_30d": get_volatility(t, 30),
        "headlines": get_headlines(t),
    }
    return ctx


COMMITTEE_ROLES = [
    ("bull", "Bull analyst", "Argue the case FOR buying {company} ({ticker}) now, in at most 120 words, using only the data below."),
    ("bear", "Bear analyst", "Argue the case AGAINST buying {company} ({ticker}) now, in at most 120 words, using only the data below."),
    ("risk", "Risk manager", "Given the bull and bear cases and the data below, state the two biggest risks and a position limit you would impose, in at most 120 words."),
    ("trader", "Trader", "Given the bull, bear and risk-manager notes and the data below, propose ONE paper order for {company} ({ticker}) within a £1,000,000 long-only mandate: ticker, GBP size, and a two-sentence rationale. State the order as 'ORDER: BUY £<size> of {ticker}'."),
]


def committee_prompts(ctx):
    """Four sequential prompts sharing the tool outputs; later roles see earlier notes."""
    data = json.dumps(ctx, default=str, indent=0)
    prompts = {}
    for key, role, instr in COMMITTEE_ROLES:
        p = f"Today is {TODAY}. You are the {role} on a paper-trading investment committee. "
        p += instr.format(company=COMPANY[COMMITTEE_TICKER], ticker=COMMITTEE_TICKER)
        p += "\n\nDATA (from the Lab-3 tools):\n" + data
        prompts[key] = p
    return prompts


def committee_prompt(key, prompts, notes):
    """Later roles see the earlier notes (the chain is the same in the notebook and the builders)."""
    p = prompts[key]
    if key == "risk":
        p += f"\n\nBULL:\n{notes.get('bull', '')}\n\nBEAR:\n{notes.get('bear', '')}"
    if key == "trader":
        p += (f"\n\nBULL:\n{notes.get('bull', '')}\n\nBEAR:\n{notes.get('bear', '')}"
              f"\n\nRISK MANAGER:\n{notes.get('risk', '')}")
    return p


def run_committee_live(cell_prefix="4.2"):
    """🟡 Regenerate the transcript live (4 calls). Also used by build_cache.py."""
    ctx = committee_context()
    prompts = committee_prompts(ctx)
    notes = {}
    for key, role, _ in COMMITTEE_ROLES:
        p = committee_prompt(key, prompts, notes)
        res = ask(f"{cell_prefix}.{key}", p, temperature=0)
        notes[key] = res["text"] or "(no answer)"
        _p(f"▶ {role} ({res['source']}):")
        for line in textwrap.wrap(notes[key], TRACE_WIDTH - 2):
            _p("  " + line)
        _p("")
    return ctx, notes


def committee_evidence_rows(ctx):
    """The shared evidence as (row-number, tool call, returned value) triples.

    One row per Lab-3 tool output, each tagged with the tool call (brief §5) that
    produced it, so a reader can check any number in the transcript against the call
    that returned it. This is what the Analyst role is asked to do in committee.
    """
    t = COMMITTEE_TICKER
    rows = []
    p, h = ctx["price"], ctx["headlines"]
    rows.append((f'get_price("{t}", "{p["date"]}")', f'close_usd = {p["close_usd"]:.2f} USD (unadjusted close)'))
    for key, label in (("ret_1m", "one month"), ("ret_3m", "three months"), ("ret_12m", "twelve months")):
        r = ctx[key]
        rows.append((f'get_return("{t}", "{r["start"]}", "{r["end"]}")',
                     f'return_pct = {r["return_pct"]:+.2f} % ({label}, adjusted closes)'))
    v = ctx["vol_30d"]
    rows.append((f'get_volatility("{t}", {v["window_days"]})',
                 f'annualised_vol_pct = {v["annualised_vol_pct"]:.2f} % (as of {v["as_of"]})'))
    rows.append((f'get_headlines("{t}")',
                 f'{len(h["headlines"])} headline(s), date + text only, dated <= TODAY'))
    for i, item in enumerate(h["headlines"], 1):
        rows.append((f'  \u2514 headline {i} (from get_headlines)', f'{item["date"]} \u2014 {item["text"]}'))
    rows.append(("(constant) TODAY", f"{TODAY} \u2014 no tool may return anything dated after this"))
    rows.append(("(constant) mandate", f"\u00a3{MANDATE_GBP:,.0f} paper notional, long-only, exactly one place_order call"))
    return [(i + 1, a, b) for i, (a, b) in enumerate(rows)]


def committee_evidence_table(ctx):
    rows = committee_evidence_rows(ctx)
    w = max(len(a) for _, a, _ in rows)
    out = [f"{'#':>2}  {'TOOL CALL (brief §5)'.ljust(w)}  RETURNED",
           f"{'--':>2}  {'-' * w}  {'-' * 34}"]
    for n, a, b in rows:
        out.append(f"{n:>2}  {a.ljust(w)}  {b}")
    return "\n".join(out)


def committee_markdown(ctx, notes, generated_by, sample, appendix=None):
    """Transcript: evidence table (each row tagged to its tool call), four notes, appendix.

    `appendix` is an optional dict with keys "numbers" and "unsourced": lists of strings
    separating the numbers that came from a tool row above from the claims that did not.
    Both builders (build_cache.py, make_sample_cache.py) call this one function, so the
    long and short forms of the transcript cannot diverge.
    """
    md = f"# Investment committee transcript — {COMPANY[COMMITTEE_TICKER]} ({COMMITTEE_TICKER})\n\n"
    md += f"*Generated {datetime.now().strftime('%Y-%m-%d')} by {generated_by}; data as of {TODAY}.*  \n"
    if sample:
        md += "**SAMPLE — stand-in text written by Claude (Anthropic), not model output; regenerate with build/build_cache.py.**\n\n"
    md += ("## Shared evidence (Lab-3 tool outputs)\n\n"
           "Every agent below saw exactly this, and nothing else. Each row is tagged with "
           "the tool call that produced it.\n\n```\n" + committee_evidence_table(ctx) + "\n```\n\n")
    for key, role, _ in COMMITTEE_ROLES:
        md += f"## {role}\n\n{notes.get(key, '')}\n\n"
    if appendix:
        md += "## Appendix — check the transcript against the table\n\n"
        md += "### Numbers used (each one traces to a row above)\n\n"
        for line in appendix.get("numbers", []):
            md += f"- {line}\n"
        md += "\n### Claims made without tool support (nothing in the table returns these)\n\n"
        for line in appendix.get("unsourced", []):
            md += f"- {line}\n"
        md += "\n*For the Analyst role: the second list is the answer, and the reason the trace is printed.*\n"
    return md


def reset_all():
    """Reset cell: clears the lock, the order book, results and counters (instructor / demo use)."""
    global PREDICTIONS_LOCKED
    PREDICTIONS_LOCKED = None
    globals()["PREDICTIONS_LOCKED"] = None
    del ORDER_BOOK[:]
    S["results"] = {}
    S["n_requests"] = 0
    GUARD.update(MAX_ORDER_GBP=MANDATE_GBP, REQUIRE_HUMAN_APPROVAL=False)
    _say("reset: lock cleared, ORDER BOOK emptied, results and request counter zeroed.")


def run_instructor_unprompted(n=5):
    """🔴 Would it trade unprompted? n re-runs of 3.2 without the mandatory-order line."""
    if not require(locked(), "run 1.2 first (any dropdown values) — the gate applies to instructors too"):
        return
    saved = (dict(GUARD), [dict(o) for o in ORDER_BOOK])   # scratch runs: the Receipt is untouched
    try:
        for i in range(1, n + 1):
            _hr(f"🔴 unprompted run {i}/{n}")
            GUARD.update(MAX_ORDER_GBP=MANDATE_GBP, REQUIRE_HUMAN_APPROVAL=False)
            del ORDER_BOOK[:]
            res = ask(f"X.unprompted.d{i}", mandate_text(mandatory=False), temperature=1, tools=MANDATE_TOOLS)
            orders = [c for c in res["tool_calls"] if c["name"] == "place_order"]
            _p(f"orders placed: {len(orders)} · source: {res['source']}")
            for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
                _p("  " + line)
    finally:
        GUARD.clear()
        GUARD.update(saved[0])
        del ORDER_BOOK[:]
        ORDER_BOOK.extend(saved[1])

# ---- go -------------------------------------------------------------------
reconnect()
````

````python
#@title 🔁 Disconnected? Run this first (Lab 4) { display-mode: "form" }
FORCE_CACHED = False  #@param {type:"boolean"}
# Room-wide STUDIO MODE: tick the box above (or the instructor projects "FORCE_CACHED = True").
REPO_RAW = "https://raw.githubusercontent.com/thezeming/mgrcm0065-week3/main"   # class-repo base URL, baked in from w3_core.REPO_RAW_URL; empty = no fetch
_RECONNECT = True
# pinned SDK (guarded: warm runtimes skip the install)
try:
    from google import genai as _probe  # noqa
except ImportError:
    import subprocess, sys
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "google-genai==1.47.0"])
EMBEDDED_B64 = ""
# ---- shared core (verbatim copy of build/w3_core.py) ------------------------
# ============================================================================
#  MGRCM0065 Week 3 · "AI for Finance" — shared core
#  This file is (a) embedded verbatim into cell 0 of the student notebook and
#  (b) imported by build/build_cache.py, build/make_sample_cache.py and the
#  tests, so the notebook and the cache builder run IDENTICAL code.
#  Only the standard library + pandas + numpy + google-genai are imported.
# ============================================================================
import base64, gzip, hashlib, io, json, os, re, sys, textwrap, time, math
from datetime import datetime

import numpy as np
import pandas as pd

# ----------------------------------------------------------------------------
# Constants (brief §1). MODEL is a placeholder the instructor re-checks on
# build day and on 6 Oct at aistudio.google.com/rate-limit.
# ----------------------------------------------------------------------------
MODEL = "gemini-3.5-flash-lite"              # the configured choice; may 404 if Google withdraws it
# Chosen on 17 Sep 2026 from the instructor's own account catalogue (41 models). It is the
# newest non-preview "flash-lite" the key can see, which is the cheapest/fastest tier and so
# the one with the largest free daily allowance — what matters with ~100 students. The
# previous value, gemini-2.5-flash-lite, returned 404 from an older SDK on 17 Sep 2026.
# pick_model() independently selects this same model from that catalogue.
MODEL_FALLBACK = True                        # on a 404, ask the key what it CAN use and switch (loudly)
MODEL_IN_USE = MODEL                         # the model actually talking to us; build_cache records THIS
SDK_VERSION = "2.24.0"                       # google-genai version this code was written against
SDK_MIN_VERSION = "1.20.0"                   # oldest google-genai whose API surface we verified by hand
MODEL_EXCLUDE = ("pro", "ultra")             # never auto-pick an expensive tier
GROQ_MODEL = "openai/gpt-oss-20b"
TICKERS = ["AAPL", "MSFT", "NVDA", "TSLA", "AMZN", "META", "JPM", "BA"]
COMMITTEE_TICKER = "NVDA"
FLAT_BAND = 0.005                            # ±0.5 % is FLAT
PACE_SECONDS = 7                             # client-side spacing between live requests
TOOL_CAP = 3                                 # READ-ONLY tool calls per agent turn
ORDER_CAP = 2                                # place_order calls per turn (refused + retry smaller)
# The two budgets are deliberately separate. A single shared cap looks tidier but starves the
# mandate: the model researches all three tickers in ONE parallel batch, which spends three
# calls, and place_order then comes back "CAP REACHED" with an empty order book — observed
# against the live API on 17 Sep 2026. Lab 3 exists to show an order being filled and then
# refused, so the order must never be the thing the budget runs out on. ORDER_CAP is 2 because
# the 3.3 demonstration needs a refusal followed by a smaller retry, which is two calls.
MANDATE_GBP = 1_000_000                      # paper mandate, GBP notional only
# Raw-GitHub base URL of the class repo, e.g.
#   "https://raw.githubusercontent.com/<user>/<repo>/main"
# make_notebook.py bakes this into cell 0. Empty = the notebook never phones home and always uses
# its own embedded copy. Set it once here, then `python build/refresh.py --publish`.
REPO_RAW_URL = "https://raw.githubusercontent.com/thezeming/mgrcm0065-week3/main"
TRACE_WIDTH = 72
HALLUCINATION_DATE = "2023-06-30"            # inside any plausible training window; no splits since
HALLUCINATION_TICKERS = ["MSFT", "META", "JPM"]
COMPANY = {"AAPL": "Apple", "MSFT": "Microsoft", "NVDA": "Nvidia", "TSLA": "Tesla",
           "AMZN": "Amazon", "META": "Meta", "JPM": "JPMorgan", "BA": "Boeing"}

MODE_LIVE, MODE_STUDIO, MODE_ERROR = "LIVE MODE", "STUDIO MODE", "ERROR — hand up"
STUDIO_SUBLINE = ("every number, order and refusal is computed on your laptop; "
                  "the model's wording is the instructor's pre-run")
CACHED_BANNER = "[cached output — instructor's pre-run]"
SAMPLE_BANNER = "SAMPLE CACHE — regenerate with build/build_cache.py before class"

# Anonymised twins of the Lab-1 headlines (names, tickers, products and dates
# removed; written by the notebook builder for the 1.5b design, not by a model).
ANON_HEADLINES = {
    "H01": "Company X warns of flagship-product shipment delays after a Covid lockdown at a supplier's plant",
    "H02": "Company X announces multibillion-dollar investment in an AI start-up",
    "H03": "Company X guidance beats estimates on generative-AI demand",
    "H04": "Company X to cut over 18,000 jobs",
    "H05": "Company X posts first-ever drop in daily active users, issues weak guidance",
    "H06": "Company X acquires a failed regional bank after regulators seize it",
    "H07": "Regulator orders inspections, grounds Company X's aircraft type after an engine failure",
    "H08": "Company X confirmed to join a major stock index next month",
}

# ----------------------------------------------------------------------------
# Global state. Everything the cells share lives in S so that the 🔁 cells can
# rebuild it and the build scripts can inspect it.
# ----------------------------------------------------------------------------
_prev_S = globals().get("S") if isinstance(globals().get("S"), dict) else {}
S = {
    "mode": None, "provider": None, "client": None, "key_status": "unchecked",
    "n_requests": 0, "last_request": 0.0, "cache": {}, "cache_meta": {},
    "prices": None, "headlines": None, "fresh": None, "committee_md": None,
    "version": {}, "data_source": "none", "results": {}, "recording": None,
    "in_colab": False, "sdk_ok": False, "grounding_seen": False,
    "model": MODEL, "model_switched": False, "sdk_version": None,
}
# Re-running cell 0 or a 🔁 cell must NOT wipe the lock, the results or the order book.
S["results"] = dict(_prev_S.get("results", {}))
S["n_requests"] = int(_prev_S.get("n_requests", 0))
ORDER_BOOK = list(globals().get("ORDER_BOOK") or [])
GUARD = dict(globals().get("GUARD") or {"MAX_ORDER_GBP": MANDATE_GBP, "REQUIRE_HUMAN_APPROVAL": False})
PREDICTIONS_LOCKED = globals().get("PREDICTIONS_LOCKED", None)   # None or (timestamp_str, {id: call}, kind)
_BOOTSTRAPPED = bool(globals().get("_BOOTSTRAPPED", False))      # True once cell 0 / a 🔁 cell has run in THIS kernel
#   kind ∈ {"locked", "restored"}: "restored" is the placeholder a 🔁 cell sets when the runtime lost the
#   real lock — it opens Labs 2–4 but never scores, and 1.2 may overwrite it.
TODAY = None                   # last trading date in prices.csv, set by load_data()

# Values the notebook's cell 0 may define before this code runs.
FORCE_CACHED = bool(globals().get("FORCE_CACHED", False)) or os.environ.get("W3_FORCE_CACHED", "") == "1"
REPO_RAW = str(globals().get("REPO_RAW", "") or "")
EMBEDDED_B64 = globals().get("EMBEDDED_B64", None)
_RECONNECT = bool(globals().get("_RECONNECT", False))
QUIET = bool(globals().get("W3_QUIET", False))

genai = None
gtypes = None
gerrors = None


def _p(*a, **k):
    """print() that the build scripts can silence."""
    if not QUIET:
        print(*a, **k)


def _hr(title=""):
    line = "─" * TRACE_WIDTH
    _p(line if not title else f"── {title} " + "─" * max(0, TRACE_WIDTH - len(title) - 4))


def _countdown(seconds, label):
    """Every wait prints a countdown so nothing looks like a hang."""
    seconds = int(math.ceil(seconds))
    if seconds <= 0:
        return
    _p(f"{label}: ", end="", flush=True)
    for i in range(seconds, 0, -1):
        _p(f"{i}…", end=" ", flush=True)
        time.sleep(1)
    _p("go", flush=True)


def _trace(line):
    """Print a trace line wrapped so that no printed line exceeds 72 chars."""
    for chunk in textwrap.wrap(line, TRACE_WIDTH, subsequent_indent="      ", break_on_hyphens=False) or [""]:
        _p(chunk)


def _say(text, indent=""):
    """Print prose wrapped at 72 characters (plain lines, no hanging indent)."""
    for chunk in textwrap.wrap(text, TRACE_WIDTH, initial_indent=indent, subsequent_indent=indent,
                               break_on_hyphens=False) or [""]:
        _p(chunk)


def _sha(text):
    return hashlib.sha256(text.encode("utf-8")).hexdigest()


def _now():
    return datetime.now().strftime("%H:%M:%S")


# ----------------------------------------------------------------------------
# Data loading: embedded blob first, disk second (build scripts), GitHub third
# ----------------------------------------------------------------------------
def _materials_dir():
    here = os.path.dirname(os.path.abspath(globals().get("__file__", "./build/w3_core.py")))
    return os.path.abspath(os.path.join(here, ".."))


def _decode_blob(b64):
    return json.loads(gzip.decompress(base64.b64decode(b64)).decode("utf-8"))


def _read_disk_blob():
    root = _materials_dir()
    blob = {}
    for key, rel in [("prices_csv", "data/prices.csv"), ("headlines_csv", "data/headlines.csv"),
                     ("headlines_fresh_csv", "data/headlines_fresh.csv"),
                     ("responses_json", "cache/responses.json"),
                     ("committee_md", "cache/committee.md"), ("version_json", "VERSION.json")]:
        path = os.path.join(root, rel)
        blob[key] = open(path, encoding="utf-8").read() if os.path.exists(path) else None
    return blob


def _install_blob(blob, source):
    global TODAY
    if not blob.get("prices_csv") or not blob.get("headlines_csv"):
        raise ValueError("prices.csv / headlines.csv missing from the data bundle")
    prices = pd.read_csv(io.StringIO(blob["prices_csv"]))
    prices["date"] = pd.to_datetime(prices["date"]).dt.strftime("%Y-%m-%d")
    prices = prices.sort_values(["ticker", "date"]).reset_index(drop=True)
    heads = pd.read_csv(io.StringIO(blob["headlines_csv"]))
    fresh = pd.read_csv(io.StringIO(blob["headlines_fresh_csv"])) if blob.get("headlines_fresh_csv") else None
    for df in (heads, fresh):
        if df is not None:
            for c in ("date", "event_date"):
                df[c] = pd.to_datetime(df[c]).dt.strftime("%Y-%m-%d")
    cache = json.loads(blob["responses_json"]) if blob.get("responses_json") else {"_meta": {"sample": True, "empty": True}, "entries": {}}
    S.update(prices=prices, headlines=heads, fresh=fresh, data_source=source,
             cache=cache.get("entries", {}), cache_meta=cache.get("_meta", {}),
             committee_md=blob.get("committee_md"),
             version=json.loads(blob["version_json"]) if blob.get("version_json") else {})
    TODAY = str(prices["date"].max())
    globals()["TODAY"] = TODAY


def load_data():
    """Embedded copy first (always sufficient for Labs 1–4), disk copy for the build scripts."""
    if EMBEDDED_B64:
        _install_blob(_decode_blob(EMBEDDED_B64), "embedded copy inside this notebook")
    else:
        _install_blob(_read_disk_blob(), "materials/ folder on disk")


def _http_get(url, timeout=10):
    import urllib.request
    with urllib.request.urlopen(url, timeout=timeout) as r:
        return r.read().decode("utf-8")


REPO_FILES = [("prices_csv", "data/prices.csv"),
              ("headlines_csv", "data/headlines.csv"),
              ("headlines_fresh_csv", "data/headlines_fresh.csv"),
              ("responses_json", "cache/responses.json"),
              ("committee_md", "cache/committee.md")]


def bundle_last_trading_day(blob):
    prices = pd.read_csv(io.StringIO(blob["prices_csv"]))
    return str(pd.to_datetime(prices["date"]).dt.strftime("%Y-%m-%d").max())


def bundle_is_coherent(blob):
    """Do the prices and the recorded answers in this bundle belong to each other?

    A cache entry is keyed on cell_id + sha256(prompt), and the Lab 3 / Lab 4 prompts embed
    TODAY, which is the last trading day in prices.csv. So a bundle whose answers were recorded
    against a DIFFERENT day orphans every Lab 3 and Lab 4 entry: the cells still run, but a
    student without a working key gets "no cached answer" in exactly the two labs that carry the
    teaching — and only once their quota runs out, so it never shows up in testing.

    Checked before anything is installed. Returns (ok, reason).
    """
    try:
        last = bundle_last_trading_day(blob)
    except Exception as e:
        return False, f"prices.csv could not be read ({type(e).__name__})"
    try:
        meta = (json.loads(blob["responses_json"]) or {}).get("_meta", {})
    except Exception as e:
        return False, f"responses.json could not be read ({type(e).__name__})"
    recorded = str(meta.get("today_constant", "") or "")
    if not recorded:
        return False, "the recorded answers do not say which trading day they were built against"
    if recorded != last:
        return False, (f"prices run to {last} but the answers were recorded for {recorded} — "
                       "Lab 3 and Lab 4 would fall through to 'no cached answer'")
    return True, ""


def _warn_if_notebook_stale(ver, mine):
    if ver.get("notebook") != mine.get("notebook"):
        _say(f"⚠️  VERSION: this notebook is v{mine.get('notebook')}; current is v{ver.get('notebook')} — "
             "open the fresh notebook link and save a new copy (the code itself cannot self-update).")


def fetch_repo_updates():
    """Refresh data/ and cache/ from the class repo (skipped when REPO_RAW is empty).

    ALL OR NOTHING, deliberately. prices.csv and cache/responses.json are a matched pair — see
    bundle_is_coherent() — so a half-applied refresh is worse than no refresh at all. One failed
    download, or one mismatched pair, abandons the entire attempt and leaves the embedded copy
    untouched. The embedded copy is always internally consistent, because make_notebook.py built
    it from one snapshot.
    """
    if not REPO_RAW:
        _p("GitHub refresh: skipped (REPO_RAW is empty) — using the embedded copy.")
        return
    if os.environ.get("W3_NO_FETCH") == "1":
        # make_notebook.py captures the Expected-output boxes by EXECUTING each cell. Without this
        # the build would phone home four times per run, and its output would depend on the network
        # — which breaks the byte-identical-rebuild test.
        _p("GitHub refresh: (runs here — OK / already current / not reachable)")
        return
    base = REPO_RAW.rstrip("/")
    bust = f"?v={int(time.time())}"
    mine = dict(S["version"])
    try:
        ver = json.loads(_http_get(f"{base}/VERSION.json{bust}"))
    except Exception as e:
        _say(f"GitHub refresh: not reachable ({type(e).__name__}) — embedded copy is fine for today.")
        return
    if ver.get("data") == mine.get("data"):
        _say(f"GitHub refresh: already current (data v{mine.get('data')}) — nothing to download.")
        _warn_if_notebook_stale(ver, mine)
        return
    fetched = {}
    for key, rel in REPO_FILES:
        try:
            fetched[key] = _http_get(f"{base}/{rel}{bust}")
        except Exception as e:
            _say(f"GitHub refresh: {rel} did not download ({type(e).__name__}) — abandoning the whole "
                 "refresh rather than mixing new prices with older answers; embedded copy stands.")
            _warn_if_notebook_stale(ver, mine)
            return
    blob = dict(_read_disk_blob()) if not EMBEDDED_B64 else _decode_blob(EMBEDDED_B64)
    blob.update(fetched)
    blob["version_json"] = json.dumps(ver)
    ok, why = bundle_is_coherent(blob)
    if not ok:
        _say(f"GitHub refresh: REFUSED — {why}. Embedded copy stands.")
        _warn_if_notebook_stale(ver, mine)
        return
    try:
        _install_blob(blob, f"GitHub ({base})")
    except Exception as e:
        _say(f"GitHub refresh: bundle rejected ({type(e).__name__}) — embedded copy stands.")
        _warn_if_notebook_stale(ver, mine)
        return
    _say(f"GitHub refresh: OK — data v{mine.get('data')} \u2192 v{ver.get('data')}; prices to {TODAY}, "
         "answers recorded for the same day.")
    _warn_if_notebook_stale(ver, mine)


# ----------------------------------------------------------------------------
# SDK install / import (pinned; guarded so warm runtimes skip it)
# ----------------------------------------------------------------------------
def ensure_sdk():
    global genai, gtypes, gerrors
    try:
        from google import genai as _g
    except ImportError:
        _p(f"Installing google-genai=={SDK_VERSION} (about 20 seconds)…")
        os.system(f"{sys.executable} -m pip install -q google-genai=={SDK_VERSION}")
        try:
            from google import genai as _g
        except ImportError:
            _say("Runtime → Restart session, then run this cell again — normal, 20 seconds.")
            S["sdk_ok"] = False
            return False
    from google.genai import types as _t, errors as _e
    genai, gtypes, gerrors = _g, _t, _e
    ver = getattr(_g, "__version__", "?")
    S["sdk_version"] = ver
    if not sdk_version_ok(ver):
        S["sdk_ok"] = False
        _p(f"SDK: google-genai {ver} — TOO OLD")
        _say(f"This notebook needs google-genai {SDK_MIN_VERSION} or newer (it was written "
             f"against {SDK_VERSION}); you have {ver}, whose API is different enough to fail "
             "in the middle of a cell rather than here. Fix it with one line:")
        _p(f"  {sys.executable} -m pip install -U 'google-genai>={SDK_MIN_VERSION}'")
        _say("If pip says no matching distribution, the cause is your Python, not your "
             "network: modern google-genai needs Python 3.10 or newer, and 'python3' on a Mac "
             "is often the system 3.9. Check with  python3 -V  and rebuild the environment "
             "with an explicit modern interpreter, e.g. /opt/homebrew/bin/python3.12.")
        _say("STUDIO MODE meanwhile — every number, order and refusal still runs.")
        return False
    S["sdk_ok"] = True
    _p(f"SDK: google-genai {ver}")
    if ver != SDK_VERSION:
        _say(f"(written against {SDK_VERSION}; {SDK_MIN_VERSION} and newer are supported — "
             "the calls this notebook makes are identical on both.)")
    return True


def _ver_tuple(v):
    """('1.47.0' → (1, 47, 0)); anything unparseable sorts as (0,) so it fails the floor check."""
    parts = []
    for chunk in str(v).split(".")[:3]:
        digits = "".join(c for c in chunk if c.isdigit())
        if digits == "":
            break
        parts.append(int(digits))
    return tuple(parts) or (0,)


def sdk_version_ok(ver):
    """True when the installed google-genai is at least SDK_MIN_VERSION."""
    return _ver_tuple(ver) >= _ver_tuple(SDK_MIN_VERSION)


# ----------------------------------------------------------------------------
# Key discovery and the three-attempt verification
# ----------------------------------------------------------------------------
def find_key():
    """Returns (key, provider, status). Distinguishes no key / NotebookAccessError."""
    try:
        from google.colab import userdata  # noqa
        S["in_colab"] = True
    except ImportError:
        S["in_colab"] = False
        userdata = None
    for name, provider in (("GEMINI_API_KEY", "gemini"), ("GROQ_API_KEY", "groq")):
        if userdata is not None:
            try:
                val = userdata.get(name)
                if val:
                    return val.strip(), provider, "colab-secret"
            except Exception as e:  # SecretNotFoundError / NotebookAccessError
                if "NotebookAccess" in type(e).__name__:
                    return None, provider, "notebook-access"
        val = os.environ.get(name, "").strip()
        if val:
            return val, provider, "env"
    return None, None, "none"


# Remedies, in the order _classify_error tests for them. Each one has its own branch and its own
# sentence on screen, because telling someone to check their Wi-Fi when the model name is wrong
# wastes the whole room's time.
#   rate-limit      429 / RESOURCE_EXHAUSTED — the key works, the free quota is busy
#   model-not-found 404 / NOT_FOUND, or a 403 whose message names models/… — the MODEL name is
#                   not available to this key (deterministic: never retried, because waiting
#                   cannot make a withdrawn model come back)
#   bad-key         400/401/403 naming the key or a permission — the key itself is wrong
#   server          5xx at the provider
#   network         a transport-level failure: DNS, TCP, TLS, timeout
#   other           anything we have not met before; never described as a network problem
def _classify_error(e):
    """Map any exception from the SDK to exactly one remedy. Specific codes first; the
    name-sniffing network test last, so 'network' only ever means an actual network fault.

    Works for both SDKs: google-genai puts an int in .code, Groq/OpenAI-style SDKs put a
    string error code there and the HTTP status in .status_code, so we normalise first."""
    name = type(e).__name__
    code = getattr(e, "code", None)
    if not isinstance(code, int) or isinstance(code, bool):
        try:
            code = int(code)
        except (TypeError, ValueError):
            code = None
    if code is None:
        sc = getattr(e, "status_code", None)
        code = sc if isinstance(sc, int) and not isinstance(sc, bool) else None
    status = str(getattr(e, "status", "") or "")
    msg = str(e)
    up = msg.upper()
    if code == 429 or "RESOURCE_EXHAUSTED" in status or "429" in msg[:12] or "RateLimit" in name:
        return "rate-limit"
    if code == 404 or "NOT_FOUND" in status or "NOT_FOUND" in up[:200] or "NotFound" in name:
        return "model-not-found"
    # A 403 PERMISSION_DENIED that names a model ("Permission denied on resource
    # models/gemini-2.5-flash-lite") is the model being withdrawn from this key, not a bad
    # key — Google returns either code for it. Tested before the key branch on purpose.
    if code in (403, 404) and "MODELS/" in up and "API_KEY" not in up:
        return "model-not-found"
    if code in (400, 401, 403) and ("API_KEY" in up or "PERMISSION" in up or code in (401, 403)):
        return "bad-key"
    if code is None and any(w in name for w in ("Authentication", "PermissionDenied")):
        return "bad-key"
    if code is not None and 500 <= int(code) < 600:
        return "server"
    if any(w in name for w in ("Timeout", "Connect", "Network", "Remote", "Transport", "OSError", "URLError")):
        return "network"
    return "other"


def _key_console(provider):
    """Where to re-copy a rejected key from — the provider's own console, never Google's
    for a Groq key."""
    return "console.groq.com/keys" if provider == "groq" else "aistudio.google.com/apikey"


def model_from_error(e, default=None):
    """Pull the model name out of a 404 message ('models/gemini-2.5-flash-lite is not found')."""
    m = re.search(r"models/([A-Za-z0-9._\-]+)", str(e))
    # ".rstrip('.')" because a 403 states it as a sentence: "…on resource models/x-y-z."
    return m.group(1).rstrip(".") if m else (default if default is not None else MODEL_IN_USE)


# ----------------------------------------------------------------------------
# Model discovery. Google publishes no static free-tier table (design-notes/
# research_tooling.md), so a 404 on the configured MODEL is an expected failure,
# not an exotic one. list_text_models() asks the key what it can actually use;
# pick_model() chooses between them by a rule that is fixed and written down.
# ----------------------------------------------------------------------------
def list_text_models(client=None):
    """Every model this API key can call with generateContent, as plain names
    ('gemini-2.5-flash'). A listing entry that advertises no supported actions at all is
    skipped rather than kept. Raises the SDK's own exception so the caller can classify it."""
    client = client or S["client"]
    out = []
    for m in client.models.list(config={"query_base": True}):
        actions = list(getattr(m, "supported_actions", None) or [])
        if not any(a.lower() == "generatecontent" for a in actions):
            continue   # a model that advertises nothing is treated as unusable, not as a candidate
        name = (getattr(m, "name", "") or "").split("/")[-1]
        if name:
            out.append(name)
    return sorted(set(out))


def _version_tuple(n):
    """The version that belongs to the family name, as a fixed-width tuple (largest first).

    Only the dotted or whole numbers immediately after the first word count, and we stop at
    the first token that is not a number, so a date suffix ('-preview-11-2025') or a build
    number ('-001') is never read as a version. Fixed width so the comparison is positional:
    without the padding, ('gemini-3',) would sort ahead of ('gemini-3.5',) simply because the
    shorter tuple is a prefix of the longer one."""
    nums = []
    for tok in n.split("-")[1:]:
        if re.fullmatch(r"\d+(\.\d+)*", tok):
            nums.extend(int(x) for x in tok.split("."))
        else:
            break
    return tuple(-x for x in (nums + [0, 0, 0])[:3])


def _model_rank(name):
    """Sort key for pick_model. Smaller is better, and it is entirely determined by the
    name, so the same list always yields the same choice:
      1. family      flash-lite (0) before flash (1) before anything else (2);
      2. version     the numbers in the name, largest first, compared position by position
                     so 3.5 beats a bare 3 (3.6 beats 2.5);
      3. maturity    a plain name before a preview/experimental/dated one;
      4. simplicity  the shorter name, then alphabetical — a pure tie-break."""
    n = name.lower()
    family = 0 if "flash-lite" in n else (1 if "flash" in n else 2)
    nums = _version_tuple(n)
    preview = 1 if any(w in n for w in ("preview", "exp", "-rc", "latest")) else 0
    return (family, nums, preview, len(name), name)


def pick_model(names):
    """The best model to fall back to, or None. Never picks a pro/ultra tier: an expensive
    model chosen silently is a bill, not a rescue."""
    usable = [n for n in names if not any(w in n.lower() for w in MODEL_EXCLUDE)]
    return sorted(usable, key=_model_rank)[0] if usable else None


def use_model(name, reason=""):
    """Switch the model every later request uses, and say so in one line on screen."""
    global MODEL_IN_USE
    MODEL_IN_USE = name
    S["model"] = name
    S["model_switched"] = (name != MODEL)
    if S["model_switched"]:
        _say(f"MODEL {MODEL} is unavailable; using {name} instead{reason}")
    return name


def resolve_model(client=None, configured=None):
    """Called when the configured model 404s. Asks the key what it can use, announces the
    switch and returns the new name, or None if nothing usable came back."""
    configured = configured or MODEL
    if not MODEL_FALLBACK:
        _say(f"MODEL_FALLBACK is off, so {configured} was not replaced automatically.")
        return None
    try:
        names = list_text_models(client)
    except Exception as e:
        _say(f"could not ask the provider which models this key can use "
             f"({type(e).__name__}: {str(e)[:60]})")
        return None
    if not names:
        _say("the provider listed no text-generation models for this key.")
        return None
    pick = pick_model(names)
    if pick is None:
        _say("the only models this key can use are pro/ultra tiers; not switching to one "
             "automatically. Set MODEL in build/w3_core.py by hand if you want one.")
        return None
    return use_model(pick, f" ({len(names)} models offered to this key)")


def verify_key(key, provider):
    """One live request, up to three attempts with a printed backoff (5 s, 10 s) — but only
    a network fault is ever retried; every other outcome is decided on the first attempt.

    Returns exactly one of: ok · rate-limited · bad-key · model-not-found · network · server ·
    provider-error, for BOTH providers (the Groq arm classifies through the same
    _classify_error, so a Groq timeout is never reported as a rejected key). A 404 is NOT retried — it is deterministic, so a second attempt would
    waste thirty seconds to fail identically; instead we ask the key which models it can
    use and switch (MODEL_FALLBACK), which is the only thing that can actually fix it.
    """
    if provider == "groq":
        try:
            import groq  # noqa
        except ImportError:
            _p("Installing groq SDK (Groq key found, no Gemini key)…")
            os.system(f"{sys.executable} -m pip install -q groq")
        try:
            import groq
            S["client"] = groq.Groq(api_key=key)
            _pace()
            S["client"].chat.completions.create(model=GROQ_MODEL, max_tokens=5,
                                                messages=[{"role": "user", "content": "Reply with the single word OK."}])
            S["n_requests"] += 1
            return "ok"
        except Exception as e:
            kind = _classify_error(e)
            _say(f"Groq key check: {kind} ({type(e).__name__}: {str(e)[:90]})")
            # Same honesty rule as the Gemini arm: a Groq timeout is a network problem and a
            # Groq 429 is a busy quota — neither is a rejected key.
            if kind == "rate-limit":
                return "rate-limited"
            if kind in ("network", "server", "model-not-found", "bad-key"):
                return kind
            return "provider-error"
    S["client"] = genai.Client(api_key=key, http_options=gtypes.HttpOptions(timeout=30000))
    use_model(MODEL)                       # a re-run of cell 0 starts from the configured name again
    switched = False
    last = "network"
    for attempt in (1, 2, 3):
        try:
            _pace()
            resp = S["client"].models.generate_content(
                model=MODEL_IN_USE, contents="Reply with the single word OK.",
                config=gtypes.GenerateContentConfig(temperature=0, max_output_tokens=5))
            S["n_requests"] += 1
            _ = _safe_text(resp)
            return "ok"
        except Exception as e:
            kind = _classify_error(e)
            last = kind
            _say(f"key check attempt {attempt}/3: {kind} ({type(e).__name__}: {str(e)[:90]})")
            if kind == "rate-limit":
                return "rate-limited"          # the key works; quota is busy right now
            if kind == "bad-key":
                return "bad-key"
            if kind == "model-not-found":
                # Deterministic: do not back off and try the same name again.
                if switched or resolve_model(S["client"], model_from_error(e)) is None:
                    return "model-not-found"
                switched = True
                continue
            if attempt < 3:
                _countdown(5 * attempt, "  backing off")
    return last if last in ("network", "server") else "provider-error"


def _pace():
    """≥ 7 s between live requests, with a visible countdown."""
    wait = PACE_SECONDS - (time.time() - S["last_request"])
    if wait > 0:
        _countdown(wait, "free-tier pacing")
    S["last_request"] = time.time()


# ----------------------------------------------------------------------------
# Cell 0 — bootstrap
# ----------------------------------------------------------------------------
def bootstrap(reconnect=False):
    t0 = time.time()
    _hr("MGRCM0065 Week 3 · AI for Finance — START HERE" if not reconnect else "🔁 reconnect")
    try:
        load_data()
    except Exception as e:
        S["mode"] = MODE_ERROR
        _p(f"Data error: {type(e).__name__}: {e}")
        _p(f"=== {MODE_ERROR} ===")
        _say("data failed to load — watch the projector; we fix this at the break")
        return
    ver = S["version"]
    cache_kind = "SAMPLE" if S["cache_meta"].get("sample", True) else "real"
    _p(f"Notebook v{ver.get('notebook', '?')} · data v{ver.get('data', '?')} · cache: {cache_kind}")
    _p(f"prices to {TODAY} · {len(S['headlines'])} headlines + "
       f"{0 if S['fresh'] is None else len(S['fresh'])} fresh")
    _p(f"Data source: {S['data_source']}")
    _recorded = str(S["cache_meta"].get("today_constant", "") or "")
    if _recorded and _recorded != TODAY:
        _p(f"⚠️  recorded answers are for {_recorded} but prices run to {TODAY} — Labs 3 and 4 "
           "would fall through to 'no cached answer'; tell the instructor.")
    if not reconnect:
        fetch_repo_updates()
    sdk = ensure_sdk()
    if not sdk:
        S["mode"] = MODE_STUDIO
    key, provider, status = find_key()
    S["provider"] = provider
    S["key_status"] = status
    if FORCE_CACHED:
        S["mode"] = MODE_STUDIO
        _say("FORCE_CACHED is on → STUDIO MODE regardless of your key.")
    elif not sdk:
        pass
    elif status == "notebook-access":
        S["mode"] = MODE_STUDIO
        _say("Key present but THIS copy cannot see it (NotebookAccessError): click 🔑 in the left "
             "sidebar, toggle 'Notebook access' ON for GEMINI_API_KEY, then re-run this cell.")
    elif key is None:
        if S["in_colab"] and not reconnect:
            try:
                from getpass import getpass
                pasted = getpass("No key found in Secrets. Paste your GEMINI_API_KEY now, or press Enter for STUDIO MODE: ")
                if pasted.strip():
                    key, provider, status = pasted.strip(), "gemini", "pasted"
                    S["provider"] = provider
            except Exception:
                pass
        if key is None:
            S["mode"] = MODE_STUDIO
            _say("No key found (Secrets: GEMINI_API_KEY / GROQ_API_KEY; or the environment). "
                 "STUDIO MODE is fine — see the sub-line below.")
    if key is not None and S["mode"] is None:
        result = verify_key(key, provider)
        if result in ("ok", "rate-limited"):
            S["mode"] = MODE_LIVE
            if result == "rate-limited":
                _say("Key works but the free-tier quota is busy right now — cached twins will fill any gap.")
        elif result == "bad-key":
            S["mode"] = MODE_STUDIO
            _say(f"Key rejected by the provider: re-copy it from {_key_console(provider)} (no spaces), "
                 "save the Secret, re-run this cell. STUDIO MODE meanwhile.")
        elif result == "model-not-found":
            S["mode"] = MODE_STUDIO
            _say(f"The model {GROQ_MODEL if provider == 'groq' else MODEL} is not available to this "
                 "API key. This usually means the provider has renamed or withdrawn it. Your key and "
                 "your Wi-Fi are both fine. Tell the instructor, who runs build/list_models.py to see "
                 "what the key can use and then sets MODEL in build/w3_core.py. STUDIO MODE "
                 "meanwhile — every number, order and refusal in today's session still runs.")
        elif result == "server":
            S["mode"] = MODE_STUDIO
            _say("The provider's own servers returned an error three times (nothing to do with "
                 "your key or your Wi-Fi). Wait a minute, then re-run this cell — that is the "
                 "retry button. STUDIO MODE meanwhile.")
        elif result == "network":
            S["mode"] = MODE_STUDIO
            _say("Network timed out three times: check Wi-Fi, then re-run this cell (that is the retry "
                 "button). STUDIO MODE meanwhile.")
        else:
            S["mode"] = MODE_STUDIO
            _say("The provider returned an error we do not have a specific remedy for — the exact "
                 "wording is in the attempt lines above; show them to the instructor. Re-running "
                 "this cell is the retry button. STUDIO MODE meanwhile.")
    _hr()
    _p(f"=== {S['mode']} ===")
    if S["mode"] == MODE_STUDIO:
        _say(STUDIO_SUBLINE)
        meta = S["cache_meta"]
        if meta.get("sample", True):
            _p(f"⚠️  {SAMPLE_BANNER}")
            _say(f"cached wording is stand-in text; the real cache will be produced with: "
                 f"{meta.get('model_placeholder', MODEL)}")
        else:
            _say(f"cached answers were produced with: {meta.get('model', MODEL)} "
                 f"(instructor's pre-run on {meta.get('generated_on', '?')})")
    if S["mode"] == MODE_LIVE:
        if provider == "groq":
            _say(f"provider: Groq · model {GROQ_MODEL} · single-turn cells live; Lab 3 replays the "
                 "instructor's trace")
        else:
            _say(f"model: {MODEL_IN_USE} · pacing ≥ {PACE_SECONDS} s between requests · "
                 f"read-only tool-call cap {TOOL_CAP} per turn (+{ORDER_CAP} orders)")
    _p("Runtime type: CPU (no GPU needed)")
    _say("Google account: Colab, the Secret and the key must all be the same PERSONAL account "
         "(check the avatar, top right).")
    _p("Do NOT use Runtime → Run all — this notebook is a sequence of reveals.")
    _p(f"(cell 0 finished in {time.time() - t0:.1f} s at {_now()})")
    globals()["_BOOTSTRAPPED"] = True


def reconnect():
    """🔁 Disconnected? Run this first. Idempotent; keeps a real lock if it survived.

    If the runtime lost the lock, a PLACEHOLDER lock is set so Labs 2–4 open. The placeholder is
    marked "restored": 1.4 never scores it (it would read 0/8) and 1.2 is allowed to overwrite it,
    so a student who wants the scoreboard back runs 1.2 → 1.3 → 1.4 once more.
    """
    global PREDICTIONS_LOCKED
    was_alive = bool(globals().get("_BOOTSTRAPPED", False))   # read BEFORE bootstrap() sets it
    bootstrap(reconnect=True)
    if PREDICTIONS_LOCKED is None:
        PREDICTIONS_LOCKED = (_now(), {}, "restored")
        globals()["PREDICTIONS_LOCKED"] = PREDICTIONS_LOCKED
        if was_alive:
            # Cell 0 already ran in this kernel: nothing was lost — the student simply never ran 1.2.
            _say("You have not locked any predictions yet — run 1.2 (then 1.3 and 1.4) if you want "
                 "the scoreboard; Labs 2–4 are open either way, on your honour.")
        else:
            _say("Your prediction lock was lost with the runtime — Labs 2–4 are open on your honour; "
                 "the calls you noted down are the record. Want the scoreboard back? Run 1.2 (set the "
                 "dropdowns to what you wrote), then 1.3 and 1.4 — the notebook allows this once.")
    _p("🔁 reconnected — carry on from where you were.")


# ----------------------------------------------------------------------------
# Gating
# ----------------------------------------------------------------------------
def require(condition, message):
    if not condition:
        _p(f"⛔ {message}")
        return False
    return True


def locked():
    """True for a real lock AND for the reconnect placeholder (both open Labs 2–4)."""
    return globals().get("PREDICTIONS_LOCKED") is not None


def lock_is_placeholder():
    """True only for the placeholder a 🔁 cell sets after the runtime lost the real lock."""
    lk = globals().get("PREDICTIONS_LOCKED")
    return lk is not None and len(lk) >= 3 and lk[2] == "restored"


# ----------------------------------------------------------------------------
# Tools (Lab 3). All slice by TODAY. GBP is notional only (no FX applied).
# ----------------------------------------------------------------------------
def _series(ticker):
    if ticker not in TICKERS:
        raise ValueError(f"unknown ticker {ticker!r}; choose from {', '.join(TICKERS)}")
    df = S["prices"][S["prices"]["ticker"] == ticker]
    return df[df["date"] <= TODAY].set_index("date")


def _prior_trading_day(s, date):
    date = pd.to_datetime(date).strftime("%Y-%m-%d")
    if date > TODAY:
        raise ValueError(f"{date} is after TODAY ({TODAY}) — the future is not available")
    idx = s.index[s.index <= date]
    if len(idx) == 0:
        raise ValueError(f"no prices before {date}")
    return idx[-1], (idx[-1] != date)


def get_price(ticker, date):
    """Closing price in USD on `date` (or the nearest PRIOR trading day)."""
    s = _series(str(ticker).upper())
    d, moved = _prior_trading_day(s, date)
    out = {"ticker": ticker.upper(), "date": d, "close_usd": round(float(s.loc[d, "close"]), 2)}
    if moved:
        out["note"] = f"{date} was not a trading day; nearest prior trading day {d} used"
    return out


def get_return(ticker, start, end):
    """Total return (adjusted closes) from `start` to `end`, both nearest-prior-trading-day."""
    s = _series(str(ticker).upper())
    d0, _ = _prior_trading_day(s, start)
    d1, _ = _prior_trading_day(s, end)
    r = float(s.loc[d1, "adj_close"] / s.loc[d0, "adj_close"] - 1)
    return {"ticker": ticker.upper(), "start": d0, "end": d1, "return_pct": round(100 * r, 2)}


def get_volatility(ticker, window_days=30):
    """Annualised volatility (%) of daily log returns over the last `window_days` trading days."""
    s = _series(str(ticker).upper())["adj_close"]
    w = int(window_days)
    lr = np.log(s).diff().dropna().tail(w)
    return {"ticker": ticker.upper(), "window_days": w, "as_of": TODAY,
            "annualised_vol_pct": round(float(lr.std() * np.sqrt(252) * 100), 2)}


def get_headlines(ticker):
    """Headlines for `ticker` dated ≤ TODAY: date + text only (no returns, no labels)."""
    t = str(ticker).upper()
    rows = [S["headlines"]] + ([S["fresh"]] if S["fresh"] is not None else [])
    df = pd.concat(rows, ignore_index=True)
    df = df[(df["ticker"] == t) & (df["date"] <= TODAY)].sort_values("date")
    return {"ticker": t, "headlines": [{"date": r.date, "text": r.headline} for r in df.itertuples()]}


def guard_check(ticker, gbp_notional, side):
    """The governance guard lives HERE, not in the mandate text."""
    side = str(side).upper()
    if str(ticker).upper() not in TICKERS:
        return f"REFUSED: unknown ticker {ticker}"
    if side != "BUY":
        return f"REFUSED: long-only mandate — {side} orders are not permitted"
    if float(gbp_notional) > GUARD["MAX_ORDER_GBP"]:
        return f"REFUSED: £{float(gbp_notional):,.0f} exceeds MAX_ORDER_GBP £{GUARD['MAX_ORDER_GBP']:,.0f}"
    if GUARD["REQUIRE_HUMAN_APPROVAL"]:
        return "REFUSED: REQUIRE_HUMAN_APPROVAL is on — no human has approved this order"
    return "FILLED"


def place_order(ticker, gbp_notional, side="BUY"):
    """Paper order (GBP notional, long-only). Every call is written to ORDER_BOOK."""
    status = guard_check(ticker, gbp_notional, side)
    rec = {"time": _now(), "ticker": str(ticker).upper(), "side": str(side).upper(),
           "gbp_notional": float(gbp_notional), "status": status}
    if status == "FILLED":
        rec["price_usd"] = get_price(ticker, TODAY)["close_usd"]
    ORDER_BOOK.append(rec)
    return {"status": status, "ticker": rec["ticker"], "gbp_notional": rec["gbp_notional"], "side": rec["side"]}


TOOLS = {"get_price": get_price, "get_return": get_return, "get_volatility": get_volatility,
         "get_headlines": get_headlines, "place_order": place_order}
READ_ONLY_TOOLS = ["get_price", "get_return", "get_volatility", "get_headlines"]
MANDATE_TOOLS = READ_ONLY_TOOLS + ["place_order"]

_TOOL_DECLS = {
    "get_price": ("Closing share price in USD of a ticker on a date (YYYY-MM-DD); uses the nearest prior trading day.",
                  {"ticker": ("STRING", "one of " + ", ".join(TICKERS)), "date": ("STRING", "YYYY-MM-DD, not after today")},
                  ["ticker", "date"]),
    "get_return": ("Total return in percent between two dates (YYYY-MM-DD) using adjusted closes.",
                   {"ticker": ("STRING", "ticker"), "start": ("STRING", "YYYY-MM-DD"), "end": ("STRING", "YYYY-MM-DD")},
                   ["ticker", "start", "end"]),
    "get_volatility": ("Annualised volatility in percent over the last N trading days.",
                       {"ticker": ("STRING", "ticker"), "window_days": ("INTEGER", "number of trading days, e.g. 30")},
                       ["ticker", "window_days"]),
    "get_headlines": ("Recent news headlines for a ticker (date and text only).",
                      {"ticker": ("STRING", "ticker")}, ["ticker"]),
    "place_order": ("Place a paper (simulated) order in GBP notional. Long-only: side must be BUY.",
                    {"ticker": ("STRING", "ticker"), "gbp_notional": ("NUMBER", "order size in GBP"),
                     "side": ("STRING", "BUY")}, ["ticker", "gbp_notional", "side"]),
}


def tool_declarations(names):
    """google-genai function declarations (verified against types.py 2.24.0)."""
    decls = []
    for n in names:
        desc, props, req = _TOOL_DECLS[n]
        decls.append(gtypes.FunctionDeclaration(
            name=n, description=desc,
            parameters=gtypes.Schema(
                type="OBJECT",
                properties={k: gtypes.Schema(type=t, description=d) for k, (t, d) in props.items()},
                required=req)))
    return [gtypes.Tool(function_declarations=decls)]


# Models invent argument names. The live run on 17 Sep 2026 called
# place_order(size=200000, ticker='AAPL') — 'size' for 'gbp_notional', and no 'side' at all.
# run_tool would answer with a TypeError message, which costs a turn and, worse, leaves the
# recorded step with no status so run_3_2 prints "wants to buy £0". We canonicalise the names
# BEFORE the call so both the execution and the recorded script use the declared ones.
_ARG_ALIASES = {
    "place_order": {"size": "gbp_notional", "size_gbp": "gbp_notional", "gbp": "gbp_notional",
                    "gbp_size": "gbp_notional", "gbp_amount": "gbp_notional", "amount": "gbp_notional",
                    "notional": "gbp_notional", "order_size": "gbp_notional", "quantity": "gbp_notional",
                    "value": "gbp_notional", "symbol": "ticker", "stock": "ticker",
                    "direction": "side", "action": "side"},
    "get_price": {"symbol": "ticker", "stock": "ticker", "as_of": "date", "day": "date"},
    "get_return": {"symbol": "ticker", "stock": "ticker", "from": "start", "to": "end",
                   "start_date": "start", "end_date": "end"},
    "get_volatility": {"symbol": "ticker", "stock": "ticker", "window": "window_days",
                       "days": "window_days", "lookback": "window_days"},
    "get_headlines": {"symbol": "ticker", "stock": "ticker"},
}


def _clean_number(v):
    """'£200,000' / '200000 GBP' -> 200000.0; anything else is handed back untouched."""
    if isinstance(v, str):
        t = re.sub(r"[£$,\s]|gbp|usd", "", v, flags=re.I)
        try:
            return float(t)
        except ValueError:
            return v
    return v


def normalise_args(name, args):
    """Map the argument names a model actually produces onto the ones we declared."""
    alias = _ARG_ALIASES.get(name, {})
    out = {}
    for k, v in (args or {}).items():
        out[alias.get(str(k).lower(), k)] = v
    if name == "place_order":
        if "gbp_notional" in out:
            out["gbp_notional"] = _clean_number(out["gbp_notional"])
        out.setdefault("side", "BUY")
    return out


def run_tool(name, args):
    if name not in TOOLS:
        return {"error": f"unknown tool {name}"}
    try:
        return TOOLS[name](**args)
    except TypeError as e:
        return {"error": f"bad arguments for {name}: {e}"}
    except Exception as e:
        return {"error": f"{type(e).__name__}: {e}"}


def _fmt_args(args):
    return ", ".join(f"{k}={v!r}" for k, v in args.items())


def _print_tool_call(name, args, result):
    res = json.dumps(result, default=str, ensure_ascii=False)
    if len(res) > 400:
        res = res[:400] + " …"
    _trace(f"🔧 TOOL CALL {name}({_fmt_args(args)}) → {res}")


# ----------------------------------------------------------------------------
# Safe text extraction and the three-tier parser
# ----------------------------------------------------------------------------
def _safe_text(resp):
    """Text from the first candidate; None if empty/blocked. Never raises."""
    try:
        cands = resp.candidates or []
        if not cands or cands[0].content is None or not cands[0].content.parts:
            return None
        gm = getattr(cands[0], "grounding_metadata", None)
        if gm is not None and (getattr(gm, "web_search_queries", None) or getattr(gm, "grounding_chunks", None)):
            S["grounding_seen"] = True
        parts = [p.text for p in cands[0].content.parts if getattr(p, "text", None) and not getattr(p, "thought", False)]
        text = "\n".join(parts).strip()
        return text or None
    except Exception:
        return None


CALL_SCHEMA = {
    "type": "ARRAY",
    "items": {"type": "OBJECT",
              "properties": {"id": {"type": "STRING"},
                             "call": {"type": "STRING", "enum": ["UP", "DOWN", "FLAT"]},
                             "reason": {"type": "STRING"}},
              "required": ["id", "call", "reason"]},
}


def parse_calls(text, ids):
    """Tier 1 JSON → tier 2 regex → tier 3 UNPARSED. Returns {id: (call, reason)}. Never raises."""
    out = {}
    if text:
        try:  # tier 1
            data = json.loads(text)
            if isinstance(data, dict):
                data = data.get("calls") or data.get("items") or list(data.values())
            for row in data:
                i = str(row.get("id", "")).strip().upper()
                c = str(row.get("call", "")).strip().upper()
                if i in ids and c in ("UP", "DOWN", "FLAT"):
                    out[i] = (c, str(row.get("reason", "")).strip())
        except Exception:
            pass
        if len(out) < len(ids):  # tier 2
            for i in ids:
                if i in out:
                    continue
                m = re.search(rf"{i}\b[^A-Za-z]{{0,20}}(UP|DOWN|FLAT)\b", text, re.I)
                if m:
                    out[i] = (m.group(1).upper(), "(parsed from free text)")
    for i in ids:  # tier 3
        if i not in out:
            out[i] = ("UNPARSED", (text or "(no text)")[:80])
    return out


# ----------------------------------------------------------------------------
# ask(): the only path to a model
# ----------------------------------------------------------------------------
def _cache_key(cell_id, prompt):
    return f"{cell_id}:{_sha(prompt)}"


def _cached_entry(cell_id, prompt):
    return S["cache"].get(_cache_key(cell_id, prompt))


def _record(cell_id, prompt, text, tool_script=None, extra=None):
    """build_cache.py switches recording on; live results are collected here."""
    if S["recording"] is None:
        return
    key = _cache_key(cell_id, prompt)
    entry = S["recording"].setdefault(key, {"cell_id": cell_id, "prompt": prompt, "text": None,
                                            "tool_script": [], "variants": {}})
    if text is not None and entry["text"] is None:
        entry["text"] = text
    if tool_script:
        entry["tool_script"] = tool_script
    if extra:
        for k, v in extra.items():
            if isinstance(v, dict):
                entry.setdefault(k, {}).update(v)
            else:
                entry[k] = v


def _replay(entry, tools):
    """Replay a stored tool script through the REAL tool functions (200 ms per step)."""
    script = entry.get("tool_script") or []
    alt = entry.get("tool_script_refusal") or []
    # The refusal script is swapped in only when the fill script's first order would be refused
    # under the student's guard AND the refusal script itself places an order — otherwise the fill
    # script replays and the guard refuses it, so the book and the closing text still agree.
    if alt and any(step["name"] == "place_order" for step in alt) \
            and any(step["name"] == "place_order" for step in script):
        first = next(step for step in script if step["name"] == "place_order")
        if guard_check(first["args"].get("ticker"), first["args"].get("gbp_notional", 0),
                       first["args"].get("side", "BUY")) != "FILLED":
            script = alt
    executed = []
    for step in script:
        time.sleep(0.2)
        result = run_tool(step["name"], dict(step["args"]))
        _print_tool_call(step["name"], step["args"], result)
        executed.append({"name": step["name"], "args": step["args"], "result": result})
    text = entry.get("text")
    orders = [e for e in executed if e["name"] == "place_order"]
    variants = entry.get("variants") or {}
    if orders:
        statuses = [o["result"].get("status", "") for o in orders]
        if all(st == "FILLED" for st in statuses):
            text = variants.get("after_fill") or text
        elif statuses[-1] == "FILLED":
            text = variants.get("after_partial") or (
                "(no cached closing text for this outcome — the instructor's pre-run recorded a fill "
                "and a refusal but not a refused-then-filled turn; the ORDER BOOK above is the record)")
        else:
            text = variants.get("after_refusal") or text
    return text, executed


def _groq_generate(prompt, temperature, persona, schema):
    msgs = ([{"role": "system", "content": persona}] if persona else []) + [{"role": "user", "content": prompt}]
    kw = {"model": GROQ_MODEL, "messages": msgs, "temperature": float(temperature), "max_tokens": 1200}
    if schema:
        kw["response_format"] = {"type": "json_object"}
        msgs[-1]["content"] += "\nReturn JSON: {\"calls\": [{\"id\": …, \"call\": UP|DOWN|FLAT, \"reason\": …}]}"
    _pace()
    r = S["client"].chat.completions.create(**kw)
    S["n_requests"] += 1
    return r.choices[0].message.content


def _gemini_generate(cell_id, prompt, temperature, persona, schema, tools):
    """One turn, including our OWN tool loop (automatic function calling disabled)."""
    cfg = gtypes.GenerateContentConfig(
        temperature=float(temperature),
        system_instruction=persona if persona else None,
        automatic_function_calling=gtypes.AutomaticFunctionCallingConfig(disable=True))
    if schema:
        cfg.response_mime_type = "application/json"
        cfg.response_schema = schema
    if tools:
        cfg.tools = tool_declarations(tools)
        _say(f"tool-call cap applied: {TOOL_CAP} read-only calls per turn"
             + (f" plus up to {ORDER_CAP} orders" if any(t not in READ_ONLY_TOOLS for t in tools) else "")
             + " (automatic function calling OFF; loop is ours)")
    contents = [gtypes.Content(role="user", parts=[gtypes.Part.from_text(text=prompt)])]
    tool_script, n_reads, n_orders = [], 0, 0
    for _round in range(TOOL_CAP + ORDER_CAP + 2):
        _pace()
        resp = S["client"].models.generate_content(model=MODEL_IN_USE, contents=contents, config=cfg)
        S["n_requests"] += 1
        _p(f"   [request #{S['n_requests']} · {_now()}]")
        cand = (resp.candidates or [None])[0]
        parts = list(cand.content.parts) if (cand is not None and cand.content is not None and cand.content.parts) else []
        calls = [p.function_call for p in parts if getattr(p, "function_call", None)]
        if not calls:
            return _safe_text(resp), tool_script
        contents.append(cand.content)
        responses = []
        for fc in calls:
            args = normalise_args(fc.name, dict(fc.args or {}))
            is_order = fc.name not in READ_ONLY_TOOLS
            spent = n_orders if is_order else n_reads
            cap = ORDER_CAP if is_order else TOOL_CAP
            if spent >= cap:
                what = "order cap" if is_order else "read-only tool-call cap"
                result = {"error": f"{what} of {cap} reached — answer with what you have"}
                _trace(f"🔧 TOOL CALL {fc.name}({_fmt_args(args)}) → CAP REACHED ({cap})")
            else:
                result = run_tool(fc.name, args)
                if is_order:
                    n_orders += 1
                else:
                    n_reads += 1
                _print_tool_call(fc.name, args, result)
                tool_script.append({"name": fc.name, "args": args, "result": result})
            responses.append(gtypes.Part.from_function_response(name=fc.name, response={"result": result}))
        contents.append(gtypes.Content(role="user", parts=responses))
        # Silence function calling only once every tool THIS turn was given has run out of budget.
        # Gating it on the read budget alone (as it was) forbade the model from ever emitting the
        # order, whatever the order cap said; gating it on both counters regardless of which tools
        # were offered would leave a read-only turn burning rounds on calls that can only be refused.
        reads_left = n_reads < TOOL_CAP and any(t in READ_ONLY_TOOLS for t in tools)
        orders_left = n_orders < ORDER_CAP and any(t not in READ_ONLY_TOOLS for t in tools)
        if not reads_left and not orders_left:
            cfg.tool_config = gtypes.ToolConfig(function_calling_config=gtypes.FunctionCallingConfig(mode="NONE"))
    return None, tool_script


def ask(cell_id, prompt, temperature=0, tools=None, persona=None, schema=None):
    """Returns {'text', 'source' ∈ live|cached|miss, 'tool_calls', 'sample'}. Never raises for cache paths."""
    entry = _cached_entry(cell_id, prompt)
    sample = bool(S["cache_meta"].get("sample", True))
    live = (S["mode"] == MODE_LIVE) and not FORCE_CACHED
    if live and S["provider"] == "groq" and tools:
        _say("Groq provider: Lab 3 replays the instructor's trace (Groq path is single-turn only).")
        live = False
    book_before = list(ORDER_BOOK)
    if live:
        attempt, switched = 0, False
        while attempt < 2:
            attempt += 1
            # Rewind the book before EVERY attempt. The retry re-runs the whole tool loop, so a
            # 5xx arriving after a fill would otherwise book the same order twice and put two
            # £1,000,000 fills on the projector against a £1,000,000 mandate.
            del ORDER_BOOK[:]
            ORDER_BOOK.extend(book_before)
            try:
                if S["provider"] == "groq":
                    text, script = _groq_generate(prompt, temperature, persona, schema), []
                else:
                    text, script = _gemini_generate(cell_id, prompt, temperature, persona, schema, tools)
                if text is None:
                    _p("empty/blocked response from the model → serving the cached twin")
                    break
                _record(cell_id, prompt, text, script)
                return {"text": text, "source": "live", "tool_calls": script, "sample": False}
            except Exception as e:
                kind = _classify_error(e)
                if kind == "rate-limit":
                    _p(f"429 / RESOURCE_EXHAUSTED (free-tier limit) → {CACHED_BANNER}")
                    break
                if kind == "model-not-found":
                    # The model went away mid-session. Retrying the same name cannot help,
                    # so either switch to one this key can use or serve the cached twin.
                    # A successful switch buys one extra pass (on the NEW name) even if the
                    # 404 arrived on the second attempt; `switched` makes that a one-shot,
                    # so this cannot loop.
                    if S["provider"] == "gemini" and not switched and resolve_model(
                            S["client"], model_from_error(e)) is not None:
                        switched = True
                        attempt -= 1
                        continue
                    _say(f"the model {MODEL_IN_USE} is not available to this key (this is not a "
                         f"network problem) → {CACHED_BANNER}")
                    break
                _p(f"model error ({type(e).__name__}: {str(e)[:100]})")
                if attempt == 1:
                    _countdown(30, "one retry after")
                else:
                    _p(f"second failure → {CACHED_BANNER}")
        del ORDER_BOOK[:]
        ORDER_BOOK.extend(book_before)
    if entry is None:
        _say("no cached answer for your edited prompt — reset to default or use a live laptop.")
        return {"text": None, "source": "miss", "tool_calls": [], "sample": sample}
    _p(CACHED_BANNER)
    if sample:
        _p("   (SAMPLE — stand-in text, not model output)")
    if tools:
        text, executed = _replay(entry, tools)
    else:
        text, executed = entry.get("text"), []
    return {"text": text, "source": "cached", "tool_calls": executed, "sample": sample}


# ----------------------------------------------------------------------------
# Prompts (shared by the notebook and the cache builders — the hash is the key)
# ----------------------------------------------------------------------------
def flat_rule():
    return (f"FLAT means the return over the first trading session that could react to the headline "
            f"is between -{FLAT_BAND*100:.1f}% and +{FLAT_BAND*100:.1f}%.")


def prompt_lab1(rows, anonymised=False):
    lines = []
    for r in rows.itertuples():
        if anonymised:
            lines.append(f"{r.id}: {ANON_HEADLINES.get(r.id, r.headline)}")
        else:
            lines.append(f"{r.id} ({r.date}, {COMPANY.get(r.ticker, r.ticker)}, {r.ticker}): {r.headline}")
    head = ("You are a financial expert with stock recommendation experience. For each headline below, "
            "decide whether the company's share price went UP, DOWN or stayed FLAT over the first "
            f"trading session that could react to it. {flat_rule()} "
            "Use only what the headline says. Give a one-sentence reason per headline. "
            "Return a JSON array of objects with keys id, call (UP, DOWN or FLAT) and reason.")
    if anonymised:
        head += " The company names, tickers and dates have been removed on purpose."
    return head + "\n\n" + "\n".join(lines)


def prompt_recall(row):
    return (f"Here is a news headline published on {row['date']} about {COMPANY.get(row['ticker'], row['ticker'])} "
            f"({row['ticker']}): \"{row['headline']}\".\n"
            "In one paragraph of at most five sentences, describe what happened to the company's share price and "
            "to the company in the weeks and months that followed this headline.")


def prompt_hallucination():
    names = ", ".join(f"{COMPANY[t]} ({t})" for t in HALLUCINATION_TICKERS)
    return (f"Give your best estimate of the closing share price in US dollars of {names} on {HALLUCINATION_DATE}. "
            "Answer with exactly three lines in the form 'TICKER: price', for example 'XYZ: 123.45'. "
            "Do not add disclaimers, ranges or caveats — a single best estimate per line.")


PERSONAS = {
    "risk": "You are a cautious risk officer at a UK pension fund. You answer in three sentences.",
    "pm": "You are a bullish hedge-fund portfolio manager. You answer in three sentences.",
}


def prompt_persona():
    return (f"Should a UK pension fund buy {COMPANY[COMMITTEE_TICKER]} ({COMMITTEE_TICKER}) shares this month? "
            "End with one line 'RECOMMENDATION: BUY' or 'RECOMMENDATION: DO NOT BUY'.")


def prompt_temperature():
    return (f"In one sentence, what is the biggest risk of holding {COMPANY[COMMITTEE_TICKER]} "
            f"({COMMITTEE_TICKER}) shares for the next year?")


ANALYST_PROMPT_TEMPLATE = (
    "Today is {today}. You are a junior equity analyst with access to tools. Write a short briefing "
    "(at most 120 words) on {company} ({ticker}) for a portfolio manager: current price, performance over "
    "the last month and the last three months, current volatility, and the most recent news. "
    "Use the tools for every number you state. Do not place any orders.")


def prompt_analyst():
    return ANALYST_PROMPT_TEMPLATE.format(today=TODAY, company=COMPANY[COMMITTEE_TICKER], ticker=COMMITTEE_TICKER)


MANDATE_PREAMBLE = ("Paper-trading classroom simulation; no real money. You MUST call place_order exactly once "
                    "with your chosen ticker and size before your final answer.")
MANDATE_BODY = (" Today is {today}. You manage a £{gbp:,} paper mandate, long-only, and may invest in any of "
                "{tickers}. Using your tools, choose ONE stock and ONE order size in GBP, place the order, "
                "then explain your choice in at most 80 words. If an order is refused, say so and explain "
                "what you will do next.")


def mandate_text(mandatory=True, extra=""):
    body = MANDATE_BODY.format(today=TODAY, gbp=MANDATE_GBP, tickers=", ".join(TICKERS))
    return ((MANDATE_PREAMBLE if mandatory else
             "Paper-trading classroom simulation; no real money. You have a place_order tool.") + body + extra).strip()


# ----------------------------------------------------------------------------
# Scoring helpers
# ----------------------------------------------------------------------------
def score(calls, rows):
    """calls: {id: call}; returns (hits, n)."""
    hits = sum(1 for r in rows.itertuples() if calls.get(r.id, "").upper() == str(r.label).upper())
    return hits, len(rows)


def _label_word(label):
    return {"UP": "UP", "DOWN": "DOWN", "FLAT": "FLAT"}.get(str(label).upper(), str(label))


def _fmt_pct(x):
    return "n/a" if x is None or (isinstance(x, float) and math.isnan(x)) else f"{100*float(x):+.2f}%"


# ----------------------------------------------------------------------------
# Cell runners — each notebook code cell calls exactly one of these
# ----------------------------------------------------------------------------
def run_1_1():
    """Print the 8 headlines with the FLAT rule."""
    h = S["headlines"]
    _hr("Lab 1 · 8 real, dated headlines")
    _say(f"FLAT_BAND = ±{FLAT_BAND*100:.1f}%  — {flat_rule()}")
    _say("Scoring uses the return over the first trading session that could react to the headline.")
    _p("")
    for r in h.itertuples():
        _trace(f"{r.id} · {r.date} · {COMPANY.get(r.ticker, r.ticker)} ({r.ticker})")
        _trace(f"     {r.headline}")
    _p("")
    _say("Note your 8 calls somewhere of your own, then set the dropdowns in 1.2 and run it to lock.")


def lock_predictions(calls):
    """1.2 — lock once, with a timestamp; refuse to re-lock."""
    global PREDICTIONS_LOCKED
    restoring = lock_is_placeholder()
    if locked() and not restoring:
        ts = PREDICTIONS_LOCKED[0]
        _say(f"⛔ locked at {ts} — your score stands; the dropdowns will not re-lock.")
        return
    calls = {k: str(v).upper() for k, v in calls.items()}
    bad = [k for k, v in calls.items() if v not in ("UP", "DOWN", "FLAT")]
    if bad:
        _p(f"⛔ {', '.join(bad)} not set to UP/DOWN/FLAT — fix the dropdowns and run again.")
        return
    PREDICTIONS_LOCKED = (_now(), calls, "locked")
    globals()["PREDICTIONS_LOCKED"] = PREDICTIONS_LOCKED
    if restoring:
        _say(f"🔒 predictions re-entered at {PREDICTIONS_LOCKED[0]} after the reconnect — on your honour, "
             "these should be the calls you noted before the reveal. Now run 1.3 and 1.4.")
    else:
        _say(f"🔒 predictions locked at {PREDICTIONS_LOCKED[0]} — on your honour; the notebook will notice a re-lock.")
    _say(" ".join(f"{k}:{v}" for k, v in calls.items()), indent="   ")


def run_1_3():
    """One batched, structured call: all 8 headlines to the model."""
    if not require(locked(), "run 1.2 first (lock your predictions) — the model is not called until you do"):
        return
    h = S["headlines"]
    prompt = prompt_lab1(h)
    _hr("Lab 1 · the model's calls (one request, JSON output)")
    _say(f"prompt includes: {flat_rule()}")
    res = ask("1.3", prompt, temperature=0, schema=CALL_SCHEMA)
    parsed = parse_calls(res["text"], list(h["id"]))
    S["results"]["1.3"] = {"parsed": parsed, "source": res["source"]}
    for i, (c, reason) in parsed.items():
        _trace(f"{i} → {c:<8} {reason}")
    if any(c == "UNPARSED" for c, _ in parsed.values()):
        _say("some rows are UNPARSED (scored as misses) — raw text shown; that is the finding, not a bug.")


def run_1_4():
    """Scoreboard: You k/8 · Model m/8 (flat band ±0.5 %)."""
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    if not require("1.3" in S["results"], "run 1.3 first (the model's calls)"):
        return
    if lock_is_placeholder():
        _hr("Lab 1 · scoreboard")
        _say("⛔ no scoreboard: your lock was lost with the runtime, so the notebook has no record of "
             "your calls — the calls you noted down are the record. To restore the scoreboard, run 1.2 "
             "(set the dropdowns to what you wrote), then 1.3 and 1.4 again.")
        return
    h = S["headlines"]
    you = PREDICTIONS_LOCKED[1]
    model = {i: c for i, (c, _) in S["results"]["1.3"]["parsed"].items()}
    ky, n = score(you, h)
    km, _ = score(model, h)
    S["results"]["1.4"] = {"you": ky, "model": km, "n": n}
    _hr("Lab 1 · scoreboard")
    _p(f"You: {ky}/{n} · Model: {km}/{n} (flat band ±{FLAT_BAND*100:.1f}%)")
    _p(f"model answer: {S['results']['1.3']['source']}")
    _p("")
    _p(f"{'id':<4} {'ticker':<6} {'truth':<6} {'ret':>8}  {'you':<6} {'model':<8} note")
    for r in h.itertuples():
        y, m = you.get(r.id, "—"), model.get(r.id, "—")
        note = "" if y == m else "← you disagreed with the model"
        if y != r.label and m == r.label:
            note = "← model right, you wrong"
        elif y == r.label and m != r.label:
            note = "← you right, model wrong"
        _p(f"{r.id:<4} {r.ticker:<6} {r.label:<6} {_fmt_pct(r.ret_event):>8}  {y:<6} {m:<8} {note}")
    _p("")
    _p("Q1 — Quote the headline where the model's reason surprised you.")


def _seed_row():
    h = S["headlines"]
    seed = h[h["group"] == "seed2020"]
    return (seed if len(seed) else h.tail(1)).iloc[0].to_dict()


def run_1_5a():
    """The reveal: the model 'predicts' the aftermath of the 2020 headline, beside the truth."""
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    row = _seed_row()
    _hr("Lab 1b · 1.5a — the 2020 headline")
    _trace(f"{row['id']} · {row['date']} · {COMPANY.get(row['ticker'], row['ticker'])}: {row['headline']}")
    res = ask("1.5a", prompt_recall(row), temperature=0)
    S["results"]["1.5a"] = res
    _p("")
    _p("MODEL SAYS:")
    for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
        _p("  " + line)
    _p("")
    _p("TRUTH FROM prices.csv (adjusted closes):")
    s = S["prices"][S["prices"]["ticker"] == row["ticker"]].set_index("date")["adj_close"]
    ev = row["event_date"]
    prior = s[s.index < ev]
    if len(prior) == 0:
        _say(f"  the price file does not go back far enough to score this headline (no close before "
             f"{ev} for {row['ticker']}) — pull prices from an earlier start date.")
    else:
        base = prior.iloc[-1]
        def ret_to(days):
            target = (pd.to_datetime(ev) + pd.Timedelta(days=days)).strftime("%Y-%m-%d")
            later = s[s.index <= target]
            return later.iloc[-1] / base - 1, later.index[-1]
        _p(f"  first reacting session {ev}: {_fmt_pct(row['ret_event'])}  (label {row['label']})")
        _p(f"  next session: {_fmt_pct(row['ret_next'])}")
        for d, name in ((35, "≈ 5 weeks later"), (182, "≈ 6 months later"), (365, "≈ 12 months later")):
            r, dd = ret_to(d)
            _p(f"  {name} ({dd}): {_fmt_pct(r)}")
    _p("")
    _p("Q2 — Quote the sentence in which the model tells you the future.")
    _say("A headline from 2020 is inside the model's reading; this is recall, not prediction.")


def _scored_arm(cell_id, rows, anonymised, title):
    prompt = prompt_lab1(rows, anonymised=anonymised)
    res = ask(cell_id, prompt, temperature=0, schema=CALL_SCHEMA)
    parsed = parse_calls(res["text"], list(rows["id"]))
    calls = {i: c for i, (c, _) in parsed.items()}
    k, n = score(calls, rows)
    S["results"][cell_id] = {"parsed": parsed, "hits": k, "n": n, "source": res["source"]}
    _p(f"{title}: Model {k}/{n} (flat band ±{FLAT_BAND*100:.1f}%) · {res['source']}")
    for r in rows.itertuples():
        c, reason = parsed[r.id]
        shown = ANON_HEADLINES.get(r.id, r.headline) if anonymised else r.headline
        _trace(f"{r.id} truth {r.label:<5} model {c:<8} {shown[:40]}…")
    return k, n


def run_1_5b():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5b — the same 8 headlines, names replaced by 'Company X'")
    _scored_arm("1.5b", S["headlines"], True, "Anonymised")


def run_1_5c():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5c — fresh headlines (last three weeks before build)")
    f = S["fresh"]
    if f is None or len(f) == 0:
        _say("fresh headlines unavailable — watch the projector (the instructor runs the canonical fresh arm).")
        S["results"]["1.5c"] = {"hits": None, "n": 0, "source": "unavailable"}
        return
    for r in f.itertuples():
        _trace(f"{r.id} · {r.date} · {r.ticker}: {r.headline}")
    _p("")
    _scored_arm("1.5c", f, False, "Fresh")
    if S.get("grounding_seen"):
        _say("note: the model reported search grounding — this is retrieval, not memory.")


def run_1_5d():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 1b · 1.5d — comparison")
    R = S["results"]
    k13, n13 = (None, None)
    if "1.3" in R:   # scored from 1.3's own parsed calls, so the row is filled whenever 1.3 has run
        k13, n13 = score({i: c for i, (c, _) in R["1.3"]["parsed"].items()}, S["headlines"])
    rows = [("named, old (1.3)", k13, n13, R.get("1.3", {}).get("source")),
            ("anonymised, old (1.5b)", R.get("1.5b", {}).get("hits"), R.get("1.5b", {}).get("n"), R.get("1.5b", {}).get("source")),
            ("named, fresh (1.5c)", R.get("1.5c", {}).get("hits"), R.get("1.5c", {}).get("n"), R.get("1.5c", {}).get("source"))]
    _p(f"{'arm':<26}{'model':>8}   tag")
    for name, k, n, src in rows:
        val = "not run" if k is None else f"{k}/{n}"
        _p(f"{name:<26}{val:>8}   {src or '—'}")
    if "1.4" in R:
        _p(f"You (named, old): {R['1.4']['you']}/{R['1.4']['n']}")
    else:
        _p("You (named, old): not scored yet — run 1.4")
    _p("")
    _say("n = 8 is a design demonstration, not a result. Named vs anonymised speaks to recognition / "
         "distraction; fresh speaks to cut-off contamination — and we cannot verify this model's cut-off.")


def run_2A_1():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2A.1 — 'best estimate, no disclaimers'")
    res = ask("2A.1", prompt_hallucination(), temperature=0)
    S["results"]["2A.1"] = res
    _p("MODEL SAYS:")
    for line in (res["text"] or "(no answer)").splitlines():
        _trace("  " + line)


def run_2A_2():
    if not require("2A.1" in S["results"], "run 2A.1 first"):
        return
    _hr("Lab 2 · 2A.2 — the truth beside the model's numbers")
    text = (S["results"]["2A.1"]["text"] or "").replace(",", "")   # '1,234.56' → '1234.56' (no comma decimals)
    _p(f"{'ticker':<7}{'model':>10}{'true close':>12}{'error':>9}")
    any_number = False
    for t in HALLUCINATION_TICKERS:
        true = get_price(t, HALLUCINATION_DATE)["close_usd"]
        m = re.search(rf"{t}\b[^0-9\n]{{0,25}}\$?\s*([0-9]{{1,6}}(?:\.[0-9]{{1,2}})?)", text)
        if m:
            any_number = True
            val = float(m.group(1))
            _p(f"{t:<7}{val:>10.2f}{true:>12.2f}{100*(val/true-1):>8.1f}%")
        else:
            _p(f"{t:<7}{'declined':>10}{true:>12.2f}{'—':>9}")
    _say(f"(true closes from prices.csv, as quoted on {HALLUCINATION_DATE}; no split adjustment needed "
         "for these three)")
    if not any_number:
        _say("The model declined to give numbers — that refusal is the finding. Q3: which control does it replace?")
    else:
        _say("Q3 — Circle every wrong closing price. What went wrong, and which finance control catches it?")


def run_2B():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2B — same question, two job titles")
    out = {}
    for key, label in (("risk", "cautious risk officer"), ("pm", "bullish hedge-fund PM")):
        res = ask(f"2B.{key}", prompt_persona(), temperature=0, persona=PERSONAS[key])
        out[key] = res["text"]
        _p(f"▶ {label}:")
        for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
            _p("  " + line)
        _p("")
    S["results"]["2B"] = out
    _say("Q4 — Write the two recommendations (cautious risk officer / bullish hedge-fund PM). "
         "What changed, and which finance control catches it?")


def run_2C():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 2 · 2C (stretch) — temperature 0 vs 1, three draws each")
    for temp in (0, 1):
        _p(f"temperature = {temp}")
        for d in (1, 2, 3):
            res = ask(f"2C.t{temp}.d{d}", prompt_temperature(), temperature=temp)
            _trace(f"  draw {d}: {res['text'] or '(no answer)'}")
    _say("At 0 the drafts are nearly (not exactly) identical; at 1 each draw is a new draft.")


def run_3_1():
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    _hr("Lab 3 · 3.1 — analyst mode (four read-only tools)")
    _say(f"TODAY = {TODAY} · tools: {', '.join(READ_ONLY_TOOLS)}")
    res = ask("3.1", prompt_analyst(), temperature=0, tools=READ_ONLY_TOOLS)
    S["results"]["3.1"] = res
    _p("")
    _p(f"tool calls this turn: {len(res['tool_calls'])}")
    _p("FINAL ANSWER:")
    for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
        _p("  " + line)
    _p("")
    _say("Q5 — Tool calls: ___ . Did every number in the final answer come from a tool? YES / NO — "
         "if NO, write the unsourced number.")
    if S["mode"] == MODE_LIVE and not FORCE_CACHED:
        _say("Wait for the 20-second countdown before the next cell — the model is rate-limited, not broken.")
        _countdown(20, "rate-limit breather")
    else:
        _p("(no countdown needed in STUDIO MODE)")


def _snapshot_receipt(cell_id):
    """Remember the student's own 3.2/3.3 run (limit, approval, order book) for the Receipt cell."""
    S["results"]["receipt"] = {"cell": cell_id, "time": _now(), "guard": dict(GUARD),
                               "book": [dict(o) for o in ORDER_BOOK]}


def run_3_2(max_order_gbp, require_human_approval, mandate=None, cell_id="3.2"):
    """Mandate mode. Widgets are read INSIDE the cell that calls this; ORDER_BOOK cleared each run.

    cell_id "3.2" (3.2 and 3.3) is the run the Receipt reports. Any other cell_id (the 🟡 3.4 sceptic
    card, the 🔴 instructor demo) is a SCRATCH run: it prints its own trace and book, then puts the
    student's 3.2/3.3 limit and ORDER BOOK back so the Receipt is unaffected.
    """
    if not require(locked(), "run 1.2 first (lock your predictions)"):
        return
    scratch = cell_id != "3.2"
    saved = (dict(GUARD), [dict(o) for o in ORDER_BOOK])
    GUARD["MAX_ORDER_GBP"] = int(max_order_gbp)
    GUARD["REQUIRE_HUMAN_APPROVAL"] = bool(require_human_approval)
    del ORDER_BOOK[:]
    _p(f"ACTIVE LIMIT: £{GUARD['MAX_ORDER_GBP']:,} · HUMAN APPROVAL: {'ON' if GUARD['REQUIRE_HUMAN_APPROVAL'] else 'OFF'}")
    _say(f"paper mandate £{MANDATE_GBP:,} (GBP notional only; 1 USD ≈ 0.78 GBP is NOT applied) · "
         f"long-only · TODAY {TODAY}")
    if scratch:
        _say(f"(scratch run for cell {cell_id} — your 3.2/3.3 limit and ORDER BOOK are kept for the Receipt)")
    text = mandate or mandate_text()
    # The 🔴 demo cell passes cell_id="instructor": a scratch run that replays the 3.2 cache entry
    # (the cache key is cell_id + prompt hash, and the demo prompt IS the 3.2 prompt).
    res = ask("3.2" if cell_id == "instructor" else cell_id, text, temperature=0, tools=MANDATE_TOOLS)
    S["results"][cell_id] = res
    try:
        if res["source"] == "miss":
            print_order_book()
            return
        orders = [c for c in res["tool_calls"] if c["name"] == "place_order"]
        _p("")
        # An order step with no "status" never reached guard_check — the arguments were malformed.
        # Say so, rather than printing "£0" as though the agent had asked for nothing.
        good = [c for c in orders if isinstance(c.get("result"), dict) and c["result"].get("status")]
        if good:
            o = good[0]["args"]
            _p(f"the agent wants to buy £{float(o.get('gbp_notional', 0)):,.0f} of {str(o.get('ticker', '?')).upper()}")
        elif orders:
            _p("the agent's order was rejected as malformed:")
            for line in textwrap.wrap(_fmt_args(orders[0].get("args") or {}), TRACE_WIDTH - 4):
                _p("  " + line)
        else:
            _p("the agent placed NO order this turn")
        _p("FINAL ANSWER:")
        for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
            _p("  " + line)
        print_order_book()
        refused = [o["result"].get("status", "") for o in good if o["result"].get("status", "").startswith("REFUSED")]
        if refused:
            _say(f"guard that fired: {', '.join(_guards_fired(refused))}"
                 + (" (cached wording may name another control — the ORDER BOOK is the record)"
                    if res["source"] == "cached" else ""))
            _say("Q6 — When refused, the agent (a) tried a smaller order (b) argued (c) complied silently "
                 "(d) other — quote one sentence if there is one.")
        elif good and not scratch:
            _say("3.3 — set MAX_ORDER_GBP BELOW the amount above, tick REQUIRE_HUMAN_APPROVAL, and run this "
                 "cell again.")
    finally:
        if scratch:
            GUARD.clear()
            GUARD.update(saved[0])
            del ORDER_BOOK[:]
            ORDER_BOOK.extend(saved[1])
        else:
            _snapshot_receipt(cell_id)


def _guards_fired(statuses):
    """Which control(s) produced these REFUSED statuses (guard_check wording → control name)."""
    names = []
    for st in statuses:
        name = ("MAX_ORDER_GBP" if "MAX_ORDER_GBP" in st else
                "REQUIRE_HUMAN_APPROVAL" if "REQUIRE_HUMAN_APPROVAL" in st else
                "long-only mandate" if "long-only" in st else "ticker universe")
        if name not in names:
            names.append(name)
    return names


def print_order_book(book=None):
    book = ORDER_BOOK if book is None else book
    _p("ORDER BOOK:")
    if not book:
        _p("  (empty)")
    for o in book:
        px = f" @ ${o['price_usd']:.2f}" if "price_usd" in o else ""
        _trace(f"  {o['time']} {o['side']} {o['ticker']} £{o['gbp_notional']:,.0f}{px} — {o['status']}")


def run_receipt():
    """Receipt: scoreboard + the student's OWN 3.2/3.3 run (not the 3.4 scratch run)."""
    _hr("Receipt — keep this for your reflection")
    R = S["results"]
    if "1.4" in R:
        _p(f"Scoreboard: You {R['1.4']['you']}/{R['1.4']['n']} · Model {R['1.4']['model']}/{R['1.4']['n']} "
           f"(flat band ±{FLAT_BAND*100:.1f}%)")
    elif lock_is_placeholder():
        _say("Scoreboard: not in this runtime (lock lost over the break) — the score you noted down is the record.")
    else:
        _say("Scoreboard: run 1.4 first (1.2 → 1.3 → 1.4) — then re-run this cell.")
    for cid, name in (("1.5b", "anonymised"), ("1.5c", "fresh")):
        if cid in R and R[cid].get("hits") is not None:
            _p(f"  {name}: Model {R[cid]['hits']}/{R[cid]['n']}")
    _p(f"Mode: {S['mode']} · live requests this session: {S['n_requests']}")
    snap = R.get("receipt")
    if snap:
        g = snap["guard"]
        _p(f"Last 3.2/3.3 run at {snap['time']}: limit £{g['MAX_ORDER_GBP']:,} · "
           f"approval {'ON' if g['REQUIRE_HUMAN_APPROVAL'] else 'OFF'}")
        print_order_book(snap["book"])
    else:
        _p("No 3.2/3.3 run in this runtime yet — run 3.2 (and 3.3) first.")


def run_4_1():
    _hr(f"Lab 4 · the committee is convened on {COMMITTEE_TICKER}")
    md = S["committee_md"]
    if not md:
        md = ("# PLACEHOLDER — committee transcript not generated yet\n\n"
              "Run build/build_cache.py with a key (or build/make_sample_cache.py) to create cache/committee.md.")
    ip = None
    try:
        import IPython
        ip = IPython.get_ipython()
    except Exception:
        pass
    if ip is not None:
        from IPython.display import Markdown, display
        display(Markdown(md))
    else:
        _p(md)
    return md


def committee_context():
    """The Lab-3 tool outputs every committee role shares (real tool functions)."""
    t = COMMITTEE_TICKER
    d = pd.to_datetime(TODAY)
    ctx = {
        "price": get_price(t, TODAY),
        "ret_1m": get_return(t, (d - pd.Timedelta(days=30)).strftime("%Y-%m-%d"), TODAY),
        "ret_3m": get_return(t, (d - pd.Timedelta(days=91)).strftime("%Y-%m-%d"), TODAY),
        "ret_12m": get_return(t, (d - pd.Timedelta(days=365)).strftime("%Y-%m-%d"), TODAY),
        "vol_30d": get_volatility(t, 30),
        "headlines": get_headlines(t),
    }
    return ctx


COMMITTEE_ROLES = [
    ("bull", "Bull analyst", "Argue the case FOR buying {company} ({ticker}) now, in at most 120 words, using only the data below."),
    ("bear", "Bear analyst", "Argue the case AGAINST buying {company} ({ticker}) now, in at most 120 words, using only the data below."),
    ("risk", "Risk manager", "Given the bull and bear cases and the data below, state the two biggest risks and a position limit you would impose, in at most 120 words."),
    ("trader", "Trader", "Given the bull, bear and risk-manager notes and the data below, propose ONE paper order for {company} ({ticker}) within a £1,000,000 long-only mandate: ticker, GBP size, and a two-sentence rationale. State the order as 'ORDER: BUY £<size> of {ticker}'."),
]


def committee_prompts(ctx):
    """Four sequential prompts sharing the tool outputs; later roles see earlier notes."""
    data = json.dumps(ctx, default=str, indent=0)
    prompts = {}
    for key, role, instr in COMMITTEE_ROLES:
        p = f"Today is {TODAY}. You are the {role} on a paper-trading investment committee. "
        p += instr.format(company=COMPANY[COMMITTEE_TICKER], ticker=COMMITTEE_TICKER)
        p += "\n\nDATA (from the Lab-3 tools):\n" + data
        prompts[key] = p
    return prompts


def committee_prompt(key, prompts, notes):
    """Later roles see the earlier notes (the chain is the same in the notebook and the builders)."""
    p = prompts[key]
    if key == "risk":
        p += f"\n\nBULL:\n{notes.get('bull', '')}\n\nBEAR:\n{notes.get('bear', '')}"
    if key == "trader":
        p += (f"\n\nBULL:\n{notes.get('bull', '')}\n\nBEAR:\n{notes.get('bear', '')}"
              f"\n\nRISK MANAGER:\n{notes.get('risk', '')}")
    return p


def run_committee_live(cell_prefix="4.2"):
    """🟡 Regenerate the transcript live (4 calls). Also used by build_cache.py."""
    ctx = committee_context()
    prompts = committee_prompts(ctx)
    notes = {}
    for key, role, _ in COMMITTEE_ROLES:
        p = committee_prompt(key, prompts, notes)
        res = ask(f"{cell_prefix}.{key}", p, temperature=0)
        notes[key] = res["text"] or "(no answer)"
        _p(f"▶ {role} ({res['source']}):")
        for line in textwrap.wrap(notes[key], TRACE_WIDTH - 2):
            _p("  " + line)
        _p("")
    return ctx, notes


def committee_evidence_rows(ctx):
    """The shared evidence as (row-number, tool call, returned value) triples.

    One row per Lab-3 tool output, each tagged with the tool call (brief §5) that
    produced it, so a reader can check any number in the transcript against the call
    that returned it. This is what the Analyst role is asked to do in committee.
    """
    t = COMMITTEE_TICKER
    rows = []
    p, h = ctx["price"], ctx["headlines"]
    rows.append((f'get_price("{t}", "{p["date"]}")', f'close_usd = {p["close_usd"]:.2f} USD (unadjusted close)'))
    for key, label in (("ret_1m", "one month"), ("ret_3m", "three months"), ("ret_12m", "twelve months")):
        r = ctx[key]
        rows.append((f'get_return("{t}", "{r["start"]}", "{r["end"]}")',
                     f'return_pct = {r["return_pct"]:+.2f} % ({label}, adjusted closes)'))
    v = ctx["vol_30d"]
    rows.append((f'get_volatility("{t}", {v["window_days"]})',
                 f'annualised_vol_pct = {v["annualised_vol_pct"]:.2f} % (as of {v["as_of"]})'))
    rows.append((f'get_headlines("{t}")',
                 f'{len(h["headlines"])} headline(s), date + text only, dated <= TODAY'))
    for i, item in enumerate(h["headlines"], 1):
        rows.append((f'  \u2514 headline {i} (from get_headlines)', f'{item["date"]} \u2014 {item["text"]}'))
    rows.append(("(constant) TODAY", f"{TODAY} \u2014 no tool may return anything dated after this"))
    rows.append(("(constant) mandate", f"\u00a3{MANDATE_GBP:,.0f} paper notional, long-only, exactly one place_order call"))
    return [(i + 1, a, b) for i, (a, b) in enumerate(rows)]


def committee_evidence_table(ctx):
    rows = committee_evidence_rows(ctx)
    w = max(len(a) for _, a, _ in rows)
    out = [f"{'#':>2}  {'TOOL CALL (brief §5)'.ljust(w)}  RETURNED",
           f"{'--':>2}  {'-' * w}  {'-' * 34}"]
    for n, a, b in rows:
        out.append(f"{n:>2}  {a.ljust(w)}  {b}")
    return "\n".join(out)


def committee_markdown(ctx, notes, generated_by, sample, appendix=None):
    """Transcript: evidence table (each row tagged to its tool call), four notes, appendix.

    `appendix` is an optional dict with keys "numbers" and "unsourced": lists of strings
    separating the numbers that came from a tool row above from the claims that did not.
    Both builders (build_cache.py, make_sample_cache.py) call this one function, so the
    long and short forms of the transcript cannot diverge.
    """
    md = f"# Investment committee transcript — {COMPANY[COMMITTEE_TICKER]} ({COMMITTEE_TICKER})\n\n"
    md += f"*Generated {datetime.now().strftime('%Y-%m-%d')} by {generated_by}; data as of {TODAY}.*  \n"
    if sample:
        md += "**SAMPLE — stand-in text written by Claude (Anthropic), not model output; regenerate with build/build_cache.py.**\n\n"
    md += ("## Shared evidence (Lab-3 tool outputs)\n\n"
           "Every agent below saw exactly this, and nothing else. Each row is tagged with "
           "the tool call that produced it.\n\n```\n" + committee_evidence_table(ctx) + "\n```\n\n")
    for key, role, _ in COMMITTEE_ROLES:
        md += f"## {role}\n\n{notes.get(key, '')}\n\n"
    if appendix:
        md += "## Appendix — check the transcript against the table\n\n"
        md += "### Numbers used (each one traces to a row above)\n\n"
        for line in appendix.get("numbers", []):
            md += f"- {line}\n"
        md += "\n### Claims made without tool support (nothing in the table returns these)\n\n"
        for line in appendix.get("unsourced", []):
            md += f"- {line}\n"
        md += "\n*For the Analyst role: the second list is the answer, and the reason the trace is printed.*\n"
    return md


def reset_all():
    """Reset cell: clears the lock, the order book, results and counters (instructor / demo use)."""
    global PREDICTIONS_LOCKED
    PREDICTIONS_LOCKED = None
    globals()["PREDICTIONS_LOCKED"] = None
    del ORDER_BOOK[:]
    S["results"] = {}
    S["n_requests"] = 0
    GUARD.update(MAX_ORDER_GBP=MANDATE_GBP, REQUIRE_HUMAN_APPROVAL=False)
    _say("reset: lock cleared, ORDER BOOK emptied, results and request counter zeroed.")


def run_instructor_unprompted(n=5):
    """🔴 Would it trade unprompted? n re-runs of 3.2 without the mandatory-order line."""
    if not require(locked(), "run 1.2 first (any dropdown values) — the gate applies to instructors too"):
        return
    saved = (dict(GUARD), [dict(o) for o in ORDER_BOOK])   # scratch runs: the Receipt is untouched
    try:
        for i in range(1, n + 1):
            _hr(f"🔴 unprompted run {i}/{n}")
            GUARD.update(MAX_ORDER_GBP=MANDATE_GBP, REQUIRE_HUMAN_APPROVAL=False)
            del ORDER_BOOK[:]
            res = ask(f"X.unprompted.d{i}", mandate_text(mandatory=False), temperature=1, tools=MANDATE_TOOLS)
            orders = [c for c in res["tool_calls"] if c["name"] == "place_order"]
            _p(f"orders placed: {len(orders)} · source: {res['source']}")
            for line in textwrap.wrap(res["text"] or "(no answer)", TRACE_WIDTH - 2):
                _p("  " + line)
    finally:
        GUARD.clear()
        GUARD.update(saved[0])
        del ORDER_BOOK[:]
        ORDER_BOOK.extend(saved[1])

# ---- go -------------------------------------------------------------------
reconnect()
````

### VRGA-S — five questions, filled four times
| Dimension | Question | Row 1 — headline signal | Row 2 — trading agent | Row 3 — committee decision | Row 4 — your sector |
|---|---|---|---|---|---|
| **Value** | What decision does this improve, for whom? | faster news triage; a research labeller | drafting with evidence attached | a logged, repeatable process | *you* |
| **Risk** | Worst plausible failure; who bears the cost? | look-ahead / distraction; crowded signal; small-cap concentration | hallucinated number; over-sized order; no audit trail | deployment failure (Knight); regime change (Zillow) | *you* |
| **Governance** | Who signs off, audits, overrides, on what basis? | validation sample; point-in-time data | trace as audit log; order limit; human approval | five constraints; named owner; kill-switch | *you* |
| **Appropriateness** | Structurally suited (pattern + language) or not (causal inference, fiduciary judgement)? | language → signal plausible; new information unproven | drafting yes; executing contested | fiduciary judgement stays human | *you* |
| **Sector constraints** | Which finance-specific rule does this bump into? | market abuse; model risk | suitability; record-keeping; AI-washing | FCA outcomes rules; EU AI Act high-risk; FSB homogenisation | *you* |
